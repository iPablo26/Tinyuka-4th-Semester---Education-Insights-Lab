## 1. Problem Definition

### 1.1 Background and Business Context

Educational institutions across the United States differ in their financial resources, staffing capacity, socioeconomic conditions, access to technology, and academic outcomes.

These differences raise important questions about the factors associated with students' academic performance and the distribution of educational opportunities.

For education policymakers and institutional decision-makers, understanding these relationships is essential for identifying potential educational disparities and determining where further investigation or support may be needed.

This project focuses on mathematics proficiency as the primary academic outcome and investigates its relationship with institutional funding, student-teacher ratios, socioeconomic disadvantage, and broadband access.

The investigation aims to provide evidence that can guide educational planning while recognizing that statistical associations do not necessarily establish cause-and-effect relationships.

The primary stakeholder for this project is an education-focused organization seeking to understand differences in institutional resources, educational opportunities, and academic performance across the United States.

Without a structured analytical investigation, education stakeholders may struggle to identify meaningful patterns or determine which institutional characteristics deserve closer attention.

This project addresses this challenge by examining relationships between mathematics proficiency and selected institutional characteristics, applying statistical methods to investigate these relationships, and developing machine learning models to assess the predictive value of available institutional information.

These limitations will be revisited as the analysis progresses, and the findings will be interpreted within the scope of the available evidence.

#### Observation

The dataset contains **5,000 rows and 35 columns**, representing institutional records and their associated characteristics.

The available variables cover institutional identification, geographical location, educational level, ownership, student enrollment, staffing, socioeconomic conditions, funding, academic performance, and technology access.

Importantly, the dataset contains `math_proficiency_pct`, which is the proposed target variable for this investigation, alongside the institutional characteristics needed to examine the research questions.

The dataset therefore provides a suitable starting point for investigating relationships between institutional resources, socioeconomic conditions, and mathematics proficiency. Further inspection is required to assess data quality and determine the appropriate population for analysis.

### 2.3 Dataset Information and Variable Types


```python

# Display a summary of the dataset structure,
# including data types and non-null counts
edu_data.info()
```

#### Observation

The dataset contains a combination of numerical and categorical variables, making it suitable for different forms of exploratory and statistical analysis.

### 2.4 Identifying Numerical and Categorical Variables


```python

# Identify numerical columns in the dataset
numerical_columns = edu_data.select_dtypes(include=["int64", "float64"]).columns

# Identify categorical columns in the dataset
categorical_columns = edu_data.select_dtypes(include=["object"]).columns

# Display the number of numerical and categorical variables
print("Total Numerical Variables:", len(numerical_columns))
print("Total Categorical Variables:", len(categorical_columns))

# Display the numerical column names
print("\nNumerical Variables:")
for column in numerical_columns:
    print(column)

# Display the categorical column names
print("\nCategorical Variables:")
for column in categorical_columns:
    print(column)
```

#### Observation

The dataset contains **25 numerical variables and 10 categorical variables**.

### 2.5 Missing Values Assessment

To assess the completeness of the dataset, I examined the number and percentage of missing observations across all variables.

This assessment aims to identify affected columns, determine the extent of missingness, and investigate whether missing values may be associated with specific institution levels.

The findings will inform appropriate missing-data treatment decisions during the data cleaning stage.

```python
# Calculate the number of missing values in each column
missing_values = edu_data.isnull().sum()

# Calculate the percentage of missing values in each column
missing_percentage = (edu_data.isnull().mean() * 100).round(2)

# Combine the missing value counts and percentages
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage (%)": missing_percentage
})

# Display only columns containing missing values
missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
]

# Sort columns by the number of missing values
missing_summary = missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)

# Display the missing value summary
missing_summary
```

#### 2.5.1 Missing Values by Institution Level

Since the dataset contains different educational institution levels, I further investigated whether the missing values were associated with institution type.

This assessment is particularly important for standardized test scores, AP/IB participation, and postsecondary enrollment indicators, which may not apply equally across all institution categories.

Understanding these patterns will help distinguish potentially inapplicable measurements from missing information that may require further treatment.

```python
# Identify columns containing missing values
missing_columns = edu_data.columns[
    edu_data.isnull().any()
]

# Count the available observations for each institution level
missing_by_level = edu_data.groupby("level")[
    missing_columns
].count()

# Display the non-missing values by institution level
missing_by_level
```

#### Observation

The missing value assessment identified **four variables with incomplete observations**:

- `avg_sat_score` — 67.50% missing.
- `avg_act_score` — 67.50% missing.
- `ap_ib_participation_pct` — 71.58% missing.
- `postsecondary_enrollment_pct` — 71.58% missing.

Further investigation revealed that the availability of these variables differs systematically across institution levels.

SAT and ACT scores are available for high schools, K–12 combined institutions, and four-year universities, while AP/IB participation and postsecondary enrollment indicators are recorded for high schools and K–12 combined institutions.

These patterns suggest that missingness may be related to the applicability or reporting of specific educational indicators rather than random data loss.

Importantly, the primary target variable, `math_proficiency_pct`, and the initial explanatory variables selected for this investigation contain no missing values.

Therefore, the missing observations in the four identified columns do not directly prevent the proposed mathematics proficiency analysis.

**Data Preparation Consideration:** The missing values will be retained during data understanding. Appropriate treatment will be determined during data cleaning, considering indicator applicability, institution level, and relevance to the research question.

### 2.6 Duplicate Records and Institution Identifier Assessment

Duplicate records can affect data quality by introducing repeated observations that may distort descriptive statistics, analytical findings, and machine learning results.

To assess the uniqueness of institutional records, I examined exact duplicate rows, repeated institution identifiers, and repeated institution names.

This assessment also considers the possibility that institutions may share similar names or characteristics without necessarily representing duplicate records.

The objective is to identify potential duplication issues and determine whether further investigation is required during data preparation.

```python

# Check for completely duplicated rows
duplicate_rows = edu_data.duplicated().sum()

# Check for repeated institution IDs
duplicate_ids = edu_data["institution_id"].duplicated().sum()

# Check for repeated institution names
duplicate_names = edu_data["institution_name"].duplicated().sum()

# Display the duplicate assessment
print("Duplicate Records Assessment")
print("----------------------------")
print("Exact Duplicate Rows:", duplicate_rows)
print("Duplicate Institution IDs:", duplicate_ids)
print("Repeated Institution Names:", duplicate_names)
```

#### 2.6.1 Investigating Potential Duplicate Institutions

Although no duplicate institution identifiers or exact duplicate records were identified, several institution names appear more than once.

To investigate these repetitions further, I examined records sharing the same institution name, state, county, educational level, and ownership category.

This combined assessment helps identify records that may represent the same institution or institutions with similar characteristics.

However, matching these attributes does not independently confirm duplication. The institution identifier and additional contextual information must be considered before removing any observations.

```python
# Define the institutional characteristics used for comparison
institution_columns = [
    "institution_name",
    "state_code",
    "county",
    "level",
    "control"
]

# Identify all records sharing the same institutional characteristics
potential_duplicates = edu_data[
    edu_data.duplicated(
        subset=institution_columns,
        keep=False
    )
]

# Display the number of potentially matching records
print("Potentially Matching Institution Records:",
      len(potential_duplicates))

# Display a sample for further investigation
potential_duplicates[
    ["institution_id"] + institution_columns
].head(10)
```

#### Observation

The duplicate assessment revealed the following:

- No exact duplicate rows were identified.
- All 5,000 institution identifiers are unique.
- A total of 4,220 records have institution names that occur earlier in the dataset.
- Further investigation identified 134 records sharing the same institution name, state, county, educational level, and ownership category.

The repeated institution names and matching institutional characteristics do not necessarily indicate duplicate observations, particularly because institution identifiers remain unique.

Removing these records without additional verification could result in the loss of valid institutional information.

**Data Preparation Decision:** No records will be removed solely on the basis of repeated institution names or matching contextual characteristics. The original records will be retained unless sufficient evidence of duplication is established during data cleaning.

This approach helps preserve data integrity and prevents unnecessary information loss.

The variables considered include per-pupil expenditure, student-teacher ratio, socioeconomic disadvantage, and broadband access.

This focused assessment provides an initial understanding of the observed values and their variability before investigating relationships among the variables.

#### 2.7.2 Preliminary Anomaly Assessment

Beyond examining descriptive statistics, I performed preliminary validation checks to identify potentially invalid numerical observations.

The assessment focuses on percentage values outside the expected 0–100 range and institutional measurements that should ordinarily be positive, such as enrollment, teacher counts, and expenditure.

These checks are intended to identify potential data-quality concerns rather than automatically classify unusual observations as errors or outliers.

```python
# Identify all percentage-based columns
percentage_columns = [
    column for column in edu_data.columns
    if column.endswith("_pct") or column == "percent_free_reduced_lunch"
]

# Check for percentages below 0 or above 100
invalid_percentages = (
    (edu_data[percentage_columns] < 0) |
    (edu_data[percentage_columns] > 100)
)

print("Total Invalid Percentage Values:",
      invalid_percentages.sum().sum())

# Check for zero or negative values in selected institutional variables
print("Nonpositive Enrollment:",
      (edu_data["total_enrollment"] <= 0).sum())

print("Nonpositive Teacher Counts:",
      (edu_data["number_of_teachers"] <= 0).sum())

print("Nonpositive Per-Pupil Expenditure:",
      (edu_data["per_pupil_expenditure_usd"] <= 0).sum())
```

#### 2.7.3 Observation and Preliminary Findings

The descriptive statistical assessment revealed considerable variation in institutional characteristics and academic performance across the dataset.

The primary findings include:

1. **Mathematics proficiency:** The average mathematics proficiency is approximately 60.01%, with values ranging from 12% to 96%, indicating differences in recorded academic outcomes.

2. **Educational expenditure:** Average per-pupil expenditure is approximately $17,113.41. The maximum recorded expenditure of $50,601 suggests the presence of high-value observations that require further contextual investigation.

3. **Student-teacher ratio:** The average ratio is approximately 14.93, with values ranging from 6 to 30 students per teacher.

4. **Socioeconomic conditions:** The percentage of students receiving free or reduced-price lunch ranges from 0% to 98%, indicating substantial variation in this socioeconomic indicator.

5. **Broadband access:** Average broadband access is approximately 90.79%, with observed values ranging from 74.5% to 99%.

The preliminary validation checks identified no negative or above-100 percentage values and no nonpositive enrollment, teacher count, or expenditure values.

However, the differences between minimum, maximum, and typical values suggest that further investigation of unusual observations is necessary.

**Data Preparation Consideration:** Potential outliers will be assessed in the context of institution level and variable distributions before deciding whether treatment is appropriate. No observations have been modified or removed at this stage.

### 3.2 Investigating and Handling Missing Values

The data understanding stage identified missing observations in four variables: average SAT scores, average ACT scores, AP/IB participation, and postsecondary enrollment.

Before applying any treatment, I reassessed these missing values in the working dataset to determine their extent, possible causes, and relevance to the research question.

Since missing values may result from differences in indicator applicability across educational levels, the treatment decision will prioritize preserving meaningful information and avoiding unnecessary data modifications.

```python
# Identify columns containing missing values
missing_values = edu_clean.isnull().sum()

# Calculate the percentage of missing values
missing_percentage = (
    edu_clean.isnull().mean() * 100
).round(2)

# Create a missing values summary
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage (%)": missing_percentage
})

# Display only columns with missing observations
missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
]

# Sort by missing percentage in descending order
missing_summary = missing_summary.sort_values(
    by="Missing Percentage (%)",
    ascending=False
)

missing_summary
```

#### 3.2.1 Missing Values Across Institution Levels

To investigate the potential causes of missingness, I examined the distribution of missing values across educational institution levels.

This assessment is important because standardized test scores, advanced academic programmes, and postsecondary indicators may not be equally applicable to elementary schools, secondary schools, colleges, and universities.

Understanding these patterns helps distinguish potentially structural missingness from missing information that may require imputation or further investigation.

```python
# Select the four variables containing missing values
missing_columns = [
    "avg_sat_score",
    "avg_act_score",
    "ap_ib_participation_pct",
    "postsecondary_enrollment_pct"
]

# Calculate the percentage of missing values within each institution level
missing_by_level = (
    edu_clean.groupby("level")[missing_columns]
    .apply(lambda group: group.isnull().mean() * 100)
    .round(2)
)

# Display missing percentages by institution level
missing_by_level
```

#### 3.2.2 Assessment of Missing Data Mechanisms

To determine an appropriate missing-data treatment strategy, I examined the observed missingness patterns in relation to the three conventional missing-data mechanisms:

- **Missing Completely At Random (MCAR):** Missingness is independent of observed and unobserved values.
- **Missing At Random (MAR):** Missingness depends on observed variables but not on the missing values themselves after conditioning on those variables.
- **Missing Not At Random (MNAR):** Missingness continues to depend on the unobserved values themselves, even after accounting for observed information.

The investigation revealed that missing observations in SAT scores, ACT scores, AP/IB participation, and postsecondary enrollment are systematically associated with educational institution level.

For instance, SAT and ACT scores are unavailable for all elementary and middle school records, while AP/IB participation and postsecondary enrollment indicators are recorded only for high schools and K–12 combined institutions.

These patterns are inconsistent with the expectation of completely random missingness and indicate that missingness is strongly associated with the observed `level` variable.

The findings are therefore **consistent with an MAR mechanism when interpreted statistically**, although structural missingness arising from indicator inapplicability is a more plausible practical explanation for several institution categories.

The available evidence does not establish whether any additional MNAR mechanism exists, and the actual reporting rules would be required to confirm the underlying cause.

**Missingness Assessment:** The four affected variables exhibit level-dependent, likely structural missingness. Their missingness is consistent with MAR conditional on institution level, but the statistical mechanism cannot be established conclusively from the observed data alone.

**Treatment Decision:** Retain the missing observations without imputation, since the affected indicators are not required for the primary research question and replacing potentially inapplicable values could introduce misleading information.

#### 3.2.3 Missing Values in the Selected Research Variables

Since the investigation focuses on mathematics proficiency and selected institutional characteristics, I separately assessed missing values in the target and initial explanatory variables.

This check helps determine whether missing-data treatment is necessary for the variables directly relevant to the planned statistical analysis and regression modelling.

```python
# Define the target and initial explanatory variables
research_variables = [
    "math_proficiency_pct",
    "per_pupil_expenditure_usd",
    "student_teacher_ratio",
    "percent_free_reduced_lunch",
    "broadband_access_pct"
]

# Check missing observations in the selected research variables
research_missing = pd.DataFrame({
    "Missing Values": edu_clean[research_variables].isnull().sum(),
    "Missing Percentage (%)": (
        edu_clean[research_variables].isnull().mean() * 100
    ).round(2)
})

# Display the results
research_missing
```

#### 3.2.4 Missing Data Treatment Decision

The missing-value investigation revealed that the four affected variables exhibit systematic missingness across educational institution levels.

The observed patterns suggest that missing observations may reflect differences in indicator applicability or reporting rather than random data loss.

Furthermore, the primary target variable and all four initial explanatory variables contain complete observations.

Based on these findings, I adopted the following treatment strategy:

1. **Retain existing missing values** in SAT scores, ACT scores, AP/IB participation, and postsecondary enrollment.
2. **Avoid numerical imputation**, since replacing potentially inapplicable measurements with mean, median, or zero values could introduce misleading information.
3. **Preserve the affected columns** in the working dataset for documentation and possible institution-specific investigations.
4. **Exclude these variables from the initial predictive feature set**, as they are not necessary for the primary research question.
5. **Retain all institutional records at this stage**, pending the separate assessment of institution-level comparability.

This decision preserves the available information without introducing unsupported estimates or unnecessary record deletion.

```python
# Verify that the working dataset retains its original dimensions
print("Original Dataset Shape:", edu_data.shape)
print("Working Dataset Shape:", edu_clean.shape)

# Check the total missing values remaining
print("\nTotal Missing Values:", edu_clean.isnull().sum().sum())

# Confirm that the original and working datasets currently match
print(
    "Original and Working Datasets Match:",
    edu_data.equals(edu_clean)
)
```

#### 3.2.5 Observation and Conclusion

The missing-data assessment confirmed that four variables contain incomplete observations, with missingness closely associated with educational institution levels.

The selected mathematics proficiency target and initial explanatory variables are complete, eliminating the need for missing-value imputation within the core research variables.

Given the possibility that the affected indicators are not applicable across all institution categories, the missing values were retained rather than replaced with unsupported estimates.

The working dataset therefore remains unchanged at **5,000 records and 35 variables**, with the original 13,908 missing cells preserved.

**Conclusion:** No missing-value imputation or record deletion is necessary at this stage. The retained missing values will be documented, and the affected columns will not be included in the initial mathematics proficiency model.

This approach preserves data integrity and supports a transparent, research-driven preparation process.

#### 3.5.2 Whitespace and Capitalization Consistency Checks

To identify potential formatting inconsistencies, I examined the categorical variables for leading or trailing whitespace, blank text values, and labels that differ only in capitalization.

These checks help detect categories that may represent the same information but are stored using different text formats.

The results will determine whether any corrective standardization is necessary.

```python
# Create an empty list to store the validation results
category_checks = []

# Examine each object-type column
for column in categorical_columns:

    # Convert the column to pandas' nullable string type
    values = edu_clean[column].astype("string")

    # Remove leading and trailing spaces temporarily
    stripped_values = values.str.strip()

    # Count values with unnecessary surrounding spaces
    whitespace_issues = (
        values.ne(stripped_values).fillna(False).sum()
    )

    # Count empty strings after removing surrounding spaces
    blank_values = stripped_values.eq("").fillna(False).sum()

    # Compare unique labels without capitalization differences
    original_unique = stripped_values.nunique()

    normalized_unique = (
        stripped_values.str.casefold().nunique()
    )

    # Store the validation findings
    category_checks.append({
        "Variable": column,
        "Whitespace Issues": int(whitespace_issues),
        "Blank Values": int(blank_values),
        "Case-Only Differences": int(
            original_unique - normalized_unique
        )
    })

# Convert the results into a DataFrame
category_validation = pd.DataFrame(category_checks)

# Display the validation results
category_validation
```

#### 3.5.3 Geographical Category Consistency

Because the dataset contains both state codes and state names, I examined whether these geographical identifiers were consistently associated with one another.

This validation is intended to identify cases where a state abbreviation may correspond to multiple state names, or where a state name may be associated with multiple abbreviations.

Consistent geographical labels are important for accurate institutional grouping and location-based analysis.

```python
# Count the unique state names associated with each state code
names_per_code = edu_clean.groupby(
    "state_code"
)["state_name"].nunique()

# Count the unique state codes associated with each state name
codes_per_name = edu_clean.groupby(
    "state_name"
)["state_code"].nunique()

# Identify inconsistent state code and name relationships
inconsistent_codes = (names_per_code > 1).sum()
inconsistent_names = (codes_per_name > 1).sum()

# Display the findings
print("State Codes with Multiple Names:", inconsistent_codes)
print("State Names with Multiple Codes:", inconsistent_names)

print("\nUnique State Codes:", edu_clean["state_code"].nunique())
print("Unique State Names:", edu_clean["state_name"].nunique())
```

#### 3.5.4 Categorical Data Cleaning Decision

The categorical validation assessment revealed that:

- All ten object-type columns contain complete observations.
- The institutional classification variables have identifiable and consistently formatted categories.
- No leading or trailing whitespace issues were detected.
- No blank text values were identified.
- No categories differing only by capitalization were detected.
- State codes and state names have consistent one-to-one relationships within the dataset.

Based on these findings, no categorical standardization or replacement is required at this stage.

**Cleaning Decision:** The existing categorical values will be retained without modification.

Institution identifiers and names will remain available for record identification, while relevant institutional categories will be preserved for subsequent analysis.

Any categorical encoding required for machine learning will be performed during model preparation rather than during this cleaning stage.

```python
# Verify that no formatting inconsistencies were detected
assert (
    category_validation[
        [
            "Whitespace Issues",
            "Blank Values",
            "Case-Only Differences"
        ]
    ].sum().sum() == 0
), "Categorical formatting inconsistencies detected."

# Verify the state code and name relationships
assert inconsistent_codes == 0
assert inconsistent_names == 0

# Confirm the working dataset dimensions
print("Working Dataset Shape:", edu_clean.shape)

print("Categorical validation completed successfully.")
print("No categorical values modified.")
```

### 3.7 Outlier Detection and Treatment

Outliers are observations that differ considerably from the majority of values in a dataset. They may result from data entry errors, measurement inconsistencies, or legitimate differences between institutions.

Because extreme observations can influence statistical summaries and machine learning models, I investigated potential outliers in the numerical variables relevant to the research question.

The Interquartile Range (IQR) method was selected to identify unusually low or high values.

Rather than automatically removing flagged observations, the investigation will consider their validity, institutional context, and potential impact on the analysis.

### 3.7.1 Detecting Outliers Using the IQR Method

The IQR method identifies observations outside the range defined by 1.5 times the interquartile range below the first quartile or above the third quartile.

The assessment focuses on enrollment, staffing, educational expenditure, mathematics proficiency, socioeconomic disadvantage, and broadband access.

#### 3.7.2 Calculate Outliers in the Selected Numerical Variables

```python
# Select numerical variables for outlier assessment
outlier_columns = [
    "total_enrollment",
    "number_of_teachers",
    "student_teacher_ratio",
    "per_pupil_expenditure_usd",
    "math_proficiency_pct",
    "percent_free_reduced_lunch",
    "broadband_access_pct"
]

# Store the outlier assessment results
outlier_results = []

# Apply the IQR method to each selected variable
for column in outlier_columns:

    # Calculate the first and third quartiles
    Q1 = edu_clean[column].quantile(0.25)
    Q3 = edu_clean[column].quantile(0.75)

    # Calculate the interquartile range
    IQR = Q3 - Q1

    # Calculate the lower and upper outlier boundaries
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify values outside the calculated boundaries
    outliers = (
        (edu_clean[column] < lower_bound) |
        (edu_clean[column] > upper_bound)
    )

    # Store the results for the current variable
    outlier_results.append({
        "Variable": column,
        "Lower Bound": round(lower_bound, 2),
        "Upper Bound": round(upper_bound, 2),
        "Outlier Count": int(outliers.sum()),
        "Outlier Percentage": round(
            outliers.mean() * 100, 2
        )
    })

# Convert the results into a summary DataFrame
outlier_summary = pd.DataFrame(outlier_results)

# Display the outlier assessment
outlier_summary
```

#### 3.7.3 Visualizing Potential Outliers

To complement the statistical assessment, I used boxplots to examine the distributions of selected numerical variables.

Boxplots provide a visual representation of the median, interquartile range, and observations beyond the whisker boundaries.

The visual assessment focuses on enrollment, per-pupil expenditure, student-teacher ratio, and mathematics proficiency.

These plots will help identify extreme observations and guide further investigation without automatically treating all flagged values as errors.

```python
# Import visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Select variables for visualization
plot_columns = [
    "total_enrollment",
    "per_pupil_expenditure_usd",
    "student_teacher_ratio",
    "math_proficiency_pct"
]

# Create separate boxplots for the selected variables
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, column in zip(axes.flat, plot_columns):

    # Plot the variable distribution
    sns.boxplot(
        x=edu_clean[column],
        ax=ax
    )

    # Add descriptive titles and axis labels
    ax.set_title(column.replace("_", " ").title())
    ax.set_xlabel(column.replace("_", " ").title())

# Improve spacing between plots
plt.tight_layout()

# Display the plots
plt.show()
```

#### 3.7.4 Investigating Outliers Across Institution Levels

Since institution size and educational resources may differ across K–12 school levels, I further examined potential expenditure outliers within each educational level.

This assessment provides additional context for determining whether observations flagged in the combined dataset also appear unusual within their own institutional categories.

The objective is to avoid treating legitimate differences between institution types as data-quality errors.

```python
# Store expenditure outlier counts by institution level
level_outlier_results = []

# Examine each educational level separately
for level, group in edu_clean.groupby("level"):

    # Calculate the IQR for per-pupil expenditure
    Q1 = group["per_pupil_expenditure_usd"].quantile(0.25)
    Q3 = group["per_pupil_expenditure_usd"].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Count outliers within the institution level
    outliers = (
        (group["per_pupil_expenditure_usd"] < lower_bound) |
        (group["per_pupil_expenditure_usd"] > upper_bound)
    )

    level_outlier_results.append({
        "Institution Level": level,
        "Total Institutions": len(group),
        "Expenditure Outliers": int(outliers.sum())
    })

# Display the group-specific results
level_outlier_summary = pd.DataFrame(level_outlier_results)

level_outlier_summary
```

#### 3.7.5 Outlier Treatment Decision

The IQR assessment identified potential outliers in enrollment, teacher counts, student-teacher ratios, educational expenditure, mathematics proficiency, and broadband access.

No statistical outliers were identified in the free/reduced-price lunch percentage variable.

Further investigation demonstrated that outlier counts vary across institution levels, indicating that institutional context is relevant when interpreting extreme observations.

Although some values are statistically unusual, the numerical validation checks did not establish that these observations represent data entry errors or invalid measurements.

Automatically removing or replacing them could eliminate meaningful institutional variation and distort the relationships under investigation.

**Cleaning Decision:** The flagged observations will be retained in the analytical dataset.

No outlier removal, median replacement, or capping will be performed at this stage.

The influence of extreme values will be considered during subsequent exploratory analysis, statistical investigation, and machine learning evaluation.

Any future outlier treatment used specifically for predictive modelling will be fitted using training data only to prevent data leakage.

This decision preserves the observed institutional variation while maintaining transparency about potential outlier effects.

#### 4.2.7 Graduation Rate Distribution

Graduation rate represents the percentage of students completing the relevant educational programme within an institution.

I examined the availability of graduation-rate measurements across the different K–12 institution levels before conducting the distribution analysis.

Since graduation is primarily associated with the completion of secondary education, I restricted the main analysis to High School institutions to maintain greater conceptual comparability.

K-12 Combined institutions will be treated separately because the dataset documentation does not establish whether their graduation-rate measurements use the same definition as those reported by standalone High Schools.

```python
# Examine graduation-rate availability across institution levels
graduation_availability = (
    edu_clean.groupby("level")["graduation_rate_pct"]
    .agg(
        Total_Institutions="size",
        Recorded_Values="count",
        Missing_Values="size"
    )
)

# Correct the missing-value calculation
graduation_availability["Missing_Values"] = (
    graduation_availability["Total_Institutions"]
    - graduation_availability["Recorded_Values"]
)

# Display the availability summary
graduation_availability
```

#### Create our graduation analysis dataset

```python
# Create a separate dataset containing High School institutions
graduation_data = edu_clean.loc[
    edu_clean["level"] == "High School"
].copy()

# Examine the resulting dataset
print("High School Institutions:", len(graduation_data))

print(
    "Missing Graduation Rate Values:",
    graduation_data["graduation_rate_pct"].isna().sum()
)

print("Dataset Shape:", graduation_data.shape)
```

#### 4.2.7.1 Descriptive Statistics of Graduation Rates

Following the institution-level assessment, I restricted the primary graduation-rate analysis to the 1,209 High School institutions.

I examined the descriptive statistics to understand the average graduation rate, median, variability, minimum and maximum values, and distribution of graduation outcomes.

This provides an initial understanding of graduation performance across the selected High School institutions.

```python
# Examine descriptive statistics for High School graduation rates
graduation_summary = (
    graduation_data["graduation_rate_pct"].describe()
)

# Display the summary statistics
graduation_summary
```

#### 4.2.7.2 Visualizing Graduation Rate Distribution

To complement the descriptive statistics, I used a histogram and boxplot to visualize graduation rates across the selected High School institutions.

The histogram illustrates the frequency and distribution of graduation rates, while the boxplot provides a summary of the median, interquartile range, and potential extreme observations.

Together, these visualizations help assess the consistency of graduation outcomes and identify institutions with unusually low or high graduation rates.

```python
# Create a figure containing a histogram and boxplot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram showing the distribution of graduation rates
sns.histplot(
    data=graduation_data,
    x="graduation_rate_pct",
    bins=30,
    kde=True,
    ax=axes[0]
)

axes[0].set_title("Distribution of Graduation Rates")
axes[0].set_xlabel("Graduation Rate (%)")
axes[0].set_ylabel("Number of High Schools")

# Boxplot showing the spread and potential outliers
sns.boxplot(
    data=graduation_data,
    x="graduation_rate_pct",
    ax=axes[1]
)

axes[1].set_title("Boxplot of Graduation Rates")
axes[1].set_xlabel("Graduation Rate (%)")

# Improve spacing between visualizations
plt.tight_layout()

# Display the plots
plt.show()
```

#### Investigate the distribution's skewness

```python
# Calculate the skewness of High School graduation rates
graduation_skewness = (
    graduation_data["graduation_rate_pct"].skew()
)

print(
    "Graduation Rate Skewness:",
    round(graduation_skewness, 2)
)
```

#### Investigate potential outliers

```python
# Calculate the first and third quartiles
Q1 = graduation_data["graduation_rate_pct"].quantile(0.25)
Q3 = graduation_data["graduation_rate_pct"].quantile(0.75)

# Calculate the interquartile range
IQR = Q3 - Q1

# Calculate the outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify potential outliers
graduation_outliers = (
    (graduation_data["graduation_rate_pct"] < lower_bound) |
    (graduation_data["graduation_rate_pct"] > upper_bound)
)

# Display the results
print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))

print("Lower Boundary:", round(lower_bound, 2))
print("Upper Boundary:", round(upper_bound, 2))

print(
    "Potential Outliers:",
    graduation_outliers.sum()
)
```

The graduation-rate analysis was conducted using the 1,209 High School institutions to maintain greater conceptual comparability between the institutions being examined.

The middle 50% of institutions have graduation rates between 80.30% and 88.30%, while the observed values range from 63.90% to 99.50%.

The histogram and calculated skewness of approximately -0.08 indicate an approximately symmetrical distribution with minimal negative skewness.

The IQR assessment identified six potential lower outliers, representing approximately 0.50% of the selected High School institutions.

**Finding:** Graduation rates are generally concentrated around the mid-80% range across the selected High School institutions, although differences exist between institutions with lower and higher graduation outcomes.

**Analytical Implication:** Graduation rates provide an additional measure of educational outcomes and can help investigate institutional characteristics associated with successful secondary-school completion.

Subsequent analysis will consider the relationship between graduation rates, institutional resources, socioeconomic conditions, and other relevant educational indicators.

The identified potential outliers were retained because unusually low graduation rates may represent meaningful educational challenges rather than erroneous observations.

The analysis was restricted to High School institutions because the available documentation does not establish the comparability of graduation-rate measurements across all K–12 institution levels.

#### Generate the consolidated descriptive statistics

```python
# Define the variables included in the univariate analysis
analysis_variables = {
    "Student Enrollment": edu_clean["total_enrollment"],
    "Student-Teacher Ratio": edu_clean["student_teacher_ratio"],
    "Per-Pupil Expenditure": edu_clean["per_pupil_expenditure_usd"],
    "Broadband Access": edu_clean["broadband_access_pct"],
    "Mathematics Proficiency": edu_clean["math_proficiency_pct"],
    "Reading Proficiency": edu_clean["reading_proficiency_pct"],
    "Graduation Rate (High School)": edu_clean.loc[
        edu_clean["level"] == "High School",
        "graduation_rate_pct"
    ],
    "Chronic Absenteeism": edu_clean["chronic_absenteeism_pct"]
}

# Create an empty list to store the summary results
summary_results = []

# Calculate descriptive statistics for each variable
for variable_name, values in analysis_variables.items():

    # Exclude missing observations for the current variable
    values = values.dropna()

    # Calculate the first and third quartiles
    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)

    # Calculate the interquartile range
    IQR = Q3 - Q1

    # Calculate the IQR outlier boundaries
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify potential outliers
    outlier_mask = (
        (values < lower_bound) |
        (values > upper_bound)
    )

    # Store the descriptive statistics
    summary_results.append({
        "Variable": variable_name,
        "Count": values.count(),
        "Mean": values.mean(),
        "Median": values.median(),
        "Standard Deviation": values.std(),
        "Minimum": values.min(),
        "Q1": Q1,
        "Q3": Q3,
        "Maximum": values.max(),
        "Skewness": values.skew(),
        "IQR Outliers": outlier_mask.sum(),
        "Outlier Percentage": outlier_mask.mean() * 100
    })

# Convert the results into a DataFrame
univariate_summary = pd.DataFrame(summary_results)

# Round numerical results for readability
univariate_summary = univariate_summary.round(2)

# Display the consolidated summary
univariate_summary
```

#### 4.2.9.1 Overall Observation and Key Findings

The univariate analysis provided an overview of institutional characteristics and educational outcomes across the 4,642 K–12 institutions.

The following findings emerged:

1. **Institutional Size:** Enrollment varies considerably across institutions, with an average of approximately 672 students and a positively skewed distribution.

2. **Staffing Conditions:** The average student-teacher ratio is approximately 14.79, with relatively little distributional asymmetry.

3. **Educational Expenditure:** Average per-pupil expenditure is approximately $16,814.07, indicating considerable variation in institutional spending.

4. **Technology Access:** Average broadband access is approximately 90.39%, although some institutions report substantially lower connectivity.

5. **Academic Performance:** Average mathematics proficiency is approximately 58.66%, while reading proficiency is approximately 63.55%. Both outcomes demonstrate considerable variation across institutions.

6. **Graduation Outcomes:** Among the 1,209 High School institutions, the average graduation rate is approximately 84.36%, with relatively limited variation compared with mathematics and reading proficiency.

7. **Attendance Conditions:** Average chronic absenteeism is approximately 20.23%, although substantial differences exist between institutions.

The analysis also identified potential statistical outliers across several variables. These observations were retained because their statistical classification does not establish that they represent invalid measurements.

**Overall Finding:** The univariate analysis reveals considerable differences in institutional size, educational resources, technology access, and academic outcomes across the selected K–12 institutions.

**Analytical Implication:** These differences provide a foundation for investigating whether institutional resources and socioeconomic conditions are associated with variations in educational outcomes.

However, univariate analysis examines variables individually and cannot establish relationships between them. Further investigation is therefore required to understand how these characteristics relate to mathematics proficiency and other academic outcomes.

The findings will guide the subsequent bivariate and multivariate analyses.

### 4.3 Bivariate Analysis

#### 4.3.1.1 Preparing Variables for Bivariate Analysis

I selected per-pupil expenditure and mathematics proficiency to investigate the relationship between institutional spending and academic performance.

Both variables are numerical, making them suitable for correlation analysis and scatterplot visualization.

I will first isolate the two variables and remove any missing observations to ensure that the statistical calculations use complete pairs of observations.

```python
# Select the two variables for the analysis
funding_math = edu_clean[
    [
        "per_pupil_expenditure_usd",
        "math_proficiency_pct"
    ]
].copy()

# Check for missing values
print("Missing values before treatment:")
print(funding_math.isnull().sum())

# Remove rows with missing values in either variable
funding_math = funding_math.dropna()

# Check the resulting dataset
print("\nDataset dimensions:", funding_math.shape)

# Display the first five observations
funding_math.head()
```

#### 4.3.1.2 Visualizing the Relationship Between Expenditure and Mathematics Proficiency

I used a scatterplot to examine whether institutions with higher per-pupil expenditure tend to report higher mathematics proficiency.

A regression line is included to help identify the overall direction of the relationship.

The visualization provides an initial indication of the relationship before conducting formal statistical analysis.

```python
# Set the figure size
plt.figure(figsize=(10, 6))

# Create a scatterplot with a regression line
sns.regplot(
    data=funding_math,
    x="per_pupil_expenditure_usd",
    y="math_proficiency_pct",
    scatter_kws={"alpha": 0.35, "s": 25},
    line_kws={"linewidth": 2}
)

# Add chart title and labels
plt.title(
    "Per-Pupil Expenditure vs Mathematics Proficiency",
    fontsize=14
)

plt.xlabel("Per-Pupil Expenditure (USD)", fontsize=11)
plt.ylabel("Mathematics Proficiency (%)", fontsize=11)

# Display the chart
plt.tight_layout()
plt.show()
```

#### 4.3.1.3 Pearson Correlation Analysis

To quantify the relationship between per-pupil expenditure and mathematics proficiency, I used Pearson's correlation coefficient.

The coefficient measures the direction and strength of the linear association between the two numerical variables.

I will also examine the p-value to determine whether the observed correlation provides sufficient statistical evidence against the null hypothesis at a 5% significance level.

```python
from scipy.stats import pearsonr

# Calculate Pearson correlation
pearson_corr, pearson_pvalue = pearsonr(
    funding_math["per_pupil_expenditure_usd"],
    funding_math["math_proficiency_pct"]
)

# Display the results
print("Pearson Correlation Analysis")
print("-" * 40)

print(f"Correlation coefficient: {pearson_corr:.4f}")
print(f"P-value: {pearson_pvalue:.6e}")

# Interpret the direction and strength
if pearson_corr > 0:
    direction = "Positive"
elif pearson_corr < 0:
    direction = "Negative"
else:
    direction = "No linear"

absolute_corr = abs(pearson_corr)

if absolute_corr < 0.1:
    strength = "Very weak"
elif absolute_corr < 0.3:
    strength = "Weak"
elif absolute_corr < 0.5:
    strength = "Moderate"
elif absolute_corr < 0.7:
    strength = "Strong"
else:
    strength = "Very strong"

print(f"\nDirection: {direction}")
print(f"Strength: {strength}")

# Interpret statistical significance
alpha = 0.05

if pearson_pvalue < alpha:
    print("Statistical decision: Reject the null hypothesis.")
    print("There is evidence of a statistically significant linear relationship.")
else:
    print("Statistical decision: Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a statistically significant linear relationship.")
```

#### 4.3.1.4 Spearman Rank Correlation

As an additional validation, I used Spearman's rank correlation to examine whether the relationship remains consistent when the observations are evaluated by their ranks rather than their original numerical values.

This provides a useful comparison with Pearson correlation and helps assess whether the positive association is also present as a general monotonic relationship.

```python
from scipy.stats import spearmanr

# Calculate Spearman rank correlation
spearman_corr, spearman_pvalue = spearmanr(
    funding_math["per_pupil_expenditure_usd"],
    funding_math["math_proficiency_pct"]
)

# Display the results
print("Spearman Correlation Analysis")
print("-" * 40)

print(f"Spearman coefficient: {spearman_corr:.4f}")
print(f"P-value: {spearman_pvalue:.6e}")
```

#### 4.3.1.5 Comparing Academic Performance Across Expenditure Quartiles

To make the relationship easier to interpret, I divided institutions into four groups based on their per-pupil expenditure.

Each group represents approximately one-quarter of the institutions.

I then calculated the average mathematics proficiency for each expenditure group to investigate whether academic performance generally differs across spending levels.

```python
# Create expenditure quartiles
funding_math["expenditure_group"] = pd.qcut(
    funding_math["per_pupil_expenditure_usd"],
    q=4,
    labels=[
        "Q1 - Lowest",
        "Q2",
        "Q3",
        "Q4 - Highest"
    ]
)

# Calculate summary statistics for each expenditure group
quartile_summary = (
    funding_math
    .groupby("expenditure_group", observed=True)
    .agg(
        Number_of_Institutions=(
            "math_proficiency_pct",
            "count"
        ),
        Average_Expenditure=(
            "per_pupil_expenditure_usd",
            "mean"
        ),
        Average_Math_Proficiency=(
            "math_proficiency_pct",
            "mean"
        ),
        Median_Math_Proficiency=(
            "math_proficiency_pct",
            "median"
        )
    )
    .round(2)
)

# Display the results
quartile_summary
```

#### 4.3.1.6 Findings and Conclusion

The investigation revealed a positive association between per-pupil expenditure and mathematics proficiency across the 4,642 institutions.

Pearson correlation produced a coefficient of 0.2874, indicating a weak positive linear relationship. The associated p-value was considerably below the 0.05 significance threshold. Therefore, I rejected the null hypothesis of no linear association.

Spearman correlation also produced a positive coefficient of 0.2688, providing additional evidence of a positive monotonic association.

The expenditure quartile comparison further supported this observation. Institutions in the lowest expenditure quartile recorded average mathematics proficiency of 53.76%, compared with 65.05% among institutions in the highest expenditure quartile.

This represents an observed difference of 11.29 percentage points.

**Conclusion:**

The findings suggest that higher per-pupil expenditure is associated with higher mathematics proficiency across the institutions examined.

However, the relatively weak correlation indicates that expenditure alone provides limited explanatory information about differences in academic performance.

The relationship may also be influenced by socioeconomic conditions, institutional characteristics, staffing, ownership, and other factors.

Importantly, this analysis establishes association rather than causation. Further multivariate analysis will be necessary to investigate whether the observed relationship remains after accounting for other relevant institutional characteristics.

#### 4.3.2.5 Pearson Correlation Analysis

To quantify the relationship between free or reduced-price lunch eligibility and mathematics proficiency, I calculated Pearson's correlation coefficient.

The analysis evaluates the direction and strength of the linear association between the socioeconomic indicator and the primary academic outcome.

I also examined the p-value using a significance level of 0.05 to determine whether the observed association provides statistically significant evidence against the null hypothesis of no linear correlation.

```python
# Select the socioeconomic indicator and mathematics outcome
x = socioeconomic_data["percent_free_reduced_lunch"]
y = socioeconomic_data["math_proficiency_pct"]

# Calculate Pearson correlation and p-value
socio_math_corr, socio_math_pvalue = pearsonr(x, y)

# Display the results
print("Pearson Correlation Analysis")
print("-" * 45)

print(f"Correlation coefficient: {socio_math_corr:.4f}")
print(f"P-value: {socio_math_pvalue:.6e}")

# Interpret the direction of the relationship
if socio_math_corr > 0:
    direction = "Positive"
elif socio_math_corr < 0:
    direction = "Negative"
else:
    direction = "No linear relationship"

# Interpret the strength using absolute correlation
absolute_corr = abs(socio_math_corr)

if absolute_corr < 0.10:
    strength = "Very weak"
elif absolute_corr < 0.30:
    strength = "Weak"
elif absolute_corr < 0.50:
    strength = "Moderate"
elif absolute_corr < 0.70:
    strength = "Strong"
else:
    strength = "Very strong"

print(f"Direction: {direction}")
print(f"Strength: {strength}")

# Test statistical significance
alpha = 0.05

if socio_math_pvalue < alpha:
    print("Decision: Reject the null hypothesis.")
    print("The linear relationship is statistically significant.")
else:
    print("Decision: Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a linear relationship.")
```

#### 4.3.2.6 Spearman Rank Correlation

To complement the Pearson correlation analysis, I calculated Spearman's rank correlation coefficient.

This assessment examines whether the negative association between free or reduced-price lunch eligibility and mathematics proficiency is also evident when the observations are compared by their ranks.

Comparing both correlation coefficients provides additional information about the consistency of the observed relationship.

```python
# Calculate Spearman rank correlation
socio_math_spearman, socio_math_spearman_pvalue = spearmanr(
    socioeconomic_data["percent_free_reduced_lunch"],
    socioeconomic_data["math_proficiency_pct"]
)

# Display the results
print("Spearman Correlation Analysis")
print("-" * 45)

print(f"Spearman coefficient: {socio_math_spearman:.4f}")
print(f"P-value: {socio_math_spearman_pvalue:.6e}")

# Compare Pearson and Spearman correlation
print("\nCorrelation Comparison")
print(f"Pearson:  {socio_math_corr:.4f}")
print(f"Spearman: {socio_math_spearman:.4f}")
```

#### 4.3.2.7 Mathematics Proficiency Across Socioeconomic Quartiles

To provide a clearer comparison of academic outcomes across different socioeconomic conditions, I divided institutions into four groups based on their free or reduced-price lunch eligibility percentages.

The first quartile represents institutions with the lowest eligibility percentages, while the fourth quartile represents institutions with the highest eligibility percentages.

I then compared average and median mathematics proficiency across these groups to investigate whether academic outcomes differ systematically with the socioeconomic indicator.

Pearson correlation produced a coefficient of approximately -0.5876, indicating a moderately strong negative linear relationship. The associated p-value was below 0.001, providing sufficient evidence to reject the null hypothesis of no linear correlation.

Spearman rank correlation produced a similar coefficient of approximately -0.5809, indicating that the negative association was also present when comparing the ranks of the observations.

The quartile comparison further supported these findings.

Institutions in the lowest free or reduced-price lunch eligibility quartile recorded average mathematics proficiency of approximately 70.87%, compared with 46.39% among institutions in the highest eligibility quartile.

**Finding:** Institutions with higher proportions of students eligible for free or reduced-price lunch generally report lower mathematics proficiency.

Further multivariate analysis will be required to investigate whether the relationship remains after accounting for institutional expenditure, staffing and broadband access.

#### Visualize the quartile comparison

```python
# Create a bar chart comparing average reading proficiency
plt.figure(figsize=(11, 6))

sns.barplot(
    data=reading_quartile_summary.reset_index(),
    x="lunch_eligibility_group",
    y="Average_Reading_Proficiency",
    color="steelblue"
)

# Add chart title and axis labels
plt.title(
    "Average Reading Proficiency by Lunch Eligibility Quartile",
    fontsize=14
)

plt.xlabel("Free/Reduced Lunch Eligibility Group")
plt.ylabel("Average Reading Proficiency (%)")

# Start the vertical axis at zero
plt.ylim(0, 90)

# Rotate labels for readability
plt.xticks(rotation=15)

# Improve chart layout
plt.tight_layout()

# Display the chart
plt.show()
```

#### 4.3.2.13 Observation and Findings

The investigation revealed a statistically significant negative association between free or reduced-price lunch eligibility and reading proficiency across the 4,642 K–12 institutions.

Pearson correlation produced a coefficient of approximately -0.5890, indicating a moderately strong negative linear relationship. The associated p-value was below the 0.05 significance threshold, providing sufficient evidence to reject the null hypothesis of no linear correlation.

Spearman rank correlation produced a similar coefficient of approximately -0.5817, providing additional evidence of a consistent negative association.

The socioeconomic quartile comparison further supported these findings.

Institutions in the lowest free or reduced-price lunch eligibility quartile recorded average reading proficiency of approximately 75.69%, compared with 51.45% among institutions in the highest eligibility quartile.

This represents a difference of approximately 24.24 percentage points.

**Finding:** Institutions with higher proportions of students eligible for free or reduced-price lunch generally report lower reading proficiency.

**Comparative Observation:** The negative socioeconomic association is evident in both mathematics and reading proficiency, suggesting that the observed pattern extends across more than one academic outcome.

**Analytical Implication:** Socioeconomic conditions, as represented by free or reduced-price lunch eligibility, appear to be meaningfully associated with differences in academic performance.

However, the analysis does not establish causation. The eligibility percentage is a proxy indicator and does not capture every dimension of socioeconomic status. Other institutional, demographic, geographic and resource-related characteristics may also contribute to the observed differences.

Further multivariate analysis will be necessary to investigate whether these associations remain after accounting for other relevant institutional characteristics.

### 4.3.3 Broadband Access and Academic Performance

**First, what are we investigating?**
Broadband access is an indicator of technological resources available within educational institutions.

Access to reliable internet connectivity may support digital learning, online educational resources, research, and technology-based instruction.

However, institutions with better broadband access may also differ in funding, socioeconomic conditions, geographic location, and other characteristics.

Therefore, we cannot assume that broadband access alone determines academic performance.
**Our research question**

*Is there a statistically significant relationship between broadband access and mathematics proficiency across K–12 institutions?*

Mathematics proficiency remains the primary academic outcome of this investigation, while broadband access represents one of the institutional characteristics identified in the research question.

**Our hypotheses**

**- Null hypothesis (H₀):** There is no linear association between broadband access and mathematics proficiency.

**- Alternative hypothesis (H₁):** There is a linear association between broadband access and mathematics proficiency.
We'll use a significance level of 0.05.

Our initial expectation is that institutions with higher broadband access may report higher mathematics proficiency.

That is a hypothesis to investigate, not an established conclusion.

Here, I investigated the relationship between broadband access and mathematics proficiency to assess whether institutions with higher reported connectivity tend to demonstrate stronger academic outcomes.

The analysis uses scatterplot visualization, Pearson and Spearman correlation, and quartile comparisons to examine the direction, strength, and consistency of the association.

The findings will help establish whether broadband access deserves further consideration in the subsequent multivariate analysis and predictive modelling.

```python

# Select broadband access and mathematics proficiency
broadband_math = edu_clean[
    [
        "broadband_access_pct",
        "math_proficiency_pct"
    ]
].copy()

# Examine missing values
print("Missing Values:")
print(broadband_math.isnull().sum())

# Remove incomplete observation pairs, if any
broadband_math = broadband_math.dropna()

# Confirm the number of observations
print("\nDataset Dimensions:", broadband_math.shape)

# Display the first five observations
broadband_math.head()
```

#### 4.3.3.1 Visualizing Broadband Access and Mathematics Proficiency

I used a scatterplot with a fitted regression line to examine the relationship between broadband access and mathematics proficiency across the selected K–12 institutions.

The visualization helps identify the direction of the association, the extent of variation in academic outcomes, and any unusual observations.

The visual assessment provides an initial understanding of the relationship before formal correlation analysis.

```python

# Create a scatterplot with a fitted regression line
plt.figure(figsize=(10, 6))

sns.regplot(
    data=broadband_math,
    x="broadband_access_pct",
    y="math_proficiency_pct",
    scatter_kws={
        "alpha": 0.35,
        "s": 25
    },
    line_kws={
        "color": "red",
        "linewidth": 2
    }
)

# Add chart title and axis labels
plt.title(
    "Broadband Access vs Mathematics Proficiency",
    fontsize=14
)

plt.xlabel("Broadband Access (%)")
plt.ylabel("Mathematics Proficiency (%)")

# Improve chart layout
plt.tight_layout()

# Display the chart
plt.show()
```

#### 4.3.3.2 Pearson Correlation Analysis

To quantify the relationship between broadband access and mathematics proficiency, I calculated Pearson's correlation coefficient.

The coefficient measures the strength and direction of the linear association between the two variables.

I also examined the p-value using a significance level of 0.05 to determine whether the observed association provides statistically significant evidence against the null hypothesis of no linear relationship.

The analysis is observational and does not establish causation.

```python

# Calculate Pearson correlation and its p-value
broadband_math_corr, broadband_math_pvalue = pearsonr(
    broadband_math["broadband_access_pct"],
    broadband_math["math_proficiency_pct"]
)

# Calculate the squared correlation coefficient
broadband_math_r2 = broadband_math_corr ** 2

# Display the correlation results
print("Pearson Correlation Analysis")
print("-" * 45)

print(f"Correlation Coefficient: {broadband_math_corr:.4f}")
print(f"P-value: {broadband_math_pvalue:.6e}")
print(f"R-squared: {broadband_math_r2:.4f}")

# Interpret the direction
if broadband_math_corr > 0:
    direction = "Positive"
elif broadband_math_corr < 0:
    direction = "Negative"
else:
    direction = "No linear relationship"

# Interpret the strength
absolute_corr = abs(broadband_math_corr)

if absolute_corr < 0.10:
    strength = "Very weak"
elif absolute_corr < 0.30:
    strength = "Weak"
elif absolute_corr < 0.50:
    strength = "Moderate"
elif absolute_corr < 0.70:
    strength = "Strong"
else:
    strength = "Very strong"

print(f"\nDirection: {direction}")
print(f"Strength: {strength}")

# Evaluate statistical significance
alpha = 0.05

if broadband_math_pvalue < alpha:
    print("Decision: Reject the null hypothesis.")
    print("The linear association is statistically significant.")
else:
    print("Decision: Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a linear association.")
```

#### 4.3.3.3 Spearman Rank Correlation

To complement the Pearson correlation results, I calculated Spearman's rank correlation coefficient.

This method evaluates whether institutions with higher broadband access generally rank higher in mathematics proficiency.

Comparing both coefficients provides additional information about the direction and consistency of the observed association.

```python
# Calculate Spearman rank correlation
broadband_math_spearman, broadband_math_spearman_pvalue = spearmanr(
    broadband_math["broadband_access_pct"],
    broadband_math["math_proficiency_pct"]
)

# Display the results
print("Spearman Correlation Analysis")
print("-" * 45)

print(f"Spearman Coefficient: {broadband_math_spearman:.4f}")
print(f"P-value: {broadband_math_spearman_pvalue:.6e}")

# Compare Pearson and Spearman correlations
print("\nCorrelation Comparison")
print(f"Pearson:  {broadband_math_corr:.4f}")
print(f"Spearman: {broadband_math_spearman:.4f}")
```

#### 4.3.3.4 Mathematics Proficiency Across Broadband Access Quartiles

To further investigate the observed positive association, I divided institutions into four groups based on their broadband access percentages.

The first quartile represents institutions with the lowest broadband access, while the fourth quartile represents those with the highest access.

I compared average and median mathematics proficiency across the groups to assess whether academic outcomes differ systematically with reported broadband connectivity.

```python

# Create a separate DataFrame for broadband quartile analysis
broadband_quartile_data = broadband_math.copy()

# Divide institutions into four broadband access groups
broadband_quartile_data["broadband_group"] = pd.qcut(
    broadband_quartile_data["broadband_access_pct"],
    q=4,
    labels=[
        "Q1 - Lowest Access",
        "Q2",
        "Q3",
        "Q4 - Highest Access"
    ]
)

# Calculate group-level descriptive statistics
broadband_quartile_summary = (
    broadband_quartile_data
    .groupby("broadband_group", observed=True)
    .agg(
        Number_of_Institutions=(
            "math_proficiency_pct",
            "count"
        ),
        Average_Broadband_Access=(
            "broadband_access_pct",
            "mean"
        ),
        Average_Math_Proficiency=(
            "math_proficiency_pct",
            "mean"
        ),
        Median_Math_Proficiency=(
            "math_proficiency_pct",
            "median"
        )
    )
    .round(2)
)

# Display the results
broadband_quartile_summary
```

```python

# Create a bar chart comparing mathematics proficiency
plt.figure(figsize=(11, 6))

sns.barplot(
    data=broadband_quartile_summary.reset_index(),
    x="broadband_group",
    y="Average_Math_Proficiency",
    color="steelblue"
)

# Add chart title and labels
plt.title(
    "Average Mathematics Proficiency by Broadband Access Quartile",
    fontsize=14
)

plt.xlabel("Broadband Access Group")
plt.ylabel("Average Mathematics Proficiency (%)")

# Begin the vertical axis at zero
plt.ylim(0, 80)

# Rotate category labels for readability
plt.xticks(rotation=15)

# Improve chart spacing
plt.tight_layout()

# Display the chart
plt.show()
```

#### 4.3.3.5 Observation and Findings

The investigation revealed a statistically significant positive association between broadband access and mathematics proficiency across the 4,642 K–12 institutions.

Pearson correlation produced a coefficient of approximately 0.4493, indicating a moderate positive linear relationship. The associated p-value was below 0.001, providing sufficient evidence to reject the null hypothesis of no linear association.

Spearman rank correlation produced a similar coefficient of approximately 0.4375, supporting the consistency of the positive association.

The broadband access quartile comparison further supported these findings.

Institutions in the lowest broadband access quartile recorded average mathematics proficiency of approximately 49.70%, compared with 68.18% among institutions in the highest access quartile.

This represents a difference of approximately 18.48 percentage points.

**Finding:** Institutions with higher reported broadband access generally demonstrate higher mathematics proficiency.

**Comparative Observation:** Broadband access shows a stronger positive correlation with mathematics proficiency than per-pupil expenditure in the individual bivariate analyses.

**Analytical Implication:** Technology access appears to be meaningfully associated with academic outcomes and may provide useful information when investigating educational opportunity and developing predictive models.

However, the analysis does not establish causation. Institutions with higher broadband access may also differ in socioeconomic conditions, expenditure, staffing, geographic location, and other characteristics.

Further multivariate analysis will be necessary to investigate whether the observed association remains after accounting for these additional factors.

##### Three important questions

**1. Is the relationship positive or negative?**

Positive. Higher broadband access generally corresponds with higher mathematics proficiency.

**2. Is the relationship statistically significant?**

Yes. The p-value is considerably smaller than 0.05.

**3. Is broadband access more important than expenditure?**

We cannot establish that from separate correlations alone. Broadband access has a larger correlation coefficient in this dataset, but we need multivariate analysis to compare their associations while accounting for other characteristics.

I compared the average and median mathematics proficiency across these groups to assess whether academic outcomes differ systematically with staffing ratios.

This comparison provides a descriptive perspective on the relationship identified through the correlation analysis.

```python

# Create a separate DataFrame for quartile analysis
ratio_quartile_data = ratio_math.copy()

# Divide institutions into four student-teacher ratio groups
ratio_quartile_data["ratio_group"] = pd.qcut(
    ratio_quartile_data["student_teacher_ratio"],
    q=4,
    labels=[
        "Q1 - Lowest Ratio",
        "Q2",
        "Q3",
        "Q4 - Highest Ratio"
    ]
)

# Calculate statistics for each group
ratio_quartile_summary = (
    ratio_quartile_data
    .groupby("ratio_group", observed=True)
    .agg(
        Number_of_Institutions=(
            "math_proficiency_pct",
            "count"
        ),
        Average_Student_Teacher_Ratio=(
            "student_teacher_ratio",
            "mean"
        ),
        Average_Math_Proficiency=(
            "math_proficiency_pct",
            "mean"
        ),
        Median_Math_Proficiency=(
            "math_proficiency_pct",
            "median"
        )
    )
    .round(2)
)

# Display the results
ratio_quartile_summary
```

##### Understanding the results

The lowest student-teacher ratio group has the highest average mathematics proficiency, at 62.01%.

The highest ratio group records an average of 57.15%.

The difference is:
\[
62.01-57.15=4.86
\]

Difference in average mathematics proficiency

**4.86 percentage points**

Between institutions in the lowest and highest student-teacher ratio quartiles
There is an important detail here.

Average mathematics proficiency decreases from Q1 to Q3, but increases slightly in Q4.

Therefore, the quartile comparison does not show a perfectly consistent decline.

This supports our interpretation that the overall relationship is negative but weak.

```python

# Create a bar chart comparing mathematics proficiency
plt.figure(figsize=(11, 6))

sns.barplot(
    data=ratio_quartile_summary.reset_index(),
    x="ratio_group",
    y="Average_Math_Proficiency",
    color="steelblue"
)

# Add chart title and axis labels
plt.title(
    "Average Mathematics Proficiency by Student-Teacher Ratio Quartile",
    fontsize=14
)

plt.xlabel("Student-Teacher Ratio Group")
plt.ylabel("Average Mathematics Proficiency (%)")

# Begin the vertical axis at zero
plt.ylim(0, 70)

# Rotate labels for readability
plt.xticks(rotation=15)

# Improve chart layout
plt.tight_layout()

# Display the chart
plt.show()
```

#### 4.3.4.5 Observation and Findings

The investigation revealed a statistically significant but weak negative association between student-teacher ratio and mathematics proficiency across the 4,642 K–12 institutions.

Pearson correlation produced a coefficient of approximately -0.1156, indicating a weak negative linear relationship. The associated p-value was below 0.001, providing sufficient evidence to reject the null hypothesis of no linear association.

Spearman rank correlation produced a similar coefficient of approximately -0.1139, supporting the consistency of the weak negative association.

The quartile comparison revealed that institutions in the lowest student-teacher ratio group recorded average mathematics proficiency of approximately 62.01%, compared with 57.15% among institutions in the highest ratio group.

This represents a difference of approximately 4.86 percentage points.

However, average mathematics proficiency did not decrease consistently across all four groups, reinforcing the relatively weak overall association.

**Finding:** Institutions with lower student-teacher ratios generally report slightly higher mathematics proficiency, although the observed relationship is weak.

**Comparative Observation:** Student-teacher ratio has a smaller absolute correlation with mathematics proficiency than per-pupil expenditure, broadband access, and free or reduced-price lunch eligibility in the individual bivariate analyses.

**Analytical Implication:** Student-teacher ratio may provide some information about differences in academic outcomes, but it has limited explanatory value when considered independently.

The analysis does not establish that reducing student-teacher ratios would directly improve mathematics proficiency. Other institutional characteristics, including funding, socioeconomic conditions, school level, and technology access, may contribute to the observed relationship.

Further multivariate analysis will be required to investigate whether student-teacher ratio remains associated with mathematics proficiency after accounting for these additional factors.

#### 4.3.5.3 STEM Certification Across Institution Levels

I examined STEM certification across the four K–12 institution levels to determine whether the proportion of certified institutions differs by educational level.

This comparison is important because institutional characteristics may vary between elementary, middle, high school, and combined K–12 institutions.

I calculated both the number of certified institutions and the certification percentage within each institution level to avoid misleading comparisons caused by differences in group sizes.

```python

# Define the institution-level order
level_order = [
    "Elementary School",
    "Middle School",
    "High School",
    "K-12 Combined"
]

# Create a cross-tabulation of school level and STEM certification
stem_level_table = pd.crosstab(
    edu_clean["level"],
    edu_clean["stem_certified_flag"]
)

# Apply a consistent institution-level order
stem_level_table = stem_level_table.reindex(level_order)

# Calculate the total number of institutions per level
stem_level_table["Total"] = (
    stem_level_table["No"] + stem_level_table["Yes"]
)

# Calculate the percentage certified within each level
stem_level_table["Certified (%)"] = (
    stem_level_table["Yes"]
    / stem_level_table["Total"]
    * 100
).round(2)

# Display the cross-tabulation
stem_level_table
```

High Schools have the highest proportion of STEM-certified institutions at approximately 22.25%.

The other institution levels have certification percentages between approximately 11% and 13%.

This is a meaningful descriptive difference.

We should not simply compare the number of certified institutions because elementary schools are much more numerous in the dataset. Instead, the within-level percentages tell us how common certification is within each category.

```python
# Prepare certification percentages for visualization
stem_level_plot = (
    stem_level_table["Certified (%)"]
    .reset_index()
)

# Create a bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=stem_level_plot,
    x="level",
    y="Certified (%)",
    order=level_order,
    color="steelblue"
)

# Add chart title and labels
plt.title(
    "STEM Certification Rate Across Institution Levels",
    fontsize=14
)

plt.xlabel("Institution Level")
plt.ylabel("Institutions STEM-Certified (%)")

# Start the vertical axis at zero
plt.ylim(0, 30)

# Rotate labels for readability
plt.xticks(rotation=15)

# Improve layout
plt.tight_layout()

# Display the chart
plt.show()
```

### 4.3.6 Bivariate Analysis Summary and Key Findings

Following the individual bivariate investigations, I consolidated the findings to evaluate the relationships between institutional characteristics, socioeconomic conditions, and academic outcomes across the selected K–12 institutions.

The analysis examined per-pupil expenditure, student-teacher ratio, free or reduced-price lunch eligibility, and broadband access in relation to mathematics proficiency, the primary academic outcome of this project.

Additional comparisons involving reading proficiency, Title I status, and STEM certification provided supporting evidence about educational opportunity and institutional differences.

The objective of this summary is to identify the direction and relative strength of the observed associations, highlight important findings, and establish the rationale for the subsequent multivariate analysis.

All relationships are interpreted as observational associations rather than evidence of causation.

#### Consolidate our primary correlation results

```python
# Define the four primary predictors
primary_predictors = {
    "Free/Reduced Lunch Eligibility": "percent_free_reduced_lunch",
    "Broadband Access": "broadband_access_pct",
    "Per-Pupil Expenditure": "per_pupil_expenditure_usd",
    "Student-Teacher Ratio": "student_teacher_ratio"
}

# Define our primary academic outcome
target_variable = "math_proficiency_pct"

# Create an empty list to store correlation results
correlation_results = []

# Calculate Pearson and Spearman correlations
for predictor_name, predictor_column in primary_predictors.items():

    # Select complete observation pairs
    paired_data = edu_clean[
        [predictor_column, target_variable]
    ].dropna()

    # Calculate Pearson correlation and p-value
    pearson_corr, pearson_pvalue = pearsonr(
        paired_data[predictor_column],
        paired_data[target_variable]
    )

    # Calculate Spearman correlation and p-value
    spearman_corr, spearman_pvalue = spearmanr(
        paired_data[predictor_column],
        paired_data[target_variable]
    )

    # Store the results
    correlation_results.append({
        "Predictor": predictor_name,
        "Observations": len(paired_data),
        "Pearson Correlation": pearson_corr,
        "Spearman Correlation": spearman_corr,
        "Absolute Pearson": abs(pearson_corr),
        "Pearson P-value": pearson_pvalue
    })

# Convert the results into a DataFrame
bivariate_correlation_summary = pd.DataFrame(
    correlation_results
)

# Sort predictors by absolute Pearson correlation
bivariate_correlation_summary = (
    bivariate_correlation_summary
    .sort_values(
        by="Absolute Pearson",
        ascending=False
    )
    .reset_index(drop=True)
)

# Display a formatted summary
bivariate_correlation_summary[
    [
        "Predictor",
        "Observations",
        "Pearson Correlation",
        "Spearman Correlation",
        "Pearson P-value"
    ]
].round(4)
```

All four Pearson correlations have p-values below 0.001.

**Understanding the results:**

The table reveals four important patterns.

**1. Free/reduced lunch eligibility has the largest absolute correlation.**

Its Pearson coefficient is -0.5876.

This indicates a moderately strong negative relationship with mathematics proficiency.

Institutions reporting higher lunch eligibility generally report lower mathematics proficiency.

**2. Broadband access has the second-largest absolute correlation.**

Its coefficient is +0.4493.

This indicates a moderate positive relationship.

Institutions with higher broadband access generally report higher mathematics proficiency.

**3. Per-pupil expenditure has a weaker positive relationship.**

Its coefficient is +0.2874.

Higher expenditure tends to be associated with higher mathematics proficiency, although the association is relatively weak.

**4. Student-teacher ratio has the smallest absolute correlation.**

Its coefficient is -0.1156.

This indicates a weak negative relationship.

Higher student-teacher ratios are associated with slightly lower mathematics proficiency, but the relationship is much weaker than the other observed associations.

**Important:** These are rankings by the strength of individual linear associations, not rankings of causal importance.

#### 4.3.6.1 Comparison of Primary Predictors

To compare the relationships identified during the bivariate analysis, I visualized the Pearson correlation coefficients for the four primary predictors of mathematics proficiency.

The comparison highlights differences in the direction and strength of the observed linear associations.

The coefficients are ordered by their absolute values to identify which institutional characteristics demonstrate the strongest individual relationships with mathematics proficiency.

These results provide preliminary evidence for selecting predictors for subsequent multivariate analysis and machine learning.

```python

# Create a horizontal bar chart comparing Pearson correlations
plt.figure(figsize=(11, 6))

# Use different colours for positive and negative correlations
bar_colors = [
    "steelblue" if value > 0 else "indianred"
    for value in bivariate_correlation_summary["Pearson Correlation"]
]

# Plot the correlation coefficients
bars = plt.barh(
    bivariate_correlation_summary["Predictor"],
    bivariate_correlation_summary["Pearson Correlation"],
    color=bar_colors
)

# Add a reference line at zero
plt.axvline(
    x=0,
    color="black",
    linewidth=1
)

# Add correlation values to each bar
for bar in bars:

    value = bar.get_width()

    # Position labels according to correlation direction
    if value >= 0:
        label_x = value + 0.02
        alignment = "left"
    else:
        label_x = value - 0.02
        alignment = "right"

    plt.text(
        label_x,
        bar.get_y() + bar.get_height() / 2,
        f"{value:+.3f}",
        va="center",
        ha=alignment
    )

# Add title and labels
plt.title(
    "Pearson Correlations with Mathematics Proficiency",
    fontsize=14
)

plt.xlabel("Pearson Correlation Coefficient")
plt.ylabel("Institutional Characteristics")

# Set comparable horizontal-axis limits
plt.xlim(-0.75, 0.75)

# Display strongest association at the top
plt.gca().invert_yaxis()

# Improve layout
plt.tight_layout()

# Display the chart
plt.show()
```

The chart makes the contrast between the four predictors easier to understand.

Free or reduced-price lunch eligibility has the largest negative correlation, while broadband access has the strongest positive correlation.

Per-pupil expenditure also has a positive association, whereas student-teacher ratio has a comparatively weak negative association.

The Pearson and Spearman results are reasonably consistent for all four predictors.

This gives us useful preliminary evidence, but it does not establish which predictors will remain informative when considered together.

### 4.4 Multivariate Analysis

The bivariate analysis established statistically significant associations between mathematics proficiency and the four primary institutional predictors: per-pupil expenditure, student-teacher ratio, free or reduced-price lunch eligibility, and broadband access.

However, these individual associations do not account for possible relationships among the predictors themselves.

The multivariate analysis will therefore examine the relationships among these institutional characteristics and investigate their combined associations with mathematics proficiency.

The objective is to determine whether the observed relationships remain when multiple institutional characteristics are considered simultaneously.

#### 4.4.1 Investigating Relationships Among the Primary Predictors

Before fitting a multivariate regression model, I examined the correlation structure of the four primary predictors.

This assessment helps identify strongly related predictors that may introduce multicollinearity and complicate the interpretation of regression coefficients.

I used a Pearson correlation matrix and heatmap to examine the direction and strength of relationships among the selected variables.

Mathematics proficiency was included in the correlation matrix to provide context, while multicollinearity assessment focuses on the relationships among the predictor variables.

#### Prepare the selected variables

```python

# Define the four primary predictors
primary_features = [
    "per_pupil_expenditure_usd",
    "student_teacher_ratio",
    "percent_free_reduced_lunch",
    "broadband_access_pct"
]

# Define our primary academic outcome
target_column = "math_proficiency_pct"

# Select the predictors and target variable
multivariate_data = edu_clean[
    primary_features + [target_column]
].copy()

# Examine missing values
print("Missing Values:")
print(multivariate_data.isnull().sum())

# Remove incomplete rows if any exist
multivariate_data = multivariate_data.dropna()

# Confirm the dataset dimensions
print("\nDataset Dimensions:", multivariate_data.shape)

# Display the first five observations
multivariate_data.head()
```

#### 4.4.1.1 Pearson Correlation Matrix

I calculated a Pearson correlation matrix for the four primary institutional predictors and mathematics proficiency.

The matrix provides a combined view of the direction and strength of the linear relationships between the selected variables.

Particular attention was given to correlations among the predictors because strong relationships between explanatory variables may create multicollinearity concerns during regression analysis.

The results provide a foundation for evaluating predictor relationships before fitting the multivariate model.

```python
# Calculate the Pearson correlation matrix
multivariate_corr = multivariate_data.corr(method="pearson")

# Display the correlation matrix rounded to four decimals
multivariate_corr.round(4)
```

#### 4.4.1.2 Correlation Heatmap

To visualize the relationships among the primary predictors and mathematics proficiency, I created a correlation heatmap.

The heatmap displays Pearson correlation coefficients using a diverging colour scale, making it easier to distinguish positive and negative relationships.

The visualization helps identify potentially overlapping predictor information and provides an initial assessment of possible multicollinearity.

The strength of relationships among predictors will be considered before fitting and interpreting the multivariate regression model.

```python

# Define shorter labels for the heatmap
correlation_labels = [
    "Per-Pupil Expenditure",
    "Student-Teacher Ratio",
    "Free/Reduced Lunch",
    "Broadband Access",
    "Math Proficiency"
]

# Create a copy of the correlation matrix
heatmap_data = multivariate_corr.copy()

# Apply readable labels
heatmap_data.index = correlation_labels
heatmap_data.columns = correlation_labels

# Create a mask for the upper triangle
# This avoids displaying duplicate correlation values
upper_triangle_mask = np.triu(
    np.ones_like(heatmap_data, dtype=bool),
    k=1
)

# Set the figure size
plt.figure(figsize=(10, 7))

# Create the correlation heatmap
sns.heatmap(
    heatmap_data,
    mask=upper_triangle_mask,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True,
    cbar_kws={"label": "Pearson Correlation"}
)

# Add chart title
plt.title(
    "Correlation Matrix of Institutional Predictors and Math Proficiency",
    fontsize=13,
    pad=15
)

# Improve label readability
plt.xticks(rotation=35, ha="right")
plt.yticks(rotation=0)

# Improve chart spacing
plt.tight_layout()

# Display the heatmap
plt.show()
```

**Three important observations emerge:**

**1. Socioeconomic conditions and broadband access are strongly related.**

This is our most important finding at this stage.

The correlation of -0.7569 suggests that institutions with higher lunch eligibility tend to have lower broadband access.

When we eventually fit a multiple regression model, we must examine whether this strong association affects the stability and interpretation of our regression coefficients.

**2. Expenditure has relatively weak relationships with the other predictors.**

Its correlations with staffing, lunch eligibility, and broadband access are all relatively small.

This suggests that expenditure may provide information that is not strongly duplicated by the other selected predictors, although further assessment is still necessary.

**3. Student-teacher ratio also has relatively weak correlations with the other predictors.**

This suggests limited pairwise overlap between staffing ratio and the other selected institutional characteristics.

However, weak pairwise correlations do not completely rule out multicollinearity involving combinations of predictors.

#### 4.4.1.3 Observation and Findings

The correlation matrix revealed differences in the relationships among the four primary institutional predictors.

The strongest relationship was observed between free or reduced-price lunch eligibility and broadband access, with a Pearson correlation coefficient of approximately -0.7569.

This indicates that institutions with higher lunch eligibility percentages generally report lower broadband access.

The relationship is particularly important because both variables demonstrated substantial individual associations with mathematics proficiency during the bivariate analysis.

Per-pupil expenditure showed relatively weak correlations with student-teacher ratio (-0.0630), lunch eligibility (-0.2344), and broadband access (0.1815).

Student-teacher ratio also showed relatively weak relationships with the other selected predictors.

**Finding:** Socioeconomic conditions and broadband access are strongly negatively correlated, suggesting that these predictors may contain overlapping information.

**Analytical Implication:** The strong relationship between lunch eligibility and broadband access raises a potential multicollinearity concern that should be investigated before interpreting their separate regression coefficients.

Although the remaining predictor pairs demonstrate relatively weak correlations, pairwise correlation analysis alone is insufficient to fully assess multicollinearity.

**Conclusion:** The selected predictors exhibit different degrees of association with one another. Further assessment is required to determine whether these relationships could affect the interpretation of a multiple regression model.

The next stage will evaluate multicollinearity more formally before investigating the combined relationships between institutional characteristics and mathematics proficiency.

#### 4.4.2 Multicollinearity Assessment Using Variance Inflation Factor (VIF)

The correlation matrix revealed a strong negative association between free or reduced-price lunch eligibility and broadband access (r = -0.7569).

Although this relationship suggests potential overlap between the two predictors, pairwise correlations alone are insufficient to fully assess multicollinearity.

To investigate this further, I calculated the Variance Inflation Factor (VIF) for each of the four primary predictors.

VIF measures the extent to which a predictor's regression coefficient variance may be inflated because of linear relationships with the other predictors.

For this investigation, I used a VIF threshold of 5 as a practical guideline for identifying potentially concerning multicollinearity.

The objective is to determine whether the selected predictors can be retained together for the subsequent multiple regression analysis.

#### Prepare the predictors for VIF calculation

```python

# Select the four primary predictors
vif_features = [
    "per_pupil_expenditure_usd",
    "student_teacher_ratio",
    "percent_free_reduced_lunch",
    "broadband_access_pct"
]

# Create a separate DataFrame containing only predictors
X_vif = multivariate_data[vif_features].copy()

# Check for missing values
print("Missing Values:")
print(X_vif.isnull().sum())

# Confirm the dataset dimensions
print("\nPredictor Dataset Dimensions:", X_vif.shape)

# Display the first five observations
X_vif.head()
```

#### 4.4.2.1 Calculating Variance Inflation Factors

I calculated the Variance Inflation Factor for each primary predictor using the `statsmodels` library.

An intercept was included in the calculation to ensure that the VIF assessment corresponds to a regression model with a constant term.

The resulting values were compared against the selected threshold of 5 to determine whether any predictor demonstrated potentially concerning multicollinearity.

The intercept was excluded from the final interpretation because the assessment focuses on the four institutional predictors.

```python

import statsmodels.api as sm

from statsmodels.stats.outliers_influence import (
    variance_inflation_factor
)

# Add an intercept column to the predictor matrix
X_vif_constant = sm.add_constant(
    X_vif,
    has_constant="add"
)

# Calculate VIF for each predictor
vif_results = []

for i, feature in enumerate(X_vif_constant.columns):

    # Skip the intercept because it is not a predictor
    if feature == "const":
        continue

    # Calculate the VIF value
    vif_value = variance_inflation_factor(
        X_vif_constant.values,
        i
    )

    # Store the result
    vif_results.append({
        "Predictor": feature,
        "VIF": vif_value
    })

# Convert results into a DataFrame
vif_table = pd.DataFrame(vif_results)

# Sort from highest to lowest VIF
vif_table = (
    vif_table
    .sort_values(by="VIF", ascending=False)
    .reset_index(drop=True)
)

# Display the results rounded to four decimal places
vif_table.round(4)
```

##### Understanding the results

**1. Free/reduced lunch eligibility: VIF = 2.4422**

This is the highest VIF in our assessment.

It indicates some overlap between lunch eligibility and the other predictors, which is consistent with the strong negative correlation observed between lunch eligibility and broadband access.

However, the value remains below our threshold of 5.

**2. Broadband access: VIF = 2.3416**

Broadband access also demonstrates some overlap with the other predictors.

Again, its VIF remains below 5.

**3. Per-pupil expenditure: VIF = 1.0584**

This is very close to 1.

It indicates relatively little linear overlap with the other predictors.

**4. Student-teacher ratio: VIF = 1.0507**

This is also very close to 1.

It suggests that student-teacher ratio contributes relatively distinct information compared with the other selected predictors.

**Our finding:** None of the four predictors exceeds our selected VIF threshold.

This means we do not have evidence of severe multicollinearity under this screening criterion.

#### 4.4.2.2 Visualizing the Variance Inflation Factors

To support the numerical assessment, I visualized the VIF values for the four primary predictors.

A horizontal reference line at VIF = 5 was included to distinguish predictors that fall below the selected screening threshold from those requiring further investigation.

The visualization provides a concise comparison of the relative degree of multicollinearity associated with each predictor.

```python

# Create readable predictor labels
vif_plot = vif_table.copy()

vif_plot["Predictor"] = vif_plot["Predictor"].replace({
    "per_pupil_expenditure_usd": "Per-Pupil Expenditure",
    "student_teacher_ratio": "Student-Teacher Ratio",
    "percent_free_reduced_lunch": "Free/Reduced Lunch Eligibility",
    "broadband_access_pct": "Broadband Access"
})

# Create the bar chart
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=vif_plot,
    x="VIF",
    y="Predictor",
    color="steelblue"
)

# Add the screening threshold
plt.axvline(
    x=5,
    color="red",
    linestyle="--",
    linewidth=2,
    label="VIF Threshold = 5"
)

# Add VIF values to the bars
for bar in ax.patches:

    value = bar.get_width()

    plt.text(
        value + 0.08,
        bar.get_y() + bar.get_height() / 2,
        f"{value:.2f}",
        va="center"
    )

# Add chart title and labels
plt.title(
    "Variance Inflation Factor of Primary Predictors",
    fontsize=14
)

plt.xlabel("Variance Inflation Factor (VIF)")
plt.ylabel("Institutional Predictors")

# Set the horizontal-axis range
plt.xlim(0, 6)

# Display the legend
plt.legend()

# Improve chart layout
plt.tight_layout()

# Display the chart
plt.show()
```

The chart clearly shows that all four predictors have VIF values below 5.

Lunch eligibility and broadband access have higher values than expenditure and student-teacher ratio, but their values are not high enough to trigger our selected multicollinearity threshold.

Therefore, we have no statistical justification from this assessment for automatically excluding either predictor.

#### 4.4.2.3 Multicollinearity Findings and Predictor-Retention Decision

The Variance Inflation Factor assessment was conducted to investigate whether the four primary predictors demonstrated potentially concerning multicollinearity.

The analysis produced the following results:

- Free or reduced-price lunch eligibility: VIF = 2.4422.
- Broadband access: VIF = 2.3416.
- Per-pupil expenditure: VIF = 1.0584.
- Student-teacher ratio: VIF = 1.0507.

All four predictors recorded VIF values below the selected threshold of 5.

Free or reduced-price lunch eligibility and broadband access demonstrated the highest values, consistent with their strong negative pairwise correlation identified in the preceding analysis.

However, neither predictor exceeded the multicollinearity screening threshold.

Per-pupil expenditure and student-teacher ratio demonstrated VIF values close to 1, indicating limited linear overlap with the other predictors.

**Finding:** The selected predictors do not demonstrate severe multicollinearity under the adopted VIF screening criterion.

**Analytical Decision:** I retained all four predictors for the subsequent multiple regression analysis.

This decision was based on their relevance to the research question, the absence of concerning VIF values under the selected threshold, and the importance of preserving potentially useful explanatory information.

**Conclusion:** The VIF assessment supports proceeding with the four primary predictors in a multiple regression model.

Although the results reduce concerns about severe multicollinearity, they do not guarantee perfectly stable regression coefficients or eliminate the need for further model diagnostics.

The next stage will investigate the combined associations between the selected predictors and mathematics proficiency.

##### Important note:

**1. What is the difference between correlation and VIF?**

Correlation measures the linear relationship between two variables.

VIF assesses how strongly one predictor can be explained by the other predictors considered together.

**2. Why didn't we remove broadband access despite its strong correlation with lunch eligibility?**

Because its VIF was only 2.3416, below our selected threshold of 5, and it is directly relevant to our research question.

**3. Does VIF tell us whether a predictor is important for predicting mathematics proficiency?**

No. VIF measures multicollinearity, not predictive importance. A predictor can have a low VIF and still have little predictive value.

### 4.4.3 Multiple Linear Regression Analysis

The previous multicollinearity assessment indicated that all four primary predictors could be retained for further analysis under the selected VIF threshold.

I therefore conducted multiple linear regression to investigate their combined associations with mathematics proficiency across the 4,642 K–12 institutions.

The model includes per-pupil expenditure, student-teacher ratio, free or reduced-price lunch eligibility, and broadband access as independent variables, with mathematics proficiency as the dependent variable.

The primary objective is to determine whether the positive relationship between per-pupil expenditure and mathematics proficiency remains after statistically accounting for socioeconomic conditions, staffing, and broadband access.

I also examined the direction, magnitude, and statistical significance of the regression coefficients, together with the model's explanatory power.

This analysis is explanatory rather than a final predictive evaluation. Its findings will inform the subsequent statistical investigation and machine learning stages.

All regression coefficients are interpreted as adjusted associations, not causal effects.


##### Define Our Regression Model

Our multiple linear regression equation is:

**Y = β₀ + β₁X₁ + β₂X₂ + β₃X₃ + β₄X₄ + ε**

##### Definition of Symbols

| Symbol | Meaning |
|:---:|---|
| Y | Mathematics proficiency (%) |
| β₀ | Intercept |
| β₁ | Coefficient for per-pupil expenditure |
| β₂ | Coefficient for student-teacher ratio |
| β₃ | Coefficient for free/reduced lunch eligibility |
| β₄ | Coefficient for broadband access |
| X₁ | Per-pupil expenditure (USD) |
| X₂ | Student-teacher ratio |
| X₃ | Free/reduced lunch eligibility (%) |
| X₄ | Broadband access (%) |
| ε | Unexplained variation (error term) |

##### Interpretation of Regression Coefficients

The regression coefficients (β₁, β₂, β₃, and β₄) describe the estimated association between each predictor and mathematics proficiency, **while holding the other three predictors constant**.

For example, the coefficient β₁ represents the estimated change in mathematics proficiency (percentage points) associated with an additional $1 of expenditure per student, holding the other predictors constant.

Similarly:

- **β₂:** Estimated change in mathematics proficiency associated with a one-unit increase in the student-teacher ratio, holding other predictors constant.

- **β₃:** Estimated change in mathematics proficiency associated with a one-percentage-point increase in free/reduced lunch eligibility, holding other predictors constant.

- **β₄:** Estimated change in mathematics proficiency associated with a one-percentage-point increase in broadband access, holding other predictors constant.

##### Important Statistical Consideration

The regression coefficients represent statistical associations, not necessarily causal relationships.

**The model does not establish that changes in these predictors directly cause changes in mathematics proficiency.**


#### 4.4.3.1 Model Specification

I specified an Ordinary Least Squares (OLS) multiple linear regression model using mathematics proficiency as the dependent variable and the four primary institutional characteristics as independent variables.

The regression model examines the association of each predictor with mathematics proficiency while statistically controlling for the other selected predictors.

I included an intercept and used heteroskedasticity-robust HC3 standard errors for coefficient inference.

The HC3 approach provides more reliable standard errors when the residual variance is not constant across observations, although it does not resolve omitted-variable bias, dependence between observations, or other model limitations.

The model was fitted to the complete analytical dataset for explanatory purposes. Out-of-sample predictive performance will be evaluated separately during the machine learning stage.

#### Fit the multiple linear regression model

```python
# Define the four independent variables
regression_features = [
    "per_pupil_expenditure_usd",
    "student_teacher_ratio",
    "percent_free_reduced_lunch",
    "broadband_access_pct"
]

# Define the dependent variable
regression_target = "math_proficiency_pct"

# Select predictors from our prepared multivariate dataset
X = multivariate_data[regression_features].copy()

# Select the target variable
y = multivariate_data[regression_target].copy()

# Add an intercept to the regression model
X_with_constant = sm.add_constant(
    X,
    has_constant="add"
)

# Fit the Ordinary Least Squares regression model
# HC3 provides heteroskedasticity-robust standard errors
ols_model = sm.OLS(
    y,
    X_with_constant
).fit(cov_type="HC3")

# Display the full regression summary
print(ols_model.summary())
```

These results reveal an important difference between our bivariate and multivariate findings.

Per-pupil expenditure and free/reduced lunch eligibility remain statistically significant.

However, student-teacher ratio and broadband access are not statistically significant at the 5% level after accounting for the other selected predictors.

#### Extract and organize the regression results

```python
# Create a summary table of regression coefficients
regression_results = pd.DataFrame({
    "Coefficient": ols_model.params,
    "Robust_Standard_Error": ols_model.bse,
    "P_Value": ols_model.pvalues,
    "CI_Lower_95": ols_model.conf_int()[0],
    "CI_Upper_95": ols_model.conf_int()[1]
})

# Identify statistically significant predictors
regression_results["Significant_at_5pct"] = (
    regression_results["P_Value"] < 0.05
)

# Display the coefficient summary
regression_results.round(6)
```

#### 4.4.3.2 Model Explanatory Power

I evaluated the explanatory power of the multiple regression model using R-squared and adjusted R-squared.

R-squared measures the proportion of variation in mathematics proficiency accounted for by the fitted linear model.

Adjusted R-squared provides a related measure that accounts for the number of predictors included in the model.

These statistics describe in-sample explanatory fit and should not be interpreted as evidence of out-of-sample predictive accuracy.

A separate machine learning evaluation will be conducted using appropriate training and testing procedures.

```python
# Extract model explanatory statistics
r_squared = ols_model.rsquared
adjusted_r_squared = ols_model.rsquared_adj

# Display the results
print("Multiple Linear Regression: Model Fit")
print("-" * 45)

print(f"Number of Observations: {int(ols_model.nobs)}")
print(f"Number of Predictors: {len(regression_features)}")

print(f"R-squared: {r_squared:.4f}")
print(f"Adjusted R-squared: {adjusted_r_squared:.4f}")

print(
    f"Variation accounted for (R-squared): "
    f"{r_squared * 100:.2f}%"
)

print(
    f"Remaining unexplained variation: "
    f"{(1 - r_squared) * 100:.2f}%"
)
```

Our R-squared is approximately 0.3693.

This means the fitted linear model accounts for approximately **36.93% of the variation in mathematics proficiency within the analysed institutions.**

The remaining 63.07% is not accounted for by the fitted model.

This may reflect other institutional characteristics, unmeasured socioeconomic factors, differences between school levels, nonlinear relationships, and random variation.

The adjusted R-squared is approximately 0.3687, very close to R-squared.

This tells us that the adjustment for the number of predictors has little effect on the overall fit statistic.

However, neither R-squared nor adjusted R-squared tells us how accurately this model will predict mathematics proficiency for previously unseen institutions.

That question belongs to our later machine learning and model evaluation stages.

#### 4.4.3.3 Actual Versus Fitted Mathematics Proficiency

To complement the regression statistics, I compared actual mathematics proficiency with the fitted values generated by the multiple regression model.

The visualization provides an indication of how closely the fitted outcomes correspond to the observed institutional proficiency levels.

A diagonal reference line represents perfect agreement between actual and fitted values.

This assessment describes the model's in-sample fit rather than its predictive performance on unseen data.

```python
# Obtain fitted mathematics proficiency values
fitted_math = ols_model.fittedvalues

# Create the scatterplot
plt.figure(figsize=(9, 7))

sns.scatterplot(
    x=y,
    y=fitted_math,
    alpha=0.35,
    s=25
)

# Determine a common range for both axes
plot_min = min(y.min(), fitted_math.min())
plot_max = max(y.max(), fitted_math.max())

# Add the perfect-agreement reference line
plt.plot(
    [plot_min, plot_max],
    [plot_min, plot_max],
    color="red",
    linestyle="--",
    linewidth=2,
    label="Perfect Agreement"
)

# Add chart title and axis labels
plt.title(
    "Actual vs Fitted Mathematics Proficiency",
    fontsize=14
)

plt.xlabel("Actual Mathematics Proficiency (%)")
plt.ylabel("Fitted Mathematics Proficiency (%)")

# Display the R-squared value on the chart
plt.text(
    0.05,
    0.95,
    f"R-squared = {ols_model.rsquared:.4f}",
    transform=plt.gca().transAxes,
    verticalalignment="top",
    bbox={
        "boxstyle": "round",
        "facecolor": "white",
        "alpha": 0.8
    }
)

# Display the legend
plt.legend()

# Improve chart layout
plt.tight_layout()

# Display the chart
plt.show()
```

### 4.4.4 Regression Diagnostics and Assumption Checks

Following the multiple linear regression analysis, I assessed the model's residuals and selected statistical assumptions to evaluate the reliability of its interpretation.

The regression model accounted for approximately 36.93% of the observed variation in mathematics proficiency, with per-pupil expenditure and free or reduced-price lunch eligibility demonstrating statistically significant adjusted associations.

However, satisfactory explanatory power and statistically significant coefficients do not automatically establish that the regression assumptions are adequately satisfied.

I therefore examined residual patterns, constant variance, approximate normality, potential serial correlation, and influential observations.

The objective is to identify potential model limitations and determine whether the fitted regression model provides a reasonable basis for statistical interpretation.

These diagnostics relate to the explanatory regression model. Predictive performance on unseen data will be assessed separately during the machine learning stage.

#### Calculate the regression residuals

```python
# Extract fitted mathematics proficiency values
fitted_values = ols_model.fittedvalues

# Extract residuals from the regression model
residuals = ols_model.resid

# Create a separate DataFrame for diagnostic analysis
diagnostic_data = pd.DataFrame({
    "Actual_Math_Proficiency": y,
    "Fitted_Math_Proficiency": fitted_values,
    "Residual": residuals
})

# Calculate descriptive statistics for residuals
print("Residual Descriptive Statistics")
print("-" * 45)

print(diagnostic_data["Residual"].describe().round(4))

# Confirm the number of observations
print("\nNumber of Observations:", len(diagnostic_data))

# Display the first five rows
diagnostic_data.head()
```

The residual mean is approximately zero.

This is expected because our Ordinary Least Squares model includes an intercept.

The residual standard deviation is approximately 12.75 percentage points.

This indicates that the differences between actual and fitted mathematics proficiency still show substantial variation.

Notice that some residuals are relatively large. We'll investigate potentially unusual or influential observations later.

Important: A residual mean close to zero does not, by itself, prove that the regression assumptions are satisfied.

#### 4.4.4.1 Linearity Assessment Using Residuals Versus Fitted Values

I examined the residuals against fitted mathematics proficiency values to investigate whether the specified linear regression model adequately represents the observed relationships.

A reasonably specified model should generally produce residuals distributed around zero without a systematic curved pattern.

A visible nonlinear pattern would suggest that the model may not fully capture the relationship between institutional characteristics and mathematics proficiency.

I used a scatterplot with a smoothed trend line to support the visual assessment.

```python
# Create the residuals versus fitted values plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=diagnostic_data,
    x="Fitted_Math_Proficiency",
    y="Residual",
    color="blue",
    alpha=0.7
)

# Add a horizontal reference line at zero
plt.axhline(
    y=0,
    color="red",
    linestyle="--",
    linewidth=2
)

# Add labels and title
plt.title("Residuals vs Fitted Values", fontsize=14)
plt.xlabel("Fitted Mathematics Proficiency (%)", fontsize=12)
plt.ylabel("Residuals", fontsize=12)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

#### Test for constant residual variance

This assumption is called homoscedasticity.

It means that residual variation should remain reasonably consistent across different fitted values.

The opposite is called heteroscedasticity, where residual variation changes systematically.

We can investigate this using the Breusch-Pagan test.
Statistical hypotheses
- H₀: The residual variance is constant.
- H₁: The residual variance is not constant.
We'll use a significance level of 0.05.

If the p-value is below 0.05, we reject the null hypothesis.

If it is above 0.05, we do not have sufficient evidence to reject the assumption of constant variance.

#### 4.4.4.2 Homoscedasticity Assessment Using the Breusch-Pagan Test

I used the Breusch-Pagan test to investigate whether the residual variance changes systematically with the model's predictors.

The null hypothesis states that the residual variance is constant, while the alternative hypothesis states that it is not constant.

The test was evaluated using a significance level of 0.05.

Although the model was fitted using HC3 robust standard errors, this diagnostic remains useful for understanding the residual variance structure.

```python
from statsmodels.stats.diagnostic import het_breuschpagan

# Perform the Breusch-Pagan test
bp_results = het_breuschpagan(
    ols_model.resid,
    ols_model.model.exog
)

# Assign the returned statistics
bp_lm_statistic = bp_results[0]
bp_lm_pvalue = bp_results[1]
bp_f_statistic = bp_results[2]
bp_f_pvalue = bp_results[3]

# Display the results
print("Breusch-Pagan Test")
print("-" * 45)

print(f"LM Statistic: {bp_lm_statistic:.4f}")
print(f"LM P-value: {bp_lm_pvalue:.4f}")
print(f"F Statistic: {bp_f_statistic:.4f}")
print(f"F P-value: {bp_f_pvalue:.4f}")

# Interpret the test using a 5% significance level
alpha = 0.05

if bp_lm_pvalue < alpha:
    print("\nDecision: Reject the null hypothesis.")
    print("Evidence of nonconstant residual variance was detected.")
else:
    print("\nDecision: Fail to reject the null hypothesis.")
    print("No statistically significant evidence of nonconstant residual variance was detected.")
```

##### Breusch-Pagan LM p-value

**0.9691**

*No significant evidence of heteroscedasticity*

Our p-value is approximately 0.9691, considerably larger than 0.05.

Therefore, we fail to reject the null hypothesis.

The test does not provide statistically significant evidence that residual variance changes with the predictors in the form examined.

Conclusion: There is no substantial heteroscedasticity concern indicated by this particular test.

However, failing to reject the null hypothesis does not prove that residual variance is perfectly constant.

#### 4.4.4.3 Residual Normality Assessment

I examined the distribution of regression residuals using a histogram, a normal Q-Q plot, and the Jarque-Bera test.

The histogram provides a visual assessment of the residual distribution, while the Q-Q plot compares the observed residual quantiles with those expected under a normal distribution.

The Jarque-Bera test evaluates departures from normality based on skewness and kurtosis.

The findings will be interpreted carefully because the large sample size may make relatively small departures from normality statistically detectable.

```python
from scipy.stats import jarque_bera

# Create a figure with two diagnostic plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot the residual distribution
sns.histplot(
    residuals,
    bins=35,
    kde=True,
    ax=axes[0],
    color="steelblue"
)

axes[0].set_title("Distribution of Regression Residuals")
axes[0].set_xlabel("Residuals (Percentage Points)")
axes[0].set_ylabel("Number of Institutions")

# Create a normal Q-Q plot
sm.qqplot(
    residuals,
    line="45",
    fit=True,
    ax=axes[1]
)

axes[1].set_title("Normal Q-Q Plot of Residuals")

# Improve chart layout
plt.tight_layout()
plt.show()

# Conduct the Jarque-Bera normality test
jb_statistic, jb_pvalue = jarque_bera(residuals)

# Calculate skewness and excess kurtosis
residual_skewness = residuals.skew()
residual_excess_kurtosis = residuals.kurtosis()

# Display results
print("Residual Normality Assessment")
print("-" * 45)

print(f"Jarque-Bera Statistic: {jb_statistic:.4f}")
print(f"P-value: {jb_pvalue:.6f}")
print(f"Skewness: {residual_skewness:.4f}")
print(f"Excess Kurtosis: {residual_excess_kurtosis:.4f}")

# Interpret the Jarque-Bera test
alpha = 0.05

if jb_pvalue < alpha:
    print("\nDecision: Reject exact residual normality.")
else:
    print("\nDecision: Fail to reject residual normality.")
```

The residual skewness is close to zero, suggesting that the residual distribution is approximately symmetric.

The excess kurtosis is slightly negative, indicating a distribution with somewhat lighter tails than a normal distribution.

However, the Jarque-Bera p-value is approximately 0.0017, below 0.05.

Therefore, we reject the null hypothesis of exact normality.

Does this mean our regression model is invalid?

No.

The residuals show relatively small deviations in skewness and kurtosis, and our sample is large.

We also fitted the regression using HC3 robust standard errors.

The results therefore do not automatically invalidate the coefficient estimates or large-sample inference.

Nevertheless, we should document that the exact normality assumption was not satisfied according to the Jarque-Bera test.

#### 4.4.4.4 Independence Assessment

I calculated the Durbin-Watson statistic to examine potential first-order serial correlation in the regression residuals.

A value near 2 generally indicates little evidence of first-order serial correlation in the observation order used.

However, the dataset contains cross-sectional institutional observations rather than a time series. Consequently, the row order does not necessarily have a meaningful temporal interpretation.

The Durbin-Watson result is therefore treated as a limited diagnostic and not as proof that institutional observations are independent.

Potential dependence among institutions within the same state or geographic area remains a limitation that cannot be resolved by this statistic alone.

```python
from statsmodels.stats.stattools import durbin_watson

# Calculate the Durbin-Watson statistic
dw_statistic = durbin_watson(ols_model.resid)

# Display the result
print("Durbin-Watson Statistic")
print("-" * 45)

print(f"Durbin-Watson: {dw_statistic:.4f}")

# Provide a cautious numerical interpretation
if 1.5 <= dw_statistic <= 2.5:
    print("Result: Little indication of first-order serial correlation in the existing row order.")
else:
    print("Result: Potential serial correlation in the existing row order should be investigated.")

print(
    "\nNote: This is cross-sectional data. "
    "The statistic does not establish independence between institutions."
)
```

##### Durbin-Watson statistic

**1.9946**

Close to 2, but interpretation is limited by the cross-sectional structure of the data.

This indicates little evidence of first-order serial correlation in the existing ordering of the observations.

However, institutions located within the same state or geographic area may share characteristics.

Therefore, we cannot conclude that all observations are statistically independent.

A more advanced investigation could use state-clustered standard errors or hierarchical modelling if the research objective and data structure justify them.

#### 4.4.4.5 Influential Observation Assessment Using Cook's Distance

I examined potentially influential observations using Cook's distance.

Cook's distance provides an indication of how strongly individual observations may influence the fitted regression model.

I used the screening threshold of 4/n, where n represents the number of observations.

Observations exceeding this threshold were flagged for further consideration rather than automatically removed.

This approach is consistent with the earlier data cleaning decision to retain statistically unusual observations when they represent plausible institutional characteristics.

```python
# Obtain influence statistics from the regression model
influence = ols_model.get_influence()

# Calculate Cook's distance for every observation
cooks_distance = influence.cooks_distance[0]

# Calculate the commonly used screening threshold
n_observations = int(ols_model.nobs)
cooks_threshold = 4 / n_observations

# Identify observations exceeding the threshold
influential_mask = cooks_distance > cooks_threshold

# Calculate summary statistics
number_influential = int(influential_mask.sum())
maximum_cooks_distance = float(cooks_distance.max())

# Display results
print("Cook's Distance Assessment")
print("-" * 45)

print(f"Number of Observations: {n_observations}")
print(f"Screening Threshold (4/n): {cooks_threshold:.6f}")
print(f"Observations Above Threshold: {number_influential}")
print(f"Maximum Cook's Distance: {maximum_cooks_distance:.6f}")

# Create a diagnostic plot
plt.figure(figsize=(12, 5))

plt.scatter(
    range(n_observations),
    cooks_distance,
    alpha=0.5,
    s=12,
    color="steelblue"
)

# Add the screening threshold
plt.axhline(
    y=cooks_threshold,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Cook's Distance Threshold (4/n)"
)

# Add chart title and axis labels
plt.title(
    "Cook's Distance for Regression Observations",
    fontsize=14
)

plt.xlabel("Observation Position")
plt.ylabel("Cook's Distance")

# Display legend
plt.legend()

# Improve chart layout
plt.tight_layout()

# Display the chart
plt.show()
```

We identified 241 observations with Cook's distance values above the screening threshold.

However, the threshold is quite small because our sample size is large.

The maximum Cook's distance is approximately 0.0046, substantially below the commonly used reference value of 1.

This suggests that no single observation demonstrates an exceptionally large Cook's distance under that reference guideline.

Nevertheless, the 241 flagged observations deserve consideration because several observations can collectively affect model estimates.

Our analytical decision: We will not automatically remove these observations.

There is no evidence from Cook's distance alone that the records are invalid.

Removing them solely to improve regression results would risk introducing bias and weakening the reproducibility of our analysis.

A sensitivity analysis could be considered later if influential observations materially affect substantive conclusions.

## 5. Statistical Analysis

Statistical analysis provides a formal framework for evaluating relationships identified during exploratory data analysis.

The preceding investigations revealed associations between institutional characteristics and mathematics proficiency. The multiple regression analysis further indicated that per-pupil expenditure and free or reduced-price lunch eligibility retained statistically significant adjusted associations.

This section focuses on formally evaluating the relationship between institutional expenditure and mathematics proficiency.

The objective is to determine whether the expenditure coefficient differs significantly from zero after accounting for socioeconomic conditions, staffing, and broadband access.

The investigation will use hypothesis testing, a 95% confidence interval, and an assessment of practical significance.

The statistical analysis will rely on the previously fitted multiple regression model and its HC3 robust standard errors, avoiding unnecessary changes to the analytical population or model specification.

### 5.1 Statistical Research Question and Hypothesis Formulation

#### Research Question

Does per-pupil expenditure demonstrate a statistically significant positive association with mathematics proficiency after accounting for free or reduced-price lunch eligibility, student-teacher ratio, and broadband access?

#### Statistical Objective

To evaluate whether the adjusted regression coefficient for per-pupil expenditure is statistically different from zero at the 5% significance level.

#### Variables

**Dependent Variable:**
- Mathematics proficiency percentage (`math_proficiency_pct`).

**Primary Independent Variable:**
- Per-pupil expenditure in US dollars (`per_pupil_expenditure_usd`).

**Control Variables:**
- Free or reduced-price lunch eligibility percentage (`percent_free_reduced_lunch`).
- Student-teacher ratio (`student_teacher_ratio`).
- Broadband access percentage (`broadband_access_pct`).

The statistical investigation uses the same 4,642 K–12 institutions selected during the data preparation and exploratory analysis stages.

#### Formulate Our Statistical Hypotheses

We will test whether the expenditure coefficient is statistically different from zero after adjustment.

Let **β₁** represent the population regression coefficient for per-pupil expenditure.

Our hypotheses are:

##### Null Hypothesis (H₀)

**H₀: β₁ = 0**

There is no adjusted linear association between per-pupil expenditure and mathematics proficiency in the specified model.

##### Alternative Hypothesis (H₁)

**H₁: β₁ ≠ 0**

There is a nonzero adjusted linear association between per-pupil expenditure and mathematics proficiency in the specified model.

##### Why Are We Using a Two-Sided Hypothesis?

Although our research expectation is that expenditure is positively associated with mathematics proficiency, a two-sided test allows us to detect statistical evidence of either a **positive or negative adjusted association**.

We then interpret the sign of the estimated regression coefficient (β₁) to determine the direction of the association.



### 5.1.1 Statistical Hypotheses

The statistical investigation evaluates the adjusted regression coefficient of per-pupil expenditure in the previously fitted multiple linear regression model.

**Null Hypothesis (H₀):**

The population regression coefficient for per-pupil expenditure equals zero after accounting for the other selected institutional characteristics.

**Alternative Hypothesis (H₁):**

The population regression coefficient for per-pupil expenditure differs from zero after accounting for the other selected institutional characteristics.

A two-sided hypothesis test was selected because the objective is to determine whether the adjusted association is statistically different from zero, regardless of its direction.

The direction of the estimated coefficient will subsequently be used to determine whether the observed relationship is positive or negative.

**Significance Level:** 0.05 (5%).

The test will be evaluated using HC3 heteroskedasticity-robust inference from the previously fitted multiple regression model.

#### 5.1.2 Selection and Justification of Statistical Method

I selected a two-sided Wald test of the per-pupil expenditure coefficient from the previously fitted multiple linear regression model.

This method is appropriate because the research question concerns the adjusted association between a continuous institutional predictor and a continuous academic outcome.

Unlike a simple correlation test, the multiple regression framework accounts for the other selected institutional characteristics when estimating the expenditure coefficient.

I retained HC3 heteroskedasticity-robust standard errors to support large-sample statistical inference.

The test will evaluate the null hypothesis that the adjusted expenditure coefficient equals zero.

A 95% confidence interval will also be reported to assess the range of plausible coefficient values.

Finally, I will translate the estimated coefficient into percentage points of mathematics proficiency per additional $1,000 of per-pupil expenditure to assess practical significance.

The statistical conclusions will be interpreted as adjusted associations rather than causal effects.

#### Establish the test configuration in Python

We already fitted the model as ols_model.

For now, we'll confirm that the model contains our expenditure variable and define the settings we'll use

```python
# Define the predictor being tested
hypothesis_predictor = "per_pupil_expenditure_usd"

# Set the statistical significance level
alpha = 0.05

# Define the expenditure increment for practical interpretation
expenditure_increment = 1000

# Confirm that the regression model is available
assert "ols_model" in globals(), (
    "The regression model has not been fitted. "
    "Run the multiple regression code in Section 4.4.3 first."
)

# Confirm that our selected predictor exists in the model
assert hypothesis_predictor in ols_model.params.index, (
    "Per-pupil expenditure is missing from the regression model."
)

# Confirm that the model uses HC3 robust standard errors
assert ols_model.cov_type == "HC3", (
    "The regression model must use HC3 robust standard errors."
)

# Display the statistical test configuration
print("Statistical Investigation Setup")
print("-" * 50)

print("Research Focus: Expenditure and Mathematics Proficiency")
print(f"Primary Predictor: {hypothesis_predictor}")
print("Null Hypothesis: Expenditure coefficient = 0")
print("Alternative Hypothesis: Expenditure coefficient != 0")
print(f"Significance Level: {alpha}")
print(f"Confidence Level: {(1 - alpha) * 100:.0f}%")
print("Test Method: Two-sided HC3 Robust Wald Test")
print(f"Number of Observations: {int(ols_model.nobs)}")
print(f"Practical Interpretation: Per ${expenditure_increment:,}")
```

#### 5.1.3 Significance Level and Decision Criteria

The hypothesis test will be evaluated using a significance level of 0.05.

The decision criteria are:

- If the p-value is below 0.05, reject the null hypothesis and conclude that the adjusted expenditure coefficient is statistically different from zero.
- If the p-value is 0.05 or greater, fail to reject the null hypothesis because the available evidence is insufficient to establish a statistically significant adjusted association.

The 95% confidence interval will be examined alongside the hypothesis test.

If the confidence interval excludes zero, it will support a statistically significant nonzero coefficient at the corresponding two-sided 5% significance level.

Statistical significance will be distinguished from practical significance by interpreting the estimated association for a $1,000 difference in per-pupil expenditure.

A statistically significant coefficient does not establish that increasing institutional expenditure would cause an equivalent improvement in mathematics proficiency.

The results will therefore be interpreted within the limitations of the observational dataset and the specified multiple regression model.

### 5.2 Hypothesis Testing

Following the hypothesis formulation, I conducted a formal statistical test to determine whether per-pupil expenditure retained a statistically significant adjusted association with mathematics proficiency.

The investigation used the coefficient estimate from the previously fitted Ordinary Least Squares multiple regression model with HC3 heteroskedasticity-robust standard errors.

The null hypothesis states that the population regression coefficient for per-pupil expenditure equals zero, while the alternative hypothesis states that the coefficient differs from zero.

I selected a two-sided robust Wald test and a significance level of 0.05.

The test evaluates the expenditure coefficient while accounting for student-teacher ratio, free or reduced-price lunch eligibility, and broadband access.

The analysis uses the same 4,642 K–12 institutions and regression specification established in the multivariate investigation, ensuring consistency across the project.

#### Conduct the robust Wald test

```python
# ---------------------------------------------------------
# 5.2.1 Conduct the Robust Wald Hypothesis Test
# ---------------------------------------------------------

# Define the null hypothesis
# H0: The expenditure regression coefficient equals zero
null_hypothesis = "per_pupil_expenditure_usd = 0"

# Conduct the two-sided robust Wald test
# This uses the HC3 covariance matrix of our fitted OLS model
wald_result = ols_model.wald_test(
    null_hypothesis,
    scalar=True
)

# Extract the test statistic and p-value
wald_statistic = float(wald_result.statistic)
wald_pvalue = float(wald_result.pvalue)

# Extract the expenditure coefficient and robust standard error
expenditure_coefficient = float(
    ols_model.params["per_pupil_expenditure_usd"]
)

expenditure_robust_se = float(
    ols_model.bse["per_pupil_expenditure_usd"]
)

# Calculate the coefficient's robust z-statistic
z_statistic = (
    expenditure_coefficient / expenditure_robust_se
)

# Display the results
print("Robust Wald Hypothesis Test")
print("-" * 50)

print(f"Null Hypothesis: {null_hypothesis}")
print(f"Significance Level: {alpha}")

print(f"\nExpenditure Coefficient: {expenditure_coefficient:.8f}")
print(f"HC3 Robust Standard Error: {expenditure_robust_se:.8f}")

print(f"\nRobust z-statistic: {z_statistic:.4f}")
print(f"Wald Chi-Square Statistic: {wald_statistic:.4f}")
print(f"P-value: {wald_pvalue:.6e}")
print("Degrees of Freedom: 1")
```

**1. Expenditure coefficient = +0.00069026**

The positive sign indicates that higher per-pupil expenditure is associated with higher mathematics proficiency after accounting for the other selected predictors.

**2. Robust z-statistic = 13.0331**

This measures how far the estimated expenditure coefficient is from zero, relative to its robust standard error.

The large positive value provides evidence against the null hypothesis.

**3. Wald chi-square statistic = 169.8616**

Because we're testing one coefficient, this is approximately the square of the robust z-statistic.

**4. P-value = 7.93 × 10⁻³⁹**

This is extremely small and substantially below 0.05.

Therefore, we reject the null hypothesis.

Our conclusion is that the adjusted expenditure coefficient is statistically different from zero.

#### 5.2.1 Statistical Decision

The robust Wald test was evaluated at a 5% significance level.

The calculated test statistic was approximately 169.8616, with a p-value of approximately 7.93 × 10⁻³⁹.

Since the p-value is substantially below 0.05, the null hypothesis is rejected.

This provides statistically significant evidence that the adjusted regression coefficient for per-pupil expenditure differs from zero.

The estimated coefficient is positive, indicating that higher per-pupil expenditure is associated with higher mathematics proficiency after accounting for the other selected predictors.

This statistical conclusion does not establish a causal effect of expenditure on academic performance.

#### Apply the statistical decision rule

```python

# ---------------------------------------------------------
# 5.2.2 Apply the Statistical Decision Rule
# ---------------------------------------------------------

# Set the significance level
alpha = 0.05

# Compare the Wald test p-value with the significance level
if wald_pvalue < alpha:

    statistical_decision = "Reject the null hypothesis"

    statistical_conclusion = (
        "There is statistically significant evidence "
        "of a nonzero adjusted association between "
        "per-pupil expenditure and mathematics proficiency."
    )

else:

    statistical_decision = "Fail to reject the null hypothesis"

    statistical_conclusion = (
        "There is insufficient statistical evidence "
        "of a nonzero adjusted association between "
        "per-pupil expenditure and mathematics proficiency."
    )

# Display the statistical decision
print("Hypothesis Testing Decision")
print("-" * 50)

print(f"Significance Level: {alpha}")
print(f"P-value: {wald_pvalue:.6e}")

print(f"\nDecision: {statistical_decision}")
print(f"\nConclusion: {statistical_conclusion}")
```

#### 5.2.3 Hypothesis Testing Findings and Conclusion

The statistical investigation evaluated whether per-pupil expenditure retained a statistically significant association with mathematics proficiency after accounting for student-teacher ratio, free or reduced-price lunch eligibility, and broadband access.

A two-sided robust Wald test was conducted using the previously fitted multiple linear regression model with HC3 robust standard errors.

The test produced a Wald chi-square statistic of approximately 169.8616 and a p-value of approximately 7.93 × 10⁻³⁹.

Because the p-value was below the selected significance level of 0.05, I rejected the null hypothesis that the adjusted expenditure coefficient equals zero.

The estimated expenditure coefficient was positive, approximately 0.00069026.

**Finding:** Per-pupil expenditure retains a positive and statistically significant adjusted association with mathematics proficiency after accounting for the other three institutional characteristics.

**Answer to the Statistical Research Question:** The results provide statistically significant evidence that the relationship between per-pupil expenditure and mathematics proficiency remains positive after adjustment.

**Analytical Implication:** Institutional expenditure provides statistically significant additional explanatory information about mathematics proficiency within the selected regression model.

However, statistical significance does not establish practical importance or causation.

The investigation uses observational institutional data, and the statistical test depends on the model specification and assumptions underlying the inference.

Furthermore, the current Wald test formalizes the coefficient evidence already identified in the multiple regression analysis. It should not be interpreted as an independent replication of that finding.

**Conclusion:** The hypothesis test supports retaining the positive expenditure association as a key statistical finding.

The next stage will examine the 95% confidence interval to evaluate the uncertainty surrounding the estimated coefficient before assessing its practical importance.

##### Note to note:

**1. Why did we reject the null hypothesis?**

Because the p-value was much smaller than our significance level of 0.05.

**2. Does the tiny p-value mean the relationship is very strong?**

No. The p-value measures evidence against the null hypothesis under the statistical model. It does not measure the strength or practical importance of the relationship.

**3. Why did we use the previous regression model rather than fitting another one?**

Because our hypothesis specifically concerns the adjusted expenditure coefficient from that model. Reusing it ensures consistency with our earlier analysis.

### 5.3 Confidence Interval Estimation

Following the hypothesis test, I estimated a 95% confidence interval for the adjusted regression coefficient of per-pupil expenditure.

The objective is to quantify the uncertainty surrounding the estimated association between institutional expenditure and mathematics proficiency.

The interval was calculated using the previously fitted multiple linear regression model with HC3 heteroskedasticity-robust standard errors.

This approach maintains consistency with the hypothesis test conducted in Section 5.2.

The confidence interval will be interpreted alongside the estimated coefficient and statistical significance results.

For practical interpretation, the coefficient and confidence interval will also be expressed as mathematics proficiency percentage points associated with a $1,000 difference in per-pupil expenditure.

The findings will be interpreted as adjusted statistical associations rather than causal effects.

#### Calculate the 95% confidence interval

```python
# ---------------------------------------------------------
# 5.3.1 Calculate the 95% Confidence Interval
# ---------------------------------------------------------

# Define the predictor of interest
expenditure_variable = "per_pupil_expenditure_usd"

# Set the significance level
alpha = 0.05

# Extract the expenditure coefficient
expenditure_estimate = float(
    ols_model.params[expenditure_variable]
)

# Extract the HC3 robust standard error
expenditure_se = float(
    ols_model.bse[expenditure_variable]
)

# Calculate the 95% confidence interval
confidence_intervals = ols_model.conf_int(alpha=alpha)

# Extract the lower and upper confidence limits
ci_lower = float(
    confidence_intervals.loc[expenditure_variable, 0]
)

ci_upper = float(
    confidence_intervals.loc[expenditure_variable, 1]
)

# Display the results
print("95% Confidence Interval: Per-Pupil Expenditure")
print("-" * 55)

print(f"Coefficient Estimate: {expenditure_estimate:.8f}")
print(f"HC3 Robust Standard Error: {expenditure_se:.8f}")

print(f"\nConfidence Level: {(1 - alpha) * 100:.0f}%")
print(f"Lower Confidence Limit: {ci_lower:.8f}")
print(f"Upper Confidence Limit: {ci_upper:.8f}")

print(
    f"\n95% Confidence Interval: "
    f"({ci_lower:.8f}, {ci_upper:.8f})"
)
```

The estimated expenditure coefficient is approximately +0.00069026.

Our 95% confidence interval extends from approximately +0.000586 to +0.000794.

Both limits are positive.

This indicates that the range of coefficient values supported by our statistical model is entirely positive.

The interval excludes zero, which is consistent with our earlier decision to reject the null hypothesis.

Important: The interval reflects statistical uncertainty under our selected model and HC3 inference assumptions. It does not account for every possible source of bias, such as omitted variables or geographic dependence.

#### 5.3.2 Practical Interpretation of the Confidence Interval

To make the expenditure coefficient easier to interpret, I converted the estimated association and its 95% confidence interval from a $1 increase to a $1,000 difference in per-pupil expenditure.

This transformation does not change the underlying statistical relationship.

It expresses the same coefficient and uncertainty interval using a more meaningful financial scale.

The resulting values describe the estimated difference in mathematics proficiency percentage points associated with a $1,000 difference in expenditure, holding the other institutional predictors constant.

```python
# ---------------------------------------------------------
# 5.3.2 Convert the Confidence Interval to a $1,000 Scale
# ---------------------------------------------------------

# Define the expenditure increment
expenditure_increment = 1000

# Convert the coefficient to a $1,000 expenditure scale
effect_per_1000 = (
    expenditure_estimate * expenditure_increment
)

# Convert the confidence interval to the same scale
ci_lower_1000 = ci_lower * expenditure_increment
ci_upper_1000 = ci_upper * expenditure_increment

# Display the results
print("Adjusted Expenditure Association per $1,000")
print("-" * 55)

print(
    f"Estimated Association: "
    f"{effect_per_1000:.4f} percentage points"
)

print(
    f"95% Confidence Interval: "
    f"({ci_lower_1000:.4f}, {ci_upper_1000:.4f}) "
    f"percentage points"
)

# Create a summary table
expenditure_ci_summary = pd.DataFrame({
    "Measurement": [
        "Estimated Association",
        "Lower 95% Confidence Limit",
        "Upper 95% Confidence Limit"
    ],
    "Percentage_Points_Per_1000_USD": [
        effect_per_1000,
        ci_lower_1000,
        ci_upper_1000
    ]
})

# Display the summary table
expenditure_ci_summary.round(4)
```

The estimated coefficient indicates that a $1,000 difference in per-pupil expenditure is associated with approximately 0.69 percentage points higher mathematics proficiency, holding student-teacher ratio, free or reduced-price lunch eligibility, and broadband access constant.

The 95% confidence interval ranges from approximately 0.59 to 0.79 percentage points.

Our study uses observational data, so the result describes an adjusted association rather than a causal effect.

### 5.4 Statistical and Practical Significance

The hypothesis test and confidence interval analysis established a statistically significant positive adjusted association between per-pupil expenditure and mathematics proficiency.

However, statistical significance alone does not determine whether the estimated association is substantial enough to inform educational decision-making.

I therefore assessed the practical magnitude of the expenditure coefficient using interpretable financial increments and the observed variability in mathematics proficiency.

The analysis distinguishes statistical evidence from practical importance and considers the limitations of translating observational associations into education policy recommendations.

The objective is to determine how the estimated relationship can be communicated responsibly to educational stakeholders without interpreting it as a causal effect.

#### Evaluate statistical significance

```python
# ---------------------------------------------------------
# 5.4.1 Statistical Significance Summary
# ---------------------------------------------------------

# Extract the expenditure coefficient and p-value
expenditure_coefficient = float(
    ols_model.params["per_pupil_expenditure_usd"]
)

expenditure_pvalue = float(
    ols_model.pvalues["per_pupil_expenditure_usd"]
)

# Define the significance level
alpha = 0.05

# Determine statistical significance
is_significant = expenditure_pvalue < alpha

# Display the results
print("Statistical Significance Assessment")
print("-" * 50)

print(f"Expenditure Coefficient: {expenditure_coefficient:.8f}")
print(f"P-value: {expenditure_pvalue:.6e}")
print(f"Significance Level: {alpha}")

print(
    "Statistically Significant:",
    "Yes" if is_significant else "No"
)

print(
    "Direction:",
    "Positive" if expenditure_coefficient > 0
    else "Negative" if expenditure_coefficient < 0
    else "Zero"
)
```

The p-value is considerably below 0.05.

Therefore, the evidence supports a nonzero adjusted association between expenditure and mathematics proficiency in the specified model.

However, the p-value does not tell us whether the relationship is educationally large, economically worthwhile, or causal.

That's why our next step is necessary.

#### 5.4.1 Practical Magnitude of the Expenditure Association

To evaluate the practical magnitude of the expenditure relationship, I expressed the adjusted regression coefficient across several per-pupil expenditure differences.

The analysis considers expenditure differences of $500, $1,000, $2,500, and $5,000 per student.

For each increment, I calculated the corresponding difference in fitted mathematics proficiency and its 95% confidence interval.

These calculations illustrate the magnitude implied by the estimated linear relationship within the model.

They do not represent predicted improvements resulting from additional funding or establish that increased expenditure would cause the calculated differences in academic performance.

```python

# ---------------------------------------------------------
# 5.4.2 Estimate Associations at Different Expenditure Scales
# ---------------------------------------------------------

# Define illustrative expenditure differences in USD per student
expenditure_differences = [500, 1000, 2500, 5000]

# Extract the expenditure coefficient
coefficient = float(
    ols_model.params["per_pupil_expenditure_usd"]
)

# Extract the 95% confidence interval
confidence_interval = ols_model.conf_int(alpha=0.05).loc[
    "per_pupil_expenditure_usd"
]

lower_coefficient = float(confidence_interval.iloc[0])
upper_coefficient = float(confidence_interval.iloc[1])

# Create a list for storing results
practical_results = []

# Calculate the adjusted association for each expenditure difference
for amount in expenditure_differences:

    # Calculate the estimated difference in mathematics proficiency
    estimated_difference = coefficient * amount

    # Calculate the confidence interval at the same scale
    lower_difference = lower_coefficient * amount
    upper_difference = upper_coefficient * amount

    # Store the results
    practical_results.append({
        "Expenditure_Difference_USD": amount,
        "Estimated_Difference_pp": estimated_difference,
        "CI_Lower_pp": lower_difference,
        "CI_Upper_pp": upper_difference
    })

# Convert results into a DataFrame
practical_significance_table = pd.DataFrame(
    practical_results
)

# Display the results
practical_significance_table.round(3)
```

**What does this tell us?**

The estimated association is relatively modest at smaller expenditure differences.

For instance, a $1,000 difference in per-pupil expenditure corresponds to approximately 0.69 percentage points of mathematics proficiency.

A $5,000 expenditure difference corresponds to approximately 3.45 percentage points under the same linear relationship.

These differences may be relevant when comparing institutions, but whether they are educationally or economically meaningful depends on the stakeholder's objectives, available resources, and alternative uses of funding.

We cannot conclude from this evidence that spending an additional $5,000 per student would increase mathematics proficiency by 3.45 percentage points.

#### 5.4.3 Contextualizing the Association Against Mathematics Proficiency Variability

I compared the estimated expenditure association with the observed standard deviation of mathematics proficiency across the analytical population.

This provides additional context for evaluating the relative magnitude of the adjusted relationship.

The comparison expresses the estimated mathematics proficiency difference associated with a $1,000 expenditure difference as a proportion of one observed standard deviation.

This is a descriptive scaling measure, not a causal effect size or a complete assessment of educational importance.

```python

# ---------------------------------------------------------
# 5.4.3 Contextualize the Association
# ---------------------------------------------------------

# Calculate descriptive statistics for mathematics proficiency
mean_math = multivariate_data["math_proficiency_pct"].mean()
std_math = multivariate_data["math_proficiency_pct"].std()

# Calculate the estimated association for a $1,000 difference
difference_per_1000 = expenditure_coefficient * 1000

# Express the association relative to one standard deviation
standardized_magnitude = difference_per_1000 / std_math

# Display the results
print("Practical Magnitude Assessment")
print("-" * 50)

print(f"Mean Mathematics Proficiency: {mean_math:.2f}%")
print(f"Standard Deviation: {std_math:.2f} percentage points")

print(
    f"\nEstimated Difference per $1,000: "
    f"{difference_per_1000:.3f} percentage points"
)

print(
    f"Relative to One Standard Deviation: "
    f"{standardized_magnitude:.4f}"
)

print(
    f"Percentage of One Standard Deviation: "
    f"{standardized_magnitude * 100:.2f}%"
)
```

A $1,000 expenditure difference is associated with approximately 0.69 percentage points of mathematics proficiency under our model.

That provides useful context: the estimated difference is small relative to the overall variation in mathematics proficiency across institutions.

However, we cannot conclude that the relationship is unimportant simply because it is small on this scale.

A relatively small average difference may still matter in some educational contexts, particularly when considering large student populations.

Determining its actual policy value would require additional evidence about funding interventions, costs, implementation, and educational outcomes.

### 5.5 Statistical Analysis Summary and Conclusion

The statistical investigation examined whether per-pupil expenditure retained a statistically significant positive association with mathematics proficiency after accounting for socioeconomic conditions, student-teacher ratio, and broadband access.

The analysis used the previously fitted Ordinary Least Squares multiple regression model with HC3 heteroskedasticity-robust standard errors.

A two-sided robust Wald test was conducted at a 5% significance level, followed by a 95% confidence interval and an assessment of practical significance.

The investigation focused on three questions:

1. Is the adjusted expenditure coefficient statistically different from zero?
2. What range of coefficient values is supported by the confidence interval?
3. How meaningful is the estimated association when expressed in practical expenditure increments?

The findings are consolidated below to provide a clear statistical conclusion and establish the foundation for the subsequent machine learning investigation.

#### Create a consolidated statistical results table

```python

# ---------------------------------------------------------
# 5.5.1 Consolidated Statistical Analysis Results
# ---------------------------------------------------------

# Define the predictor of interest
expenditure_variable = "per_pupil_expenditure_usd"

# Define the statistical settings
alpha = 0.05
expenditure_increment = 1000

# Extract the expenditure coefficient
coefficient = float(
    ols_model.params[expenditure_variable]
)

# Extract its HC3 robust standard error and p-value
robust_se = float(
    ols_model.bse[expenditure_variable]
)

p_value = float(
    ols_model.pvalues[expenditure_variable]
)

# Calculate the two-sided robust Wald test
wald_result = ols_model.wald_test(
    f"{expenditure_variable} = 0",
    scalar=True
)

wald_statistic = float(wald_result.statistic)

# Extract the 95% confidence interval
confidence_interval = ols_model.conf_int(
    alpha=alpha
).loc[expenditure_variable]

ci_lower = float(confidence_interval.iloc[0])
ci_upper = float(confidence_interval.iloc[1])

# Express the coefficient and interval per $1,000
effect_per_1000 = coefficient * expenditure_increment
lower_per_1000 = ci_lower * expenditure_increment
upper_per_1000 = ci_upper * expenditure_increment

# Calculate the relative magnitude
math_std = multivariate_data["math_proficiency_pct"].std()

relative_magnitude = (
    effect_per_1000 / math_std
)

# Record the statistical decision
decision = (
    "Reject H0"
    if p_value < alpha
    else "Fail to reject H0"
)

# Construct a consolidated summary table
statistical_summary = pd.DataFrame({
    "Metric": [
        "Number of Institutions",
        "Significance Level",
        "Adjusted Expenditure Coefficient",
        "HC3 Robust Standard Error",
        "Wald Chi-Square Statistic",
        "P-value",
        "Statistical Decision",
        "Association per $1,000 (pp)",
        "95% CI Lower per $1,000 (pp)",
        "95% CI Upper per $1,000 (pp)",
        "Magnitude Relative to Outcome SD (%)"
    ],
    "Value": [
        str(int(ols_model.nobs)),
        f"{alpha:.2f}",
        f"{coefficient:.8f}",
        f"{robust_se:.8f}",
        f"{wald_statistic:.4f}",
        f"{p_value:.6e}",
        decision,
        f"{effect_per_1000:.4f}",
        f"{lower_per_1000:.4f}",
        f"{upper_per_1000:.4f}",
        f"{relative_magnitude * 100:.2f}"
    ]
})

# Display the consolidated results
statistical_summary
```

#### 5.5.2 Key Statistical Findings

The statistical analysis produced the following findings:

**1. Statistically Significant Expenditure Association**

The robust Wald test produced a chi-square statistic of approximately 169.86 and a p-value substantially below 0.05.

I therefore rejected the null hypothesis that the adjusted expenditure coefficient equals zero.

This provides statistically significant evidence of an association between per-pupil expenditure and mathematics proficiency after accounting for the other selected institutional characteristics.

**2. Positive Adjusted Relationship**

The estimated regression coefficient was approximately +0.00069026.

This indicates that higher per-pupil expenditure is associated with higher mathematics proficiency when student-teacher ratio, free or reduced-price lunch eligibility, and broadband access are held constant in the fitted model.

**3. Confidence Interval Supports the Positive Association**

The 95% confidence interval for the expenditure coefficient was entirely positive.

Expressed per $1,000 of expenditure, the estimated association was approximately +0.69 percentage points, with a 95% confidence interval of approximately +0.59 to +0.79 percentage points.

This finding is consistent with the hypothesis test.

**4. Relatively Modest Practical Magnitude**

The estimated association for a $1,000 expenditure difference represented approximately 4.30% of one observed standard deviation in mathematics proficiency.

This suggests that the estimated difference is relatively modest compared with the overall variation in academic outcomes.

Statistical significance therefore should not be confused with educational or economic importance.

**5. Implications for Educational Decision-Making**

The findings suggest that institutional expenditure provides useful explanatory information about mathematics proficiency beyond the other selected predictors.

However, the observational analysis does not establish that increasing funding would produce the estimated improvement in academic outcomes.

The evidence may inform further investigation of educational resources and institutional performance, but it does not establish the effectiveness or cost-effectiveness of specific funding interventions.

#### 5.5.3 Statistical Limitations

The findings should be interpreted within the following limitations:

**Observational Data:** The dataset describes institutional characteristics and academic outcomes. The analysis establishes statistical associations rather than causal effects.

**Potential Omitted Variables:** The regression model includes four selected predictors. Other factors, such as geographic conditions, institutional characteristics, educational programmes, and unmeasured student circumstances, may also be associated with mathematics proficiency.

**Model Specification:** The statistical investigation assumes a linear relationship between the selected predictors and the conditional mean of mathematics proficiency. More complex relationships may not be fully represented.

**Statistical Assumptions:** Previous regression diagnostics identified a departure from exact residual normality. Although HC3 robust standard errors were used, they do not eliminate every possible source of statistical uncertainty.

**Potential Geographic Dependence:** Institutions within the same state or geographic area may share characteristics. This potential dependence was not fully addressed by the HC3 covariance estimator.

**Practical Significance:** The estimated expenditure association is relatively modest per $1,000 and cannot, by itself, establish whether additional spending would be economically worthwhile.

**Generalizability:** The findings describe the selected analytical dataset and depend on its representativeness and measurement quality.

These limitations will be considered when interpreting the predictive models and developing evidence-based recommendations.

#### 5.5.4 Overall Statistical Conclusion

The statistical investigation evaluated whether per-pupil expenditure retained a statistically significant association with mathematics proficiency after accounting for socioeconomic conditions, student-teacher ratio, and broadband access.

#### Confirm Our Target and Predictors

#### Variable Definitions

| Role | Variable | Meaning |
|---|---|---|
| Target (Y) | `math_proficiency_pct` | Mathematics proficiency percentage |

**Why are we retaining all four predictors?**

A predictor that is not statistically significant in a linear regression model may still contribute useful information to a different predictive model.

Additionally, all four variables are central to our original research question.

Therefore, we'll retain them initially and evaluate their predictive usefulness through model comparison.m

#### 6.1.1 Target and Predictor Selection Rationale

Mathematics proficiency percentage was selected as the prediction target because it directly addresses the primary research question and represents a numerical academic outcome.

The four predictors were selected based on their relevance to institutional funding, staffing, socioeconomic conditions, and technology access.

These characteristics were investigated during exploratory data analysis and the statistical analysis stages.

Although student-teacher ratio and broadband access did not demonstrate statistically significant adjusted linear associations in the explanatory regression model, they were retained for the predictive investigation.

This decision recognizes that statistical significance in an explanatory model does not necessarily determine a variable's usefulness for prediction.

Retaining the four predictors also ensures that the machine learning investigation remains consistent with the original research question.

Additional academic outcome variables were not included as predictors because the objective is to estimate mathematics proficiency from institutional characteristics rather than from other recorded academic outcomes.

#### 6.1.2 Selection of Machine Learning Models

Three regression approaches were selected for the predictive investigation.

**Baseline Model**

A Dummy Regressor using the mean of the training target will serve as the baseline.

This provides a reference point for determining whether more advanced models offer meaningful improvements in predictive performance.

**Linear Regression**

Linear Regression was selected because it provides an interpretable approach to predicting mathematics proficiency using the four institutional predictors.

It also provides continuity with the earlier multiple linear regression investigation.

However, the machine learning implementation will be evaluated using a separate test dataset rather than relying on in-sample explanatory statistics.

**Random Forest Regressor**

Random Forest Regressor was selected as a nonlinear comparison model.

Unlike Linear Regression, Random Forest can capture more complex relationships and interactions between predictors without requiring an explicitly specified linear equation.

Comparing these models will help determine whether a more flexible predictive approach improves performance.

The final model will be selected based on predictive evaluation results rather than model complexity alone.

#### 6.1.3 Training, Validation, and Testing Strategy

The dataset will be divided into training and testing subsets using an 80:20 split.

The training subset will contain approximately 80% of the institutional observations and will be used to fit the machine learning models.

The remaining 20% will be reserved for evaluating predictive performance on institutions not used during training.

A fixed random state of 42 will be used to ensure reproducibility.

Model development and any necessary tuning will be conducted without using the held-out test targets.

Cross-validation will be performed using the training data to assess model stability and support model selection.

The final test set will be used to evaluate how well the selected models generalize to unseen institutional observations within the supplied dataset.

Because the data are cross-sectional, this evaluation does not establish predictive reliability across future academic years or entirely different geographic populations.

#### 6.1.4 Model Evaluation Criteria

The predictive models will be evaluated using three regression metrics.

**Mean Absolute Error (MAE)**

MAE measures the average absolute difference between actual and predicted mathematics proficiency.

It will serve as the primary evaluation metric because its value can be interpreted directly in mathematics proficiency percentage points.

Lower MAE indicates better average predictive accuracy.

**Root Mean Squared Error (RMSE)**

RMSE also measures prediction error but places greater emphasis on larger errors.

It will be used to assess whether the models produce substantial individual prediction errors.

Lower RMSE indicates better performance.

**R-squared (R²)**

R-squared measures the relative reduction in squared prediction error compared with a mean-based reference for the evaluated observations.

Higher R-squared generally indicates better predictive performance.

Unlike classification accuracy, R-squared is not the percentage of predictions that are correct.

**Primary Selection Criterion**

The preferred model will be selected primarily using the lowest test MAE, supported by RMSE, R-squared, cross-validation results, and model interpretability.

Predictive performance will be evaluated using observations that were not used to fit the model.

#### 6.1.5 Data Leakage Prevention and Reproducibility

The modelling workflow will be designed to prevent data leakage and support reproducible results.

Mathematics proficiency will be used exclusively as the prediction target and will not be included among the independent features.

Other recorded academic outcomes, including reading proficiency and graduation rate, will not be used as predictors in the initial modelling investigation.

This decision maintains the focus on institutional characteristics and avoids relying on additional outcome measurements that may not be available when predictions are required.

The dataset will be separated into training and testing subsets before any preprocessing steps that require learning information from the data.

Any such preprocessing will be fitted using the training data only and applied consistently to the test data.

A fixed random state will be used for reproducibility.

Cross-validation and any model selection procedures will be performed without exposing the held-out test set to the model development process.

These decisions are intended to provide a fairer assessment of predictive performance and reduce the risk of overly optimistic results.

The machine learning investigation will predict mathematics proficiency using four institutional characteristics: per-pupil expenditure, student-teacher ratio, free or reduced-price lunch eligibility, and broadband access.

The problem is defined as supervised regression because the target variable is numerical.

Three models were selected: a mean-based Dummy Regressor, Linear Regression, and Random Forest Regressor.

The dataset will be divided into training and testing subsets using an 80:20 split and a fixed random state of 42.

The Dummy Regressor will provide a baseline for evaluating whether the trained models improve predictive performance.

Linear Regression will provide an interpretable model, while Random Forest Regressor will investigate whether nonlinear relationships improve prediction.

Mean Absolute Error will serve as the primary evaluation metric, supported by Root Mean Squared Error and R-squared.

#### 6.2.2 Missing Values and Data Type Validation

#### 6.2.3 Duplicate and Numerical Range Validation

I conducted a final validation of the modelling dataset to confirm institution-level uniqueness and plausible numerical values.

Institution identifiers were checked for duplicates to reduce the risk of repeated institutions appearing in the modelling data.

I also verified that expenditure and student-teacher ratios were positive and that percentage-based variables fell between 0 and 100.

This validation complements the earlier data cleaning process and ensures that the selected modelling variables satisfy basic numerical requirements.

Statistically unusual but plausible values were retained in accordance with the previously documented outlier-handling decisions.

```python

# ---------------------------------------------------------
# 6.2.3 Validate Duplicates and Numerical Ranges
# ---------------------------------------------------------

# Check duplicate institution identifiers
duplicate_ids = ml_data["institution_id"].duplicated().sum()

# Check completely duplicated records
duplicate_rows = ml_data.duplicated().sum()

print("Duplicate Validation")
print("-" * 50)

print(f"Duplicate Institution IDs: {duplicate_ids}")
print(f"Duplicate Rows: {duplicate_rows}")

# Define validity conditions for each modelling variable
validity_checks = {
    "Positive Per-Pupil Expenditure": (
        ml_data["per_pupil_expenditure_usd"] > 0
    ),
    "Positive Student-Teacher Ratio": (
        ml_data["student_teacher_ratio"] > 0
    ),
    "Valid Lunch Eligibility Percentage": (
        ml_data["percent_free_reduced_lunch"].between(0, 100)
    ),
    "Valid Broadband Percentage": (
        ml_data["broadband_access_pct"].between(0, 100)
    ),
    "Valid Mathematics Proficiency Percentage": (
        ml_data["math_proficiency_pct"].between(0, 100)
    )
}

# Count invalid observations for each check
validation_results = []

for check_name, valid_condition in validity_checks.items():

    invalid_count = int((~valid_condition).sum())

    validation_results.append({
        "Validation_Check": check_name,
        "Invalid_Observations": invalid_count,
        "Status": "Passed" if invalid_count == 0 else "Review Required"
    })

# Convert results into a DataFrame
ml_validation_summary = pd.DataFrame(validation_results)

# Display the validation results
ml_validation_summary
```

Following the validation process, I separated the modelling dataset into an independent feature matrix and a target variable.

The feature matrix, X, contains the four institutional characteristics selected for predicting mathematics proficiency.

```python

# ---------------------------------------------------------
# 6.2.4 Separate Features and Target
# ---------------------------------------------------------

# Create the feature matrix
X_ml = ml_data[ml_features].copy()

# Create the target variable
y_ml = ml_data[ml_target].copy()

# Confirm dimensions
print("Feature and Target Preparation")
print("-" * 50)

print(f"Feature Matrix Shape: {X_ml.shape}")
print(f"Target Shape: {y_ml.shape}")

# Confirm that the target is not included in the features
assert ml_target not in X_ml.columns, (
    "Data leakage detected: target included in features."
)

# Confirm that the institution ID is excluded
assert "institution_id" not in X_ml.columns, (
    "Institution identifier should not be used as a predictor."
)

# Confirm all modelling values are complete
assert X_ml.notna().all().all(), (
    "Missing values detected in features."
)

assert y_ml.notna().all(), (
    "Missing values detected in target."
)

# Confirm observations align
assert len(X_ml) == len(y_ml), (
    "Feature and target observation counts do not match."
)

print("\nFeature and Target Validation: Passed")

# Display the feature matrix
X_ml.head()
```

### 6.3 Train-Test Split

Following feature selection and data preparation, I divided the modelling dataset into training and testing subsets.

The objective is to evaluate whether the selected institutional characteristics can predict mathematics proficiency for observations that were not used during model training.

I adopted an 80:20 train-test split, allocating approximately 80% of the observations to training and the remaining 20% to testing.

A fixed random state of 42 was used to ensure reproducibility.

The training dataset will be used for fitting the baseline, Linear Regression, and Random Forest models, as well as subsequent cross-validation and any required model tuning.

The test dataset will be reserved for the final evaluation of predictive performance.

This separation helps reduce overly optimistic performance estimates that may arise from evaluating a model using the same data on which it was trained.

The evaluation will measure generalization to held-out institutions within the supplied dataset, rather than performance across future academic years or entirely different educational populations.

#### Perform the 80:20 train-test split

```python

# ---------------------------------------------------------
# 6.3.1 Split the Dataset into Training and Testing Sets
# ---------------------------------------------------------

from sklearn.model_selection import train_test_split

# Define the train-test split configuration
test_size = 0.20
random_state = 42

# Split the features and target simultaneously
X_train, X_test, y_train, y_test = train_test_split(
    X_ml,
    y_ml,
    test_size=test_size,
    random_state=random_state,
    shuffle=True
)

# Display the split dimensions
print("Train-Test Split Results")
print("-" * 50)

print(f"Total Observations: {len(X_ml)}")
print(f"Training Observations: {len(X_train)}")
print(f"Testing Observations: {len(X_test)}")

print("\nFeature Matrix Dimensions")
print(f"X_train: {X_train.shape}")
print(f"X_test: {X_test.shape}")

print("\nTarget Dimensions")
print(f"y_train: {y_train.shape}")
print(f"y_test: {y_test.shape}")

print("\nSplit Configuration")
print(f"Test Size: {test_size:.0%}")
print(f"Random State: {random_state}")
```

#### 6.3.2 Training and Testing Proportion Validation

I verified the number and proportion of observations allocated to each subset.

The purpose of this validation is to confirm that the train-test split was performed as intended and that all 4,642 institutional observations were accounted for.

The validation also ensures that the feature matrices and target variables contain corresponding numbers of observations.

```python

# ---------------------------------------------------------
# 6.3.2 Verify Training and Testing Proportions
# ---------------------------------------------------------

import pandas as pd

# Calculate total observations
total_observations = len(X_ml)

# Calculate the number of observations in each subset
training_count = len(X_train)
testing_count = len(X_test)

# Calculate actual percentages
training_percentage = (
    training_count / total_observations
) * 100

testing_percentage = (
    testing_count / total_observations
) * 100

# Create a summary table
split_summary = pd.DataFrame({
    "Dataset": ["Training", "Testing", "Total"],
    "Observations": [
        training_count,
        testing_count,
        total_observations
    ],
    "Percentage": [
        training_percentage,
        testing_percentage,
        training_percentage + testing_percentage
    ]
})

# Confirm that no observations were lost
assert training_count + testing_count == total_observations, (
    "The training and testing counts do not match the original dataset."
)

# Confirm that features and targets are aligned
assert len(X_train) == len(y_train), (
    "Training features and targets are not aligned."
)

assert len(X_test) == len(y_test), (
    "Testing features and targets are not aligned."
)

# Display the results
print("Training and Testing Proportions")
print("-" * 50)

print(split_summary.round(2))
```

The slight difference from exactly 80% and 20% occurs because we cannot divide an institution into fractional observations.

Our split is correct.

#### 6.3.3 Training and Testing Overlap Validation

I checked whether any institutional observations appeared in both the training and testing subsets.

The purpose of this validation is to confirm that the held-out test observations remain separate from the training data.

I compared the original row indexes of the training and testing feature matrices and checked the alignment of their corresponding target values.

The split was performed using the previously validated institution-level dataset, which contained no duplicate institution identifiers.

A successful validation requires zero overlapping observations between the training and testing subsets.

```python

# ---------------------------------------------------------
# 6.3.3 Verify No Overlap Between Training and Testing Data
# ---------------------------------------------------------

# Check for overlapping original DataFrame indexes
overlapping_indexes = X_train.index.intersection(
    X_test.index
)

# Count overlapping observations
overlap_count = len(overlapping_indexes)

# Verify that training features and targets align
train_indexes_match = X_train.index.equals(
    y_train.index
)

# Verify that testing features and targets align
test_indexes_match = X_test.index.equals(
    y_test.index
)

# Confirm that no institution appears in both subsets
assert overlap_count == 0, (
    "Data leakage risk: observations overlap between training and testing sets."
)

# Confirm alignment between features and targets
assert train_indexes_match, (
    "Training feature and target indexes do not match."
)

assert test_indexes_match, (
    "Testing feature and target indexes do not match."
)

# Verify that the combined indexes reproduce the original rows
combined_indexes = X_train.index.append(X_test.index)

assert set(combined_indexes) == set(X_ml.index), (
    "Some original observations are missing from the split."
)

# Display validation results
print("Train-Test Overlap Validation")
print("-" * 50)

print(f"Overlapping Observations: {overlap_count}")
print(f"Training Index Alignment: {train_indexes_match}")
print(f"Testing Index Alignment: {test_indexes_match}")

print("\nValidation Status: Passed")
```

There are no overlapping observations between the training and testing datasets.

This confirms that the split does not reuse the same institutional records in both subsets.

However, it does not completely eliminate every possible source of data leakage. We must also avoid using test-set information during preprocessing, tuning, and model selection.

### 6.4 Baseline Regression Model

Before training the selected machine learning models, I established a baseline regression model to provide a reference point for evaluating predictive performance.

The baseline was developed using scikit-learn's `DummyRegressor` with the mean prediction strategy.

This approach predicts the average mathematics proficiency calculated from the training target for every observation, without learning relationships between the institutional predictors and the target.

The purpose is to determine whether Linear Regression and Random Forest Regressor provide improvements over a simple prediction strategy.

The baseline model will be fitted using the training dataset only.

Its performance will be examined using Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and R-squared (R²).

The baseline will serve as a benchmark for subsequent model comparison, rather than as the preferred predictive solution.

```python

# ---------------------------------------------------------
# 6.4.1 Train the Baseline Regression Model
# ---------------------------------------------------------

from sklearn.dummy import DummyRegressor

# Initialize the baseline model
# The mean strategy predicts the training target average
baseline_model = DummyRegressor(strategy="mean")

# Fit the baseline using training data only
baseline_model.fit(X_train, y_train)

# Extract the average mathematics proficiency learned by the model
baseline_prediction_value = float(
    baseline_model.constant_.item()
)

# Display the learned baseline value
print("Baseline Regression Model")
print("-" * 50)

print("Model: DummyRegressor")
print("Strategy: Mean")

print(f"Training Observations: {len(X_train)}")

print(
    f"Training Mathematics Proficiency Mean: "
    f"{y_train.mean():.4f}%"
)

print(
    f"Baseline Prediction Value: "
    f"{baseline_prediction_value:.4f}%"
)
```

Our baseline model learned that average mathematics proficiency in the training dataset was approximately 58.62%.

It will predict 58.62% for every institution.

This model does not consider individual institutional characteristics.

That limitation is intentional: it provides a simple benchmark for evaluating whether more sophisticated models actually improve prediction.

#### 6.4.2 Generating Baseline Predictions

After fitting the baseline model, I generated predictions for the training and testing subsets.

The model predicts the same training-set mean mathematics proficiency for every observation.

Training predictions will be used to understand baseline behaviour and establish a training-performance reference.

Test predictions will be stored for the final model evaluation stage, where all selected models will be compared using the same held-out observations.

No model development decisions will be made using the test-set prediction errors at this stage.

```python

# ---------------------------------------------------------
# 6.4.2 Generate Baseline Predictions
# ---------------------------------------------------------

# Generate baseline predictions on the training set
baseline_train_predictions = baseline_model.predict(X_train)

# Generate predictions for the reserved test set
# These will be formally evaluated in Section 7
baseline_test_predictions = baseline_model.predict(X_test)

# Confirm the number of predictions
print("Baseline Prediction Summary")
print("-" * 50)

print(
    f"Number of Training Predictions: "
    f"{len(baseline_train_predictions)}"
)

print(
    f"Number of Test Predictions: "
    f"{len(baseline_test_predictions)}"
)

# Display the first five training predictions
baseline_prediction_sample = pd.DataFrame({
    "Actual_Math_Proficiency": y_train.iloc[:5].to_numpy(),
    "Baseline_Prediction": baseline_train_predictions[:5]
})

print("\nFirst Five Training Predictions:")
print(baseline_prediction_sample.round(2))

# Confirm that every prediction equals the learned mean
assert (
    abs(
        baseline_train_predictions - baseline_prediction_value
    ) < 1e-10
).all(), "Unexpected variation in baseline training predictions."

assert (
    abs(
        baseline_test_predictions - baseline_prediction_value
    ) < 1e-10
).all(), "Unexpected variation in baseline test predictions."

print("\nPrediction Validation: Passed")
```

#### 6.4.3 Baseline Training Performance

I evaluated the baseline model on the training dataset using Mean Absolute Error, Root Mean Squared Error, and R-squared.

These metrics establish a training-performance reference for the subsequent regression models.

MAE and RMSE are expressed in mathematics proficiency percentage points, making the prediction errors directly interpretable.

The mean-based Dummy Regressor is expected to produce an R-squared value of approximately zero on the training dataset because it predicts the training target mean.

This assessment describes baseline training performance only.

The final comparison of held-out test performance will be conducted in Section 7 after the remaining models have been trained and their development decisions finalized.

```python

# ---------------------------------------------------------
# 6.4.3 Evaluate Baseline Training Performance
# ---------------------------------------------------------

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Calculate Mean Absolute Error
baseline_train_mae = mean_absolute_error(
    y_train,
    baseline_train_predictions
)

# Calculate Root Mean Squared Error
baseline_train_rmse = np.sqrt(
    mean_squared_error(
        y_train,
        baseline_train_predictions
    )
)

# Calculate R-squared
baseline_train_r2 = r2_score(
    y_train,
    baseline_train_predictions
)

# Display the training metrics
print("Baseline Model: Training Performance")
print("-" * 50)

print(f"MAE: {baseline_train_mae:.4f} percentage points")
print(f"RMSE: {baseline_train_rmse:.4f} percentage points")
print(f"R-squared: {baseline_train_r2:.4f}")
```

**MAE = 12.91 percentage points**

The baseline predictions differ from actual training mathematics proficiency by approximately 12.91 percentage points on average, when we ignore whether an error is positive or negative.

**RMSE = 15.94 percentage points**

This metric places greater weight on larger prediction errors. Its value exceeds MAE, indicating that some predictions have relatively large errors.

**R² = 0.0000**

This is expected because the model predicts the training-set mean.

It does not capture institution-specific differences in mathematics proficiency.

Our Linear Regression and Random Forest models should ideally achieve lower errors and higher validation R² than the baseline.

However, we'll let the actual results determine whether they do.

### 6.5 Linear Regression Model

Following the development of the baseline model, I trained a Linear Regression model to predict mathematics proficiency using four institutional characteristics.

The predictors were per-pupil expenditure, student-teacher ratio, free or reduced-price lunch eligibility, and broadband access.

Linear Regression was selected because it provides an interpretable approach to estimating numerical outcomes and serves as an appropriate comparison with the mean-based baseline model.

The model was trained exclusively on the 3,713 training observations obtained from the established 80:20 train-test split.

Its training performance was evaluated using Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and R-squared (R²).

The objective is to determine whether incorporating institutional characteristics improves training performance compared with the baseline prediction strategy.

The held-out test dataset remains reserved for the final model evaluation stage.

#### Train the Linear Regression model

```python

# ---------------------------------------------------------
# 6.5.1 Train the Linear Regression Model
# ---------------------------------------------------------

from sklearn.linear_model import LinearRegression

# Initialize the Linear Regression model
linear_model = LinearRegression()

# Train the model using training data only
linear_model.fit(X_train, y_train)

# Confirm that the model has been trained
print("Linear Regression Model Training")
print("-" * 50)

print("Model: LinearRegression")
print(f"Training Observations: {len(X_train)}")
print(f"Number of Features: {X_train.shape[1]}")

print("\nModel Training: Completed")
```

The .fit() method examined the relationship between the four institutional predictors and mathematics proficiency in the training dataset.

It then calculated the coefficients that minimize the sum of squared training prediction errors.

Note: This model is different from the explanatory OLS model in Section 4.4 because it was fitted only on our training subset, rather than the complete analytical dataset.

That's the correct approach for machine learning evaluation.

#### 6.5.2 Fitted Linear Regression Coefficients

After training the model, I examined the intercept and regression coefficients to understand how the selected institutional characteristics contribute to its fitted predictions.

Each coefficient represents the estimated change in predicted mathematics proficiency associated with a one-unit difference in the corresponding predictor, holding the other model predictors constant.

The coefficients are expressed in their original measurement units.

Their signs indicate the direction of the fitted linear relationships.

These coefficients describe predictive model behaviour and should not be interpreted as causal effects or as independent evidence of statistical significance.

```python

# ---------------------------------------------------------
# 6.5.2 Extract Linear Regression Coefficients
# ---------------------------------------------------------

# Extract the fitted intercept
linear_intercept = float(linear_model.intercept_)

# Extract the fitted coefficients
linear_coefficients = pd.DataFrame({
    "Feature": X_train.columns,
    "Coefficient": linear_model.coef_
})

# Add a readable direction column
linear_coefficients["Direction"] = (
    linear_coefficients["Coefficient"].apply(
        lambda value: (
            "Positive" if value > 0
            else "Negative" if value < 0
            else "Zero"
        )
    )
)

# Display the intercept
print("Linear Regression Intercept")
print("-" * 50)

print(f"Intercept: {linear_intercept:.6f}")

# Display the feature coefficients
print("\nLinear Regression Coefficients")
print("-" * 50)

print(linear_coefficients.round(6))
```

##### How do we interpret the coefficients?

**Per-pupil expenditure:** Holding the other predictors constant, a $1,000 difference in expenditure corresponds to approximately 0.694 percentage points in fitted mathematics proficiency.

**Student-teacher ratio:** A one-unit difference in student-teacher ratio corresponds to approximately +0.038 percentage points in fitted proficiency, holding the other predictors constant.

**Free/reduced lunch eligibility:** A one-percentage-point difference in eligibility corresponds to approximately -0.399 percentage points in fitted proficiency, holding the other predictors constant.

**Broadband access:** A one-percentage-point difference in broadband access corresponds to approximately +0.005 percentage points in fitted proficiency, holding the other predictors constant.

These coefficients are broadly consistent in direction with our earlier multiple regression results.

However, they are not identical because the machine learning model was trained on 80% of the observations.

The coefficients also do not tell us whether the model predicts unseen institutions accurately. We'll assess that separately.

#### 6.5.3 Generating Linear Regression Training Predictions

I generated mathematics proficiency predictions for the training observations using the fitted Linear Regression model.

Unlike the baseline model, which predicts a constant value, Linear Regression uses each institution's expenditure, student-teacher ratio, lunch eligibility, and broadband access to estimate mathematics proficiency.

I compared the predicted values with the actual training outcomes to examine the model's fitted behaviour.

These predictions will be used to calculate training performance metrics.

The held-out test observations remain excluded from the model training and evaluation activities conducted in this subsection.

```python

# ---------------------------------------------------------
# 6.5.3 Generate Linear Regression Training Predictions
# ---------------------------------------------------------

# Generate predictions using the training features
linear_train_predictions = linear_model.predict(X_train)

# Create a comparison of actual and predicted values
linear_prediction_sample = pd.DataFrame({
    "Actual_Math_Proficiency": y_train.to_numpy(),
    "Predicted_Math_Proficiency": linear_train_predictions
})

# Calculate the prediction errors
linear_prediction_sample["Prediction_Error"] = (
    linear_prediction_sample["Actual_Math_Proficiency"]
    - linear_prediction_sample["Predicted_Math_Proficiency"]
)

# Display prediction information
print("Linear Regression Training Predictions")
print("-" * 50)

print(f"Number of Predictions: {len(linear_train_predictions)}")

print("\nFirst Five Predictions:")
print(linear_prediction_sample.head().round(2))

# Confirm that the number of predictions matches the training target
assert len(linear_train_predictions) == len(y_train), (
    "The number of predictions does not match the training observations."
)

# Confirm that predictions contain finite numerical values
import numpy as np

assert np.isfinite(linear_train_predictions).all(), (
    "Invalid numerical values detected in predictions."
)

print("\nPrediction Validation: Passed")
```

#### 6.5.4 Linear Regression Training Performance

I evaluated the fitted Linear Regression model on the training dataset using Mean Absolute Error, Root Mean Squared Error, and R-squared.

The metrics were calculated from the differences between actual and predicted mathematics proficiency values.

MAE provides the average absolute prediction error in mathematics proficiency percentage points.

RMSE places greater emphasis on larger prediction errors.

R-squared measures the proportion of training outcome variation accounted for by the model relative to predicting the training target mean.

These results will be compared with the baseline training metrics to determine whether the institutional predictors improve in-sample fit.

However, training performance alone will not determine the preferred model because it does not establish predictive performance on unseen institutions.

```python

# ---------------------------------------------------------
# 6.5.4 Evaluate Linear Regression Training Performance
# ---------------------------------------------------------

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Calculate Mean Absolute Error
linear_train_mae = mean_absolute_error(
    y_train,
    linear_train_predictions
)

# Calculate Root Mean Squared Error
linear_train_rmse = np.sqrt(
    mean_squared_error(
        y_train,
        linear_train_predictions
    )
)

# Calculate R-squared
linear_train_r2 = r2_score(
    y_train,
    linear_train_predictions
)

# Display the evaluation results
print("Linear Regression: Training Performance")
print("-" * 50)

print(f"MAE: {linear_train_mae:.4f} percentage points")
print(f"RMSE: {linear_train_rmse:.4f} percentage points")
print(f"R-squared: {linear_train_r2:.4f}")
```

**MAE = 10.08 percentage points**

The model's predictions differ from actual training mathematics proficiency by approximately 10.08 percentage points on average.

**RMSE = 12.51 percentage points**

The model still produces meaningful prediction errors, including some relatively large errors.

**R² = 0.3838**

The Linear Regression model accounts for approximately 38.38% of the observed variation in training mathematics proficiency.

This is an improvement over our baseline training R² of zero.

However, it also indicates that substantial variation remains unexplained by the selected linear model.

Note: These are training results. They do not yet establish how well the model performs on unseen institutions.

#### 6.5.5 Training Performance Comparison with the Baseline

I compared the Linear Regression model with the mean-based baseline using the same 3,713 training observations.

Both models were evaluated using MAE, RMSE, and R-squared.

The comparison was conducted to determine whether incorporating the four selected institutional characteristics improves training performance over predicting the average mathematics proficiency.

Any improvement observed at this stage is an in-sample improvement and must be confirmed through cross-validation and held-out test evaluation before selecting a final predictive model.

```python

# ---------------------------------------------------------
# 6.5.5 Compare Baseline and Linear Regression Training Results
# ---------------------------------------------------------

# Rebuild the comparison table directly from saved model metrics
# This avoids creating duplicate rows when the cell is rerun
training_results = pd.DataFrame([
    {
        "Model": "Baseline (Mean DummyRegressor)",
        "Train_MAE": baseline_train_mae,
        "Train_RMSE": baseline_train_rmse,
        "Train_R2": baseline_train_r2
    },
    {
        "Model": "Linear Regression",
        "Train_MAE": linear_train_mae,
        "Train_RMSE": linear_train_rmse,
        "Train_R2": linear_train_r2
    }
])

# Calculate the improvement in MAE relative to baseline
mae_improvement = (
    baseline_train_mae - linear_train_mae
)

mae_improvement_pct = (
    mae_improvement / baseline_train_mae
) * 100

# Calculate the reduction in RMSE
rmse_improvement = (
    baseline_train_rmse - linear_train_rmse
)

# Display the model comparison
print("Training Performance Comparison")
print("-" * 60)

print(training_results.round(4).to_string(index=False))

print("\nImprovement Relative to Baseline")
print("-" * 60)

print(
    f"MAE Reduction: "
    f"{mae_improvement:.4f} percentage points"
)

print(
    f"Percentage Reduction in MAE: "
    f"{mae_improvement_pct:.2f}%"
)

print(
    f"RMSE Reduction: "
    f"{rmse_improvement:.4f} percentage points"
)
```

Linear Regression performed better than the baseline on all three training metrics.

Its training MAE decreased from approximately 12.91 to 10.08 percentage points.

Its training RMSE also decreased, while R² increased from zero to approximately 0.3838.

Finding: The four institutional characteristics improve the model's ability to represent variation in training mathematics proficiency compared with predicting the mean.

Limitation: Since this comparison uses training observations, we cannot yet conclude that Linear Regression will outperform the baseline on unseen institutions.

#### 6.5.6 Actual Versus Predicted Mathematics Proficiency

I visualized the relationship between actual and predicted mathematics proficiency for the training observations.

The scatterplot compares the recorded proficiency values with the predictions generated by the Linear Regression model.

A diagonal reference line represents perfect agreement between actual and predicted values.

This visualization provides an indication of the model's training fit and helps identify the extent of unexplained variation.

The results will not be interpreted as evidence of generalization to unseen institutions.

```python

# ---------------------------------------------------------
# 6.5.6 Visualize Actual vs Predicted Training Values
# ---------------------------------------------------------

# Convert actual training values to a NumPy array
actual_values = y_train.to_numpy()

# Determine common plotting limits
plot_min = min(
    actual_values.min(),
    linear_train_predictions.min()
)

plot_max = max(
    actual_values.max(),
    linear_train_predictions.max()
)

# Create the scatterplot
plt.figure(figsize=(9, 7))

plt.scatter(
    actual_values,
    linear_train_predictions,
    alpha=0.35,
    s=25,
    color="steelblue"
)

# Add the perfect-agreement reference line
plt.plot(
    [plot_min, plot_max],
    [plot_min, plot_max],
    color="red",
    linestyle="--",
    linewidth=2,
    label="Perfect Agreement"
)

# Add the training R-squared value
plt.text(
    0.05,
    0.95,
    f"Training R-squared = {linear_train_r2:.4f}",
    transform=plt.gca().transAxes,
    verticalalignment="top",
    bbox={
        "boxstyle": "round",
        "facecolor": "white",
        "alpha": 0.85
    }
)

# Add the chart title and labels
plt.title(
    "Linear Regression: Actual vs Predicted Mathematics Proficiency",
    fontsize=14
)

plt.xlabel("Actual Mathematics Proficiency (%)")
plt.ylabel("Predicted Mathematics Proficiency (%)")

# Display legend
plt.legend()

# Improve the layout
plt.tight_layout()

# Display the chart
plt.show()
```

#### 6.5.7 Linear Regression Model: Findings and Conclusion

I trained a Linear Regression model using the four selected institutional predictors and 3,713 training observations.

The model was fitted exclusively on the training data to preserve the held-out test set for final evaluation.

The fitted coefficients indicated positive relationships between mathematics proficiency and per-pupil expenditure, student-teacher ratio, and broadband access, while free or reduced-price lunch eligibility demonstrated a negative relationship.

These coefficients represent adjusted relationships within the fitted model and should not be interpreted as causal effects.

The model produced the following training results:

- **Mean Absolute Error (MAE):** 10.08 percentage points.
- **Root Mean Squared Error (RMSE):** 12.51 percentage points.
- **R-squared (R²):** 0.3838.

Compared with the baseline model, Linear Regression reduced training MAE from approximately 12.91 to 10.08 percentage points.

**Finding:** The four institutional predictors provide useful information for fitting mathematics proficiency compared with a constant mean prediction.

**Analytical Decision:** I retained Linear Regression as an interpretable predictive candidate for subsequent model evaluation.

**Limitation:** The results describe training performance only. They do not establish whether Linear Regression will outperform the baseline on unseen institutional observations.

**Conclusion:** Linear Regression improved training performance compared with the baseline, demonstrating the potential predictive value of the selected institutional characteristics.

The next stage will train a Random Forest Regressor to investigate whether a model capable of capturing nonlinear relationships and feature interactions can provide additional predictive improvements.

Final model selection will be based on cross-validation and held-out test evaluation rather than training performance alone.

### 6.6 Random Forest Regressor

Following the Linear Regression investigation, I developed a Random Forest Regressor to examine whether a nonlinear machine learning model could improve predictions of mathematics proficiency.

Random Forest combines predictions from multiple decision trees and can represent nonlinear relationships and interactions between institutional characteristics.

The model uses the same four predictors: per-pupil expenditure, student-teacher ratio, free or reduced-price lunch eligibility, and broadband access.

To ensure a consistent comparison, I retained the established training dataset containing 3,713 institutions and the fixed random state of 42.

I also introduced constraints on tree depth and minimum leaf size to limit model complexity and reduce the risk of overfitting.

The model will first be assessed using training MAE, RMSE, and R-squared.

These findings will be compared with the baseline and Linear Regression training results. Cross-validation and final held-out test evaluation will follow in Section 7.

#### 6.6.1 Random Forest Model Configuration

I configured the Random Forest Regressor with 200 decision trees, a maximum tree depth of 10, and a minimum of five training observations per leaf.

These settings were selected as an initial constrained configuration to balance model flexibility with the risk of overfitting.

A fixed random state of 42 was used for reproducibility.

The model was configured to use available processing cores during training.

These hyperparameters are initial modelling choices rather than optimized values.

The model's suitability will be assessed using cross-validation and subsequent evaluation on held-out observations.

#### Train the Random Forest model

```python

# ---------------------------------------------------------
# 6.6.2 Train the Random Forest Regressor
# ---------------------------------------------------------

from sklearn.ensemble import RandomForestRegressor

# Initialize Random Forest with controlled model complexity
rf_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=10,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

# Fit the model using training data only
rf_model.fit(X_train, y_train)

# Display training information
print("Random Forest Regressor Training")
print("-" * 50)

print("Model: RandomForestRegressor")
print(f"Training Observations: {len(X_train)}")
print(f"Number of Features: {X_train.shape[1]}")

print("\nModel Configuration")
print(f"Number of Trees: {rf_model.n_estimators}")
print(f"Maximum Tree Depth: {rf_model.max_depth}")
print(f"Minimum Samples per Leaf: {rf_model.min_samples_leaf}")
print(f"Random State: {rf_model.random_state}")

print("\nModel Training: Completed")
```

#### 6.6.3 Random Forest Training Predictions

Following model training, I generated mathematics proficiency predictions for the training observations using the fitted Random Forest Regressor.

The model combines predictions from multiple decision trees to produce a numerical estimate for each institution.

I compared the predicted values with actual mathematics proficiency and calculated individual prediction errors.

The purpose of this assessment is to examine the fitted model's training behaviour before evaluating its generalization performance through cross-validation and held-out testing.

```python

# ---------------------------------------------------------
# 6.6.3 Generate Random Forest Training Predictions
# ---------------------------------------------------------

# Generate predictions for training institutions
rf_train_predictions = rf_model.predict(X_train)

# Compare actual and predicted values
rf_prediction_sample = pd.DataFrame({
    "Actual_Math_Proficiency": y_train.to_numpy(),
    "Predicted_Math_Proficiency": rf_train_predictions
})

# Calculate signed prediction errors
rf_prediction_sample["Prediction_Error"] = (
    rf_prediction_sample["Actual_Math_Proficiency"]
    - rf_prediction_sample["Predicted_Math_Proficiency"]
)

# Confirm the number and validity of predictions
assert len(rf_train_predictions) == len(y_train), (
    "Prediction count does not match training observations."
)

assert np.isfinite(rf_train_predictions).all(), (
    "Random Forest generated invalid numerical predictions."
)

# Display the results
print("Random Forest Training Predictions")
print("-" * 50)

print(f"Number of Predictions: {len(rf_train_predictions)}")

print("\nFirst Five Predictions:")
print(rf_prediction_sample.head().round(2))

print("\nPrediction Validation: Passed")
```

#### 6.6.4 Random Forest Training Performance

I evaluated the Random Forest Regressor using Mean Absolute Error, Root Mean Squared Error, and R-squared on the training observations.

These metrics were selected to maintain consistency with the previous models.

MAE measures the average absolute prediction error in mathematics proficiency percentage points.

RMSE places greater emphasis on larger prediction errors.

R-squared measures the model's training fit relative to predicting the training target mean.

The results will be compared with the baseline and Linear Regression training performances.

However, strong training performance does not necessarily indicate strong generalization, particularly for flexible tree-based models.

Cross-validation and held-out test evaluation remain necessary before selecting a final model.

```python

# ---------------------------------------------------------
# 6.6.4 Evaluate Random Forest Training Performance
# ---------------------------------------------------------

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Calculate Mean Absolute Error
rf_train_mae = mean_absolute_error(
    y_train,
    rf_train_predictions
)

# Calculate Root Mean Squared Error
rf_train_rmse = np.sqrt(
    mean_squared_error(
        y_train,
        rf_train_predictions
    )
)

# Calculate R-squared
rf_train_r2 = r2_score(
    y_train,
    rf_train_predictions
)

# Display the evaluation results
print("Random Forest: Training Performance")
print("-" * 50)

print(f"MAE: {rf_train_mae:.4f} percentage points")
print(f"RMSE: {rf_train_rmse:.4f} percentage points")
print(f"R-squared: {rf_train_r2:.4f}")
```

**MAE = 8.04 percentage points**

The Random Forest predictions differ from the actual training mathematics proficiency by approximately 8.04 percentage points on average.

**RMSE = 10.05 percentage points**

The model still makes errors, including some relatively large ones, but its training RMSE is lower than that of Linear Regression.

**R² = 0.6023**

Random Forest accounts for approximately 60.23% of the observed variation in training mathematics proficiency.

This is higher than the Linear Regression training R² of 0.3838.

Note: Random Forest is a more flexible model, so stronger training performance alone does not establish that it is better for unseen institutions.

#### 6.6.5 Training Performance Comparison

I compared the Random Forest Regressor with the previously trained baseline and Linear Regression models.

All three models were trained using the same 3,713 institutional observations and evaluated using the same training target.

The comparison focuses on MAE, RMSE, and R-squared to establish how well each model represents the training data.

Because Random Forest can capture nonlinear relationships and interactions, it may achieve stronger training performance than Linear Regression.

However, the comparison will not be used to select the final model. Predictive reliability must be assessed through cross-validation and held-out evaluation.

```python

# ---------------------------------------------------------
# 6.6.5 Compare All Three Models on Training Data
# ---------------------------------------------------------

# Build the training comparison table
# Rebuilding prevents duplicate rows if the cell is rerun
training_results = pd.DataFrame([
    {
        "Model": "Baseline (Mean DummyRegressor)",
        "Train_MAE": baseline_train_mae,
        "Train_RMSE": baseline_train_rmse,
        "Train_R2": baseline_train_r2
    },
    {
        "Model": "Linear Regression",
        "Train_MAE": linear_train_mae,
        "Train_RMSE": linear_train_rmse,
        "Train_R2": linear_train_r2
    },
    {
        "Model": "Random Forest",
        "Train_MAE": rf_train_mae,
        "Train_RMSE": rf_train_rmse,
        "Train_R2": rf_train_r2
    }
])

# Sort models by training MAE
training_results = training_results.sort_values(
    by="Train_MAE",
    ascending=True
).reset_index(drop=True)

# Calculate Random Forest improvement relative to Linear Regression
rf_vs_linear_mae_reduction = (
    linear_train_mae - rf_train_mae
)

rf_vs_linear_pct = (
    rf_vs_linear_mae_reduction / linear_train_mae
) * 100

# Calculate Random Forest improvement relative to the baseline
rf_vs_baseline_pct = (
    (baseline_train_mae - rf_train_mae) / baseline_train_mae
) * 100

# Display the comparison
print("Machine Learning Training Performance Comparison")
print("-" * 65)

print(training_results.round(4).to_string(index=False))

print("\nRandom Forest Training MAE Improvements")
print("-" * 65)

print(
    f"Reduction vs Linear Regression: "
    f"{rf_vs_linear_mae_reduction:.4f} percentage points"
)

print(
    f"Relative Improvement vs Linear Regression: "
    f"{rf_vs_linear_pct:.2f}%"
)

print(
    f"Relative Improvement vs Baseline: "
    f"{rf_vs_baseline_pct:.2f}%"
)
```

Random Forest achieved the lowest training MAE and RMSE and the highest training R².

Compared with Linear Regression, its training MAE was lower by approximately 2.03 percentage points, representing a 20.17% reduction.

Compared with the baseline, its training MAE was approximately 37.70% lower.

These results suggest that Random Forest can fit additional patterns in the training observations.

However, the improvement could partly reflect its greater flexibility.

We still need to establish whether it performs consistently on validation folds and unseen institutions.

#### 6.6.6 Actual Versus Predicted Mathematics Proficiency

I visualized the relationship between actual and predicted mathematics proficiency for the Random Forest training observations.

The scatterplot uses a diagonal reference line to represent perfect agreement between observed and predicted values.

Predictions closer to this line indicate smaller errors, while greater distances indicate larger discrepancies.

## 7. Model Evaluation

The machine learning development stage produced three regression models for predicting mathematics proficiency using institutional characteristics.

The models were trained using 3,713 institutional observations, while 929 observations were reserved for held-out evaluation.

During training, Random Forest achieved the lowest prediction errors and highest R-squared, followed by Linear Regression and the mean-based baseline.

However, training performance alone does not establish whether a model can generalize to unseen institutions.

This section therefore evaluates predictive performance and model stability using appropriate regression metrics, cross-validation, and the reserved test dataset.

The evaluation will prioritize accurate predictions, consistent performance, and interpretability.

The objective is to identify a suitable model for estimating mathematics proficiency while clearly documenting its limitations.

### 7.1 Evaluation Metrics and Criteria

Before comparing the models, I established a consistent evaluation framework.

Three regression metrics were selected: Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and R-squared (R²).

MAE will serve as the primary model-selection metric because it expresses average prediction errors directly in mathematics proficiency percentage points.

RMSE and R-squared will provide additional information about larger errors and the overall quality of model predictions.

Cross-validation will be used to assess model stability before final evaluation on the held-out test observations.

#### 7.1.1 Mean Absolute Error (MAE)

A lower MAE indicates smaller average absolute prediction errors.

I selected MAE as the primary model-selection metric because it provides a straightforward measure of prediction accuracy in the same units as the target variable.

The preferred model should demonstrate a lower MAE than the baseline and provide reasonably consistent performance across validation folds.

#### 7.1.2 Root Mean Squared Error (RMSE)

Unlike MAE, RMSE places greater emphasis on larger errors because the differences are squared before averaging.

RMSE is expressed in mathematics proficiency percentage points.

A lower RMSE indicates better predictive performance, particularly when large errors are undesirable.

I included RMSE as a supporting evaluation metric to examine whether the models produce substantial individual prediction errors.

Comparing MAE and RMSE will help assess whether a model's overall performance may be affected by relatively large prediction errors.

#### 7.1.3 R-squared (R²)

A value of 1 indicates perfect predictions, while a value of 0 indicates the same squared-error performance as the evaluated target mean.

A negative value indicates that the model performs worse than this mean-based reference.

Higher R-squared values generally indicate better predictive performance.

However, R-squared is not a percentage of correctly predicted observations and should not be interpreted as classification accuracy.

I included R-squared as a supporting evaluation metric to assess how well each model accounts for variation in mathematics proficiency on the evaluated observations.

The final comparison will prioritize MAE while considering RMSE and R-squared together.

#### 7.1.4 Model Evaluation and Selection Criteria

The model comparison will use a predefined evaluation strategy to reduce the risk of selecting a model based on favourable results from the held-out test dataset.

**Primary Metric**

Cross-validation MAE will serve as the primary criterion for selecting the preferred model.

The model with the lowest mean validation MAE will be considered the leading candidate.

**Supporting Metrics**

RMSE and R-squared will be examined to provide additional evidence about prediction errors and model performance.

**Cross-Validation Stability**

The mean and variation of validation scores across folds will be examined to assess the consistency of model performance.

**Overfitting Assessment**

Training performance will be compared with cross-validation performance.

Large differences between training and validation results may indicate overfitting.

**Model Interpretability**

Where models demonstrate comparable validation performance, model simplicity and interpretability will be considered in the selection decision.

**Held-Out Test Evaluation**

Following model selection, the reserved test set will be used to evaluate predictive performance on unseen observations.

Test-set results will be reported transparently and will not be used to repeatedly revise model configurations.

The final conclusions will consider predictive accuracy, stability, interpretability, and limitations.

#### 7.1.5 Reusable Regression Evaluation Function

To ensure consistent model evaluation, I defined a reusable function for calculating MAE, RMSE, and R-squared.

The function accepts actual outcomes, predicted outcomes, and a model name.

It returns the three regression metrics in a structured format.

This approach reduces code duplication and supports reproducible comparisons across the baseline, Linear Regression, and Random Forest models.

The function will be used during subsequent validation and test-set evaluation without changing the metric definitions between models.

```python

# ---------------------------------------------------------
# 7.1.5 Define a Reusable Regression Evaluation Function
# ---------------------------------------------------------

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

def evaluate_regression_model(y_true, y_pred, model_name):
    """
    Evaluate a regression model using MAE, RMSE, and R-squared.

    Parameters
    ----------
    y_true : array-like
        Actual mathematics proficiency values.

    y_pred : array-like
        Predicted mathematics proficiency values.

    model_name : str
        Name of the regression model.

    Returns
    -------
    dict
        Dictionary containing model name and evaluation metrics.
    """

    # Convert actual and predicted values to one-dimensional arrays
    actual = np.asarray(y_true, dtype=float).ravel()
    predicted = np.asarray(y_pred, dtype=float).ravel()

    # Validate that the arrays contain corresponding observations
    if len(actual) != len(predicted):
        raise ValueError(
            "Actual and predicted arrays must have the same length."
        )

    # Validate that the arrays are not empty
    if len(actual) == 0:
        raise ValueError(
            "Evaluation requires at least one observation."
        )

    # Validate numerical values
    if not np.isfinite(actual).all():
        raise ValueError(
            "Actual values contain missing or infinite numbers."
        )

    if not np.isfinite(predicted).all():
        raise ValueError(
            "Predicted values contain missing or infinite numbers."
        )

    # Calculate Mean Absolute Error
    mae = mean_absolute_error(
        actual,
        predicted
    )

    # Calculate Root Mean Squared Error
    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )

    # Calculate R-squared when it is defined
    r2 = (
        r2_score(actual, predicted)
        if len(actual) >= 2
        else np.nan
    )

    # Store the results
    results = {
        "Model": model_name,
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2)
    }

    return results


# Confirm that the evaluation function is available
print("Regression evaluation function defined successfully.")
print("Metrics: MAE, RMSE, R-squared")
```

Without a reusable function, we might accidentally calculate metrics differently for different models.

By defining one function, we ensure the same metric calculations are applied consistently.

This also makes our notebook easier to maintain and reproduce.

#### Validate our evaluation function

```python

# ---------------------------------------------------------
# 7.1.6 Validate the Evaluation Function
# ---------------------------------------------------------

# Define a small example of actual values
example_actual = [70, 50, 80]

# Define corresponding predictions
example_predicted = [65, 60, 75]

# Evaluate the example
example_results = evaluate_regression_model(
    y_true=example_actual,
    y_pred=example_predicted,
    model_name="Example Model"
)

# Convert the results to a DataFrame
example_results_table = pd.DataFrame(
    [example_results]
)

# Validate the known MAE and RMSE calculations
assert np.isclose(
    example_results["MAE"],
    20 / 3
), "MAE validation failed."

assert np.isclose(
    example_results["RMSE"],
    np.sqrt(50)
), "RMSE validation failed."

print("Evaluation Function Validation")
print("-" * 50)

print(example_results_table.round(4).to_string(index=False))

print("\nValidation Status: Passed")
```

#### 7.1.6 Evaluation Framework: Findings and Conclusion

The model evaluation framework was established using three regression metrics: Mean Absolute Error, Root Mean Squared Error, and R-squared.

MAE was selected as the primary model-selection metric because it expresses average prediction errors directly in mathematics proficiency percentage points.

RMSE was included to provide additional information about larger prediction errors.

R-squared was selected to assess performance relative to a mean-based reference on the evaluated observations.

A reusable Python function was developed to calculate these metrics consistently across all models.

The function was validated using a small numerical example with known evaluation results.

**Model Selection Strategy**

Cross-validation will be conducted using the training observations to compare model performance and examine stability.

Mean cross-validation MAE will serve as the primary model-selection criterion.

RMSE, R-squared, variation across folds, and model interpretability will provide supporting evidence.

The held-out test dataset will remain reserved until the model development and selection decisions have been finalized.

**Conclusion**

The evaluation framework provides a consistent and reproducible approach for comparing the baseline, Linear Regression, and Random Forest models.

The next stage will conduct cross-validation to investigate how reliably the models predict mathematics proficiency across different validation subsets.

The results will support model selection before final evaluation on held-out institutional observations.

### 7.2 Cross-Validation and Model Stability

Following the establishment of the model evaluation framework, I conducted 5-fold cross-validation to assess the predictive performance and stability of the three regression models.

The models evaluated were the mean-based Dummy Regressor, Linear Regression, and Random Forest Regressor.

Although Random Forest demonstrated the strongest performance on the complete training dataset, this result alone was insufficient to establish its ability to generalize.

I therefore used 5-fold cross-validation on the 3,713 training observations, ensuring that the 929 held-out test observations remained excluded.

For each fold, the models were trained using four portions of the training dataset and evaluated on the remaining portion.

The process was repeated until each portion had served as the validation fold.

I evaluated performance using MAE, RMSE, and R-squared, with mean validation MAE serving as the primary comparison criterion.

The variability of validation scores across folds was also examined to assess model stability.

The objective was to determine whether the stronger Random Forest training performance translated into more reliable validation performance.

#### Configure 5-fold cross-validation

```python

# ---------------------------------------------------------
# 7.2.1 Configure Five-Fold Cross-Validation
# ---------------------------------------------------------

from sklearn.model_selection import KFold

# Define a reproducible 5-fold cross-validation strategy
cv_strategy = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Define the regression scoring metrics
cv_scoring = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}

# Confirm the training dataset
print("Cross-Validation Configuration")
print("-" * 50)

print(f"Training Observations: {len(X_train)}")
print(f"Held-Out Test Observations: {len(X_test)}")

print(f"Number of Folds: {cv_strategy.n_splits}")
print(f"Shuffle Enabled: {cv_strategy.shuffle}")
print(f"Random State: {cv_strategy.random_state}")

print("\nEvaluation Metrics:")
for metric_name in cv_scoring:
    print("-", metric_name)

# Confirm training data and target alignment
assert X_train.index.equals(y_train.index), (
    "Training features and target indexes do not match."
)

print("\nConfiguration Validation: Passed")
```

#### 7.2.1 Five-Fold Cross-Validation Procedure

I evaluated all three regression models using the same five training-validation folds.

The cross-validation procedure refitted each model independently during every fold to ensure that the validation observations were not used for fitting in that iteration.

MAE, RMSE, and R-squared were calculated on each validation fold.

Training-fold scores were also recorded to support the assessment of potential overfitting.

The Random Forest retained its initial configuration of 200 trees, maximum depth of 10, minimum leaf size of five, and random state of 42.

No hyperparameter changes were made using the held-out test dataset.

The resulting scores were stored for comparison of average validation performance, variability, and training-validation differences.

#### Run cross-validation for all three models

```python

# ---------------------------------------------------------
# 7.2.2 Run Five-Fold Cross-Validation
# ---------------------------------------------------------

from sklearn.model_selection import cross_validate

# Define the three models already trained in Section 6
cv_models = {
    "Baseline": baseline_model,
    "Linear Regression": linear_model,
    "Random Forest": rf_model
}

# Store fold-level scores for all models
cv_fold_results = []

# Evaluate each model using the same cross-validation strategy
for model_name, model in cv_models.items():

    # Perform cross-validation using training observations only
    cv_result = cross_validate(
        estimator=model,
        X=X_train,
        y=y_train,
        cv=cv_strategy,
        scoring=cv_scoring,
        return_train_score=True,
        n_jobs=1,
        error_score="raise"
    )

    # Extract scores from each of the five folds
    for fold_number in range(5):

        cv_fold_results.append({
            "Model": model_name,
            "Fold": fold_number + 1,

            # Convert negative error scores to positive values
            "Train_MAE": -cv_result["train_MAE"][fold_number],
            "Validation_MAE": -cv_result["test_MAE"][fold_number],

            "Train_RMSE": -cv_result["train_RMSE"][fold_number],
            "Validation_RMSE": -cv_result["test_RMSE"][fold_number],

            # R-squared scores are already in the correct direction
            "Train_R2": cv_result["train_R2"][fold_number],
            "Validation_R2": cv_result["test_R2"][fold_number]
        })

    print(f"Completed 5-fold cross-validation: {model_name}")

# Convert the results to a DataFrame
cv_fold_results = pd.DataFrame(cv_fold_results)

# Check that all three models have exactly five folds
assert cv_fold_results.shape[0] == 15, (
    "Expected 15 fold-level results across three models."
)

assert (
    cv_fold_results.groupby("Model")["Fold"].nunique() == 5
).all(), (
    "Each model must have exactly five validation folds."
)

print("\nCross-Validation Completed")
print("-" * 50)

print(f"Models Evaluated: {len(cv_models)}")
print(f"Folds per Model: {cv_strategy.n_splits}")
print(f"Total Fold-Level Results: {len(cv_fold_results)}")
```

Note: This code trains temporary model copies inside cross-validation. It does not replace our already fitted baseline_model, linear_model, or rf_model.

#### Examine the five validation MAE scores

```python

# ---------------------------------------------------------
# 7.2.3 Compare Validation MAE Across Folds
# ---------------------------------------------------------

# Create a table with models as rows and folds as columns
cv_mae_comparison = cv_fold_results.pivot(
    index="Model",
    columns="Fold",
    values="Validation_MAE"
)

# Arrange models in a consistent order
model_order = [
    "Baseline",
    "Linear Regression",
    "Random Forest"
]

cv_mae_comparison = cv_mae_comparison.loc[model_order]

# Rename columns for clarity
cv_mae_comparison.columns = [
    f"Fold_{fold}" for fold in cv_mae_comparison.columns
]

# Display results
print("Five-Fold Validation MAE Comparison")
print("-" * 75)

print(cv_mae_comparison.round(4).to_string())
```

All three models produced relatively consistent results across the five validation folds.

The baseline had the largest validation errors.

Both trained models performed better than the baseline in every fold.

Interestingly, Linear Regression achieved a lower MAE than Random Forest in all five validation folds.

This differs from the training results, where Random Forest performed better.

That difference is useful evidence that the more flexible Random Forest model may be learning some patterns specific to its training observations.

We need to examine the average scores and training-validation gaps before drawing a stronger conclusion.

#### 7.2.2 Cross-Validation Performance and Stability

Following the five-fold cross-validation procedure, I summarized the models' validation performance using the mean and standard deviation of their fold scores.

Mean validation MAE was used as the primary comparison metric.

Mean validation RMSE and R-squared were included to provide additional information about larger prediction errors and overall predictive performance.

The standard deviation of validation MAE was calculated to assess how much performance varied across the five folds.

Lower mean validation errors indicate better average predictive performance, while smaller variations across folds indicate more consistent results under the selected cross-validation procedure.

The findings will be used to identify the leading predictive model before held-out test evaluation.

```python

# ---------------------------------------------------------
# 7.2.4 Summarize Cross-Validation Performance
# ---------------------------------------------------------

# Calculate mean and standard deviation across validation folds
cv_summary = (
    cv_fold_results
    .groupby("Model", as_index=False)
    .agg(
        Mean_CV_MAE=("Validation_MAE", "mean"),
        SD_CV_MAE=("Validation_MAE", "std"),

        Mean_CV_RMSE=("Validation_RMSE", "mean"),
        SD_CV_RMSE=("Validation_RMSE", "std"),

        Mean_CV_R2=("Validation_R2", "mean"),
        SD_CV_R2=("Validation_R2", "std")
    )
)

# Sort according to our predefined primary metric
cv_summary = (
    cv_summary
    .sort_values("Mean_CV_MAE", ascending=True)
    .reset_index(drop=True)
)

# Display results
print("Cross-Validation Performance Summary")
print("-" * 85)

print(cv_summary.round(4).to_string(index=False))
```

Linear Regression achieved the lowest average validation MAE of approximately 10.09 percentage points.

Random Forest achieved approximately 10.33, while the baseline achieved approximately 12.92.

Linear Regression also had the lowest average validation RMSE and highest average validation R².

The standard deviations of validation MAE were relatively small for all models, suggesting that their performance did not fluctuate dramatically across the five folds.

However, five-fold consistency should not be interpreted as proof of stability across different states, future years, or entirely new educational populations.

Our preliminary decision: Linear Regression is currently the preferred candidate because it achieved the lowest mean cross-validation MAE.

We have not yet evaluated the final held-out test set.

**Why did Random Forest perform better during training but worse during cross-validation?**

One possible explanation is overfitting.

We can investigate this by comparing the average training-fold MAE against the average validation-fold MAE.

A large positive difference suggests that a model performs much better on its training observations than on observations excluded from fitting.

#### 7.2.3 Training-Validation Performance Gap

To investigate potential overfitting, I compared the average training-fold MAE with the average validation-fold MAE for each model.

The difference between validation MAE and training MAE was used as an indication of generalization difficulty.

A large positive gap suggests that the model achieves substantially lower errors on observations used during fitting than on observations excluded from training.

This comparison is particularly important for Random Forest because its flexibility can allow it to fit training-specific patterns.

The results will be interpreted as evidence of potential overfitting rather than as a definitive diagnosis based on a single statistic.

```python

# ---------------------------------------------------------
# 7.2.5 Assess Training-Validation Performance Gaps
# ---------------------------------------------------------

# Calculate mean training and validation MAE across folds
cv_overfitting_summary = (
    cv_fold_results
    .groupby("Model", as_index=False)
    .agg(
        Mean_Train_MAE=("Train_MAE", "mean"),
        Mean_Validation_MAE=("Validation_MAE", "mean")
    )
)

# Calculate the validation minus training MAE gap
cv_overfitting_summary["MAE_Gap"] = (
    cv_overfitting_summary["Mean_Validation_MAE"]
    - cv_overfitting_summary["Mean_Train_MAE"]
)

# Calculate the gap as a percentage of training MAE
cv_overfitting_summary["MAE_Gap_Percent"] = (
    cv_overfitting_summary["MAE_Gap"]
    / cv_overfitting_summary["Mean_Train_MAE"]
) * 100

# Arrange models in the same order
cv_overfitting_summary = (
    cv_overfitting_summary
    .set_index("Model")
    .loc[
        [
            "Baseline",
            "Linear Regression",
            "Random Forest"
        ]
    ]
    .reset_index()
)

# Display the results
print("Training-Validation MAE Gap")
print("-" * 75)

print(cv_overfitting_summary.round(4).to_string(index=False))
```

The baseline has almost no training-validation gap.

Linear Regression also has a very small gap, suggesting that its performance generalizes relatively consistently across the validation folds.

Random Forest, however, has a considerably larger gap of approximately 2.40 percentage points.

It predicts training observations much more accurately than validation observations.

This provides evidence of overfitting in the current Random Forest configuration.

Although Random Forest achieved the strongest training results, its additional flexibility did not translate into better validation performance.

That is why using training MAE alone to choose our final model would have been misleading.

#### 7.2.4 Cross-Validation Findings

The five-fold cross-validation assessment evaluated the predictive performance and stability of the baseline, Linear Regression, and Random Forest models using the 3,713 training institutions.

All three models were evaluated using identical folds and the same MAE, RMSE, and R-squared metrics.

**Baseline Model**

The baseline achieved a mean validation MAE of approximately 12.92 percentage points and a mean R-squared slightly below zero.

This indicates that the constant mean prediction strategy provided limited predictive value.

**Linear Regression**

Linear Regression achieved the lowest mean validation MAE of approximately 10.09 percentage points, mean RMSE of 12.52, and mean R-squared of 0.3821.

Its validation MAE standard deviation was approximately 0.35 percentage points.

The difference between mean training-fold and validation-fold MAE was only about 0.01 percentage points, indicating relatively consistent performance across the selected folds.

**Random Forest Regressor**

Random Forest achieved a mean validation MAE of approximately 10.33 percentage points, mean RMSE of 12.85, and mean R-squared of 0.3494.

Although its training errors were substantially lower than those of Linear Regression, its validation performance was weaker.

The average training-validation MAE gap was approximately 2.40 percentage points, indicating potential overfitting under the selected configuration.

**Key Finding**

Linear Regression outperformed Random Forest in all five validation folds based on MAE.

It also achieved the best mean validation MAE, RMSE, and R-squared among the three models.

**Analytical Implication**

The results demonstrate that stronger training performance does not necessarily translate into stronger generalization.

Random Forest's additional flexibility improved training fit but did not improve validation performance in this investigation.

**Preliminary Model Preference**

Based on the predefined primary criterion of mean cross-validation MAE, Linear Regression is the leading candidate.

However, final model evaluation and selection will follow the established held-out test procedure.

The results reflect generalization across randomly selected institutional observations within the supplied dataset, not across independently held-out states or future academic years.

#### 7.2.5 Cross-Validation and Model Stability: Conclusion

I conducted five-fold cross-validation on the training dataset to assess the performance and stability of the three regression models.

Linear Regression achieved the strongest validation performance, with a mean MAE of approximately 10.09 percentage points and mean R-squared of 0.3821.

Random Forest achieved a mean validation MAE of approximately 10.33 percentage points and exhibited a substantially larger training-validation performance gap.

The findings indicate that the current Random Forest configuration may be overfitting the training observations.

The baseline produced the largest validation errors, confirming that the selected institutional characteristics provide useful predictive information beyond a constant mean prediction.

**Conclusion:** Linear Regression is the preferred candidate based on cross-validation MAE, model stability, and interpretability.

### 7.3 Held-Out Test Set Evaluation

The cross-validation analysis identified Linear Regression as the preferred candidate because it achieved the lowest mean validation MAE and demonstrated relatively consistent performance across folds.

I retained that selection decision before examining the held-out test results.

The evaluation compares Linear Regression, Random Forest, and the mean-based baseline using MAE, RMSE, and R-squared.

All three models were evaluated on the same test observations to maintain consistency.

The test results provide an estimate of predictive performance on unseen institutions drawn from the supplied dataset. They do not establish predictive accuracy across future years or entirely new educational populations.

```python

# ---------------------------------------------------------
# 7.3.1 Generate Held-Out Test Predictions
# ---------------------------------------------------------

# Confirm that our test features and target are aligned
assert X_test.index.equals(y_test.index), (
    "Test features and target indexes do not match."
)

# Generate predictions from the previously trained models
baseline_test_predictions = baseline_model.predict(X_test)

linear_test_predictions = linear_model.predict(X_test)

rf_test_predictions = rf_model.predict(X_test)

# Store the predictions in a dictionary
test_predictions = {
    "Baseline": baseline_test_predictions,
    "Linear Regression": linear_test_predictions,
    "Random Forest": rf_test_predictions
}

# Validate the predictions
for model_name, predictions in test_predictions.items():

    assert len(predictions) == len(y_test), (
        f"{model_name}: prediction count does not match the test data."
    )

    assert np.isfinite(predictions).all(), (
        f"{model_name}: invalid numerical predictions detected."
    )

# Display prediction counts
print("Held-Out Test Prediction Summary")
print("-" * 55)

print(f"Test Institutions: {len(y_test)}")

for model_name, predictions in test_predictions.items():
    print(f"{model_name}: {len(predictions)} predictions")

print("\nPrediction Validation: Passed")
```

Each model makes predictions for the same 929 institutions.

#### 7.3.1 Held-Out Test Performance

I evaluated the predictions generated by the three trained models using the 929 held-out institutional observations.

Mean Absolute Error was used as the primary performance metric because it expresses average prediction errors directly in mathematics proficiency percentage points.

Root Mean Squared Error was used to examine larger prediction errors, while R-squared measured performance relative to a mean-based reference on the test observations.

```python

# ---------------------------------------------------------
# 7.3.2 Evaluate All Models on the Held-Out Test Set
# ---------------------------------------------------------

# Create a list for storing evaluation results
test_results_list = []

# Evaluate each model using the same test target
for model_name, predictions in test_predictions.items():

    results = evaluate_regression_model(
        y_true=y_test,
        y_pred=predictions,
        model_name=model_name
    )

    test_results_list.append(results)

# Create a comparison DataFrame
test_results = pd.DataFrame(test_results_list)

# Sort by the predefined primary metric: MAE
test_results = (
    test_results
    .sort_values(by="MAE", ascending=True)
    .reset_index(drop=True)
)

# Display test results
print("Held-Out Test Performance Comparison")
print("-" * 70)

print(test_results.round(4).to_string(index=False))
```

```python

# ---------------------------------------------------------
# 7.3.3 Calculate Improvements Over the Test Baseline
# ---------------------------------------------------------

# Extract model MAE values from the test results table
test_mae_lookup = test_results.set_index("Model")["MAE"]

baseline_test_mae = float(test_mae_lookup["Baseline"])
linear_test_mae = float(test_mae_lookup["Linear Regression"])
rf_test_mae = float(test_mae_lookup["Random Forest"])

# Calculate absolute MAE reductions
linear_mae_reduction = baseline_test_mae - linear_test_mae
rf_mae_reduction = baseline_test_mae - rf_test_mae

# Calculate percentage reductions relative to baseline
linear_mae_reduction_pct = (
    linear_mae_reduction / baseline_test_mae
) * 100

rf_mae_reduction_pct = (
    rf_mae_reduction / baseline_test_mae
) * 100

# Display improvements
print("Held-Out Test MAE Improvement Over Baseline")
print("-" * 60)

print(
    f"Linear Regression MAE Reduction: "
    f"{linear_mae_reduction:.4f} percentage points"
)

print(
    f"Linear Regression Relative Reduction: "
    f"{linear_mae_reduction_pct:.2f}%"
)

print(
    f"\nRandom Forest MAE Reduction: "
    f"{rf_mae_reduction:.4f} percentage points"
)

print(
    f"Random Forest Relative Reduction: "
    f"{rf_mae_reduction_pct:.2f}%"
)
```

### 7.4 Model Comparison and Overfitting Assessment

Following cross-validation and held-out test evaluation, I compared the performance of the three regression models across training, validation, and testing.

The objective was to determine whether improvements observed during training translated into reliable predictions on unseen institutions.

The comparison focused on the baseline Dummy Regressor, Linear Regression, and Random Forest Regressor.

I examined Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and R-squared (R²), with particular attention to differences between training and validation performance.

I also examined the performance gap between training and held-out testing to identify potential generalization difficulties.

The analysis aims to distinguish genuine predictive improvement from apparent improvements caused by fitting the training observations too closely.

The model preference established through cross-validation will remain unchanged during this assessment. Test-set results will be used to evaluate that decision rather than to introduce additional model tuning.

#### Consolidate training, validation, and test performance

```python

# ---------------------------------------------------------
# 7.4.1 Consolidate Training, Validation, and Test Results
# ---------------------------------------------------------

# Select and rename the training metrics
train_comparison = training_results[
    ["Model", "Train_MAE", "Train_RMSE", "Train_R2"]
].copy()

# Select the cross-validation summary metrics
cv_comparison = cv_summary[
    [
        "Model",
        "Mean_CV_MAE",
        "SD_CV_MAE",
        "Mean_CV_RMSE",
        "Mean_CV_R2"
    ]
].copy()

# Rename the held-out test metrics
test_comparison = test_results.rename(
    columns={
        "MAE": "Test_MAE",
        "RMSE": "Test_RMSE",
        "R2": "Test_R2"
    }
)

# Combine the three sets of evaluation results
model_comparison = (
    train_comparison
    .merge(cv_comparison, on="Model", validate="one_to_one")
    .merge(test_comparison, on="Model", validate="one_to_one")
)

# Define a consistent display order
model_order = [
    "Baseline",
    "Linear Regression",
    "Random Forest"
]

model_comparison = (
    model_comparison
    .set_index("Model")
    .loc[model_order]
    .reset_index()
)

# Confirm that exactly three models are present
assert len(model_comparison) == 3, (
    "Expected evaluation results for exactly three models."
)

# Display the consolidated results
print("Consolidated Model Performance")
print("-" * 105)

print(model_comparison.round(4).to_string(index=False))
```

Random Forest achieved the lowest training MAE and highest training R².

However, Linear Regression performed better during cross-validation and on the held-out test set.

This is an important finding because it shows that a more flexible model does not necessarily provide more accurate predictions on unseen data.

#### 7.4.1 Training-to-Test Performance Gap

I calculated the difference between each model's held-out test MAE and training MAE.

The comparison measures how much average absolute prediction error increased when the fitted model was evaluated on institutions excluded from training.

A relatively small gap suggests that training and test errors are similar.

A larger positive gap indicates that the model performs more accurately on its training observations than on the held-out test observations.

This difference may indicate overfitting, although sampling variation and differences between the datasets may also contribute.

The results will be interpreted alongside the cross-validation findings to assess the strength of the evidence for overfitting.

#### Calculate training-to-test performance gaps

```python

# ---------------------------------------------------------
# 7.4.2 Calculate Training-to-Test MAE Gaps
# ---------------------------------------------------------

# Create a separate table for generalization gaps
generalization_summary = model_comparison[
    [
        "Model",
        "Train_MAE",
        "Mean_CV_MAE",
        "Test_MAE"
    ]
].copy()

# Calculate the training-to-test MAE gap
generalization_summary["Train_Test_MAE_Gap"] = (
    generalization_summary["Test_MAE"]
    - generalization_summary["Train_MAE"]
)

# Calculate the training-to-validation MAE gap
# These values use the full-training fit and the mean CV score
generalization_summary["Train_CV_MAE_Gap"] = (
    generalization_summary["Mean_CV_MAE"]
    - generalization_summary["Train_MAE"]
)

# Calculate validation-to-test MAE difference
generalization_summary["CV_Test_MAE_Gap"] = (
    generalization_summary["Test_MAE"]
    - generalization_summary["Mean_CV_MAE"]
)

# Display results
print("Model Generalization Gap Comparison")
print("-" * 95)

print(
    generalization_summary
    .round(4)
    .to_string(index=False)
)
```

Baseline: Test MAE was approximately 0.43 percentage points higher than training MAE.

Linear Regression: Test MAE was approximately 0.94 percentage points higher than training MAE.

Random Forest: Test MAE was approximately 3.28 percentage points higher than training MAE.

Random Forest exhibited the largest training-to-test gap.

This is consistent with our earlier cross-validation finding that Random Forest fits training observations substantially better than validation observations.

However, the test MAE of all three models was somewhat higher than the mean cross-validation MAE.

This suggests that the held-out test institutions were, on average, somewhat harder to predict than the validation folds, although the precise reason cannot be established from these metrics alone.

Important distinction: The training-to-test gap compares a model fitted on all training observations with its test performance. The earlier fold-level gap compares training and validation performance within cross-validation. Both are useful, but they measure slightly different things.

#### 7.4.2 Visual Comparison of Model Generalization

I visualized training, cross-validation, and held-out test MAE for the three regression models.

The objective was to identify differences between in-sample fit and performance on observations excluded from model fitting.

A model whose training MAE is substantially lower than its validation and test MAE may be overfitting the training observations.

The comparison also shows whether the models provide consistent improvements over the baseline across different evaluation stages.

#### Compare training, validation, and test MAE visually

```python

# ---------------------------------------------------------
# 7.4.3 Visualize Training, CV, and Test MAE
# ---------------------------------------------------------

# Define the model order
model_names = model_comparison["Model"].tolist()

# Create positions for grouped bars
positions = np.arange(len(model_names))
bar_width = 0.25

# Create the chart
fig, ax = plt.subplots(figsize=(11, 6))

# Training MAE bars
train_bars = ax.bar(
    positions - bar_width,
    model_comparison["Train_MAE"],
    width=bar_width,
    label="Training MAE",
    color="steelblue"
)

# Cross-validation MAE bars
cv_bars = ax.bar(
    positions,
    model_comparison["Mean_CV_MAE"],
    width=bar_width,
    label="Mean CV MAE",
    color="darkorange"
)

# Held-out test MAE bars
test_bars = ax.bar(
    positions + bar_width,
    model_comparison["Test_MAE"],
    width=bar_width,
    label="Test MAE",
    color="seagreen"
)

# Add values above the bars
ax.bar_label(train_bars, fmt="%.2f", padding=3)
ax.bar_label(cv_bars, fmt="%.2f", padding=3)
ax.bar_label(test_bars, fmt="%.2f", padding=3)

# Add title and labels
ax.set_title(
    "Model Performance Across Training, Validation, and Testing",
    fontsize=14
)

ax.set_xlabel("Regression Model")
ax.set_ylabel("Mean Absolute Error (Percentage Points)")

ax.set_xticks(positions)
ax.set_xticklabels(model_names)

# Add a legend
ax.legend()

# Leave space above the bars
ax.set_ylim(
    0,
    model_comparison[
        ["Train_MAE", "Mean_CV_MAE", "Test_MAE"]
    ].to_numpy().max() * 1.15
)

# Improve chart layout
plt.tight_layout()

# Display the chart
plt.show()
```

Random Forest bars; Its training MAE is considerably lower than its validation and test MAE.

Linear Regression has much more similar training and validation scores, although its test error is somewhat higher.

#### 7.4.3 Cross-Validation Overfitting Assessment

I examined the differences between mean training-fold MAE and mean validation-fold MAE to assess potential overfitting.

The assessment used the five-fold cross-validation results calculated exclusively from the training dataset.

Linear Regression demonstrated a small difference between training and validation MAE.

Random Forest demonstrated a substantially larger difference, indicating that its strong training performance did not transfer equally to the validation folds.

These findings provide stronger evidence of potential overfitting than relying only on the original training-to-test gap.

The results will inform the final model recommendation, together with predictive accuracy and interpretability.

```python

# ---------------------------------------------------------
# 7.4.4 Summarize Cross-Validation Overfitting Evidence
# ---------------------------------------------------------

# Calculate fold-level training and validation means
overfitting_results = (
    cv_fold_results
    .groupby("Model", as_index=False)
    .agg(
        Mean_Fold_Train_MAE=("Train_MAE", "mean"),
        Mean_Fold_Validation_MAE=("Validation_MAE", "mean"),
        SD_Fold_Validation_MAE=("Validation_MAE", "std")
    )
)

# Calculate the average training-to-validation gap
overfitting_results["CV_MAE_Gap"] = (
    overfitting_results["Mean_Fold_Validation_MAE"]
    - overfitting_results["Mean_Fold_Train_MAE"]
)

# Arrange the models in a consistent order
overfitting_results = (
    overfitting_results
    .set_index("Model")
    .loc[
        [
            "Baseline",
            "Linear Regression",
            "Random Forest"
        ]
    ]
    .reset_index()
)

# Display results
print("Five-Fold Training-Validation Gap Assessment")
print("-" * 85)

print(overfitting_results.round(4).to_string(index=False))
```

Linear Regression's gap is approximately 0.01 percentage points, indicating that its training and validation errors are very similar.

Random Forest's gap is approximately 2.40 percentage points, indicating a more substantial deterioration when evaluated on observations not used in fitting.

Taken together with the test results, this provides consistent evidence of overfitting in the current Random Forest configuration.

This does not mean Random Forest is generally an unsuitable algorithm. Different features, hyperparameters, or training data could produce different results.

We should also avoid tuning Random Forest now based on the observed test scores. A new tuning investigation would require a properly defined training-only selection procedure, and the already inspected test set would no longer be entirely untouched by subsequent development decisions.

The objective of this subsection is to consolidate the evaluation evidence, document the preferred model, and answer the predictive component of the original research question.

#### Create our final model evaluation table

```python

# ---------------------------------------------------------
# 7.6.1 Consolidate Final Model Evaluation Results
# ---------------------------------------------------------

# Extract training performance
final_train_metrics = training_results[
    ["Model", "Train_MAE", "Train_RMSE", "Train_R2"]
].copy()

# Extract cross-validation performance
final_cv_metrics = cv_summary[
    [
        "Model",
        "Mean_CV_MAE",
        "SD_CV_MAE",
        "Mean_CV_RMSE",
        "Mean_CV_R2"
    ]
].copy()

# Extract held-out test performance
final_test_metrics = test_results[
    ["Model", "MAE", "RMSE", "R2"]
].rename(
    columns={
        "MAE": "Test_MAE",
        "RMSE": "Test_RMSE",
        "R2": "Test_R2"
    }
)

# Combine evaluation results
final_evaluation_summary = (
    final_train_metrics
    .merge(
        final_cv_metrics,
        on="Model",
        validate="one_to_one"
    )
    .merge(
        final_test_metrics,
        on="Model",
        validate="one_to_one"
    )
)

# Calculate test-minus-training MAE gaps
final_evaluation_summary["Train_Test_MAE_Gap"] = (
    final_evaluation_summary["Test_MAE"]
    - final_evaluation_summary["Train_MAE"]
)

# Rank models using our predefined selection criterion
final_evaluation_summary = (
    final_evaluation_summary
    .sort_values(
        by="Mean_CV_MAE",
        ascending=True
    )
    .reset_index(drop=True)
)

final_evaluation_summary.insert(
    0,
    "CV_Rank",
    range(1, len(final_evaluation_summary) + 1)
)

# Check that the expected models are included
expected_models = {
    "Baseline",
    "Linear Regression",
    "Random Forest"
}

assert set(final_evaluation_summary["Model"]) == expected_models, (
    "The final comparison is missing one or more models."
)

# Confirm that our selected model remains consistent
assert final_evaluation_summary.loc[0, "Model"] == selected_model_name, (
    "Final model selection does not match cross-validation ranking."
)

# Display the complete comparison
print("Final Model Evaluation Summary")
print("-" * 115)

print(
    final_evaluation_summary
    .round(4)
    .to_string(index=False)
)
```

Linear Regression ranked first using the predefined cross-validation MAE criterion.

Its held-out test performance supported that selection.

Random Forest achieved a better fit to the training observations, but its validation and test errors were larger.

The baseline performed worst, confirming that the four institutional predictors provided useful information for predicting mathematics proficiency.

#### Quantify the predictive improvement

```python

# ---------------------------------------------------------
# 7.6.2 Quantify Final Predictive Improvement
# ---------------------------------------------------------

# Create a convenient lookup table
model_metrics_lookup = final_evaluation_summary.set_index("Model")

# Extract held-out test MAE values
baseline_mae = float(
    model_metrics_lookup.loc["Baseline", "Test_MAE"]
)

selected_mae = float(
    model_metrics_lookup.loc[selected_model_name, "Test_MAE"]
)

# Calculate absolute MAE improvement
absolute_mae_improvement = baseline_mae - selected_mae

# Calculate percentage improvement relative to the baseline
percentage_mae_improvement = (
    absolute_mae_improvement / baseline_mae
) * 100

# Extract additional selected-model test metrics
selected_test_rmse = float(
    model_metrics_lookup.loc[selected_model_name, "Test_RMSE"]
)

selected_test_r2 = float(
    model_metrics_lookup.loc[selected_model_name, "Test_R2"]
)

# Display the findings
print("Final Predictive Performance")
print("-" * 60)

print(f"Selected Model: {selected_model_name}")
print(f"Test Institutions: {len(y_test)}")

print(f"\nBaseline Test MAE: {baseline_mae:.4f} pp")
print(f"Selected Model Test MAE: {selected_mae:.4f} pp")

print(
    f"\nAbsolute MAE Reduction: "
    f"{absolute_mae_improvement:.4f} pp"
)

print(
    f"Relative MAE Reduction: "
    f"{percentage_mae_improvement:.2f}%"
)

print(f"\nSelected Model Test RMSE: {selected_test_rmse:.4f} pp")
print(f"Selected Model Test R-squared: {selected_test_r2:.4f}")
```

Our selected Linear Regression model reduced average absolute prediction error by approximately 2.33 percentage points, or 17.45%, compared with the baseline.

This demonstrates measurable predictive value from the four institutional characteristics.

However, the model still has an average absolute prediction error of approximately 11 percentage points.

Therefore, we should not claim that it provides highly precise institution-level predictions.

```python

# ---------------------------------------------------------
# 8.2.1 Extract the Selected Model's Fitted Coefficients
# ---------------------------------------------------------


# Extract the intercept from the selected model
model_intercept = float(interpretation_model.intercept_)

# Create a table of fitted coefficients
coefficient_results = pd.DataFrame({
    "Feature": interpretation_features,
    "Coefficient": interpretation_model.coef_
})

# Add the direction of each relationship
coefficient_results["Direction"] = (
    coefficient_results["Coefficient"].apply(
        lambda value: (
            "Positive" if value > 0
            else "Negative" if value < 0
            else "Zero"
        )
    )
)

# Add readable predictor names
feature_labels = {
    "per_pupil_expenditure_usd": "Per-Pupil Expenditure",
    "student_teacher_ratio": "Student-Teacher Ratio",
    "percent_free_reduced_lunch": "Free/Reduced Lunch Eligibility",
    "broadband_access_pct": "Broadband Access"
}

coefficient_results["Predictor"] = (
    coefficient_results["Feature"].map(feature_labels)
)

# Arrange columns
coefficient_results = coefficient_results[
    ["Predictor", "Feature", "Coefficient", "Direction"]
]

# Validate the number of coefficients
assert len(coefficient_results) == len(interpretation_features), (
    "The number of coefficients does not match the predictors."
)

# Display the fitted intercept
print("Selected Linear Regression Model")
print("-" * 65)

print(f"Intercept: {model_intercept:.6f}")

# Display the coefficients
print("\nFitted Regression Coefficients")
print("-" * 65)

print(coefficient_results.round(6).to_string(index=False))
```

### 8.3 Feature Importance and Contribution Analysis

Following the interpretation of the Linear Regression coefficients, I investigated the relative contributions of the four institutional predictors to the selected model's predictions.

The model includes per-pupil expenditure, student-teacher ratio, free or reduced-price lunch eligibility, and broadband access.

Because these predictors have different measurement units, their raw regression coefficients cannot be directly compared to determine relative importance.

I therefore used standardized coefficients and permutation feature importance to examine their contributions from two complementary perspectives.

Standardized coefficients describe fitted relationships after accounting for differences in predictor scales.

Permutation importance measures changes in prediction error when the values of an individual predictor are randomly shuffled.

I also examined correlations among the predictors to identify relationships that could affect the interpretation of feature importance.

The analysis used the existing fitted Linear Regression model and the training observations. The held-out test data were not used to calculate the feature importance rankings.

The objective is to identify the predictors that contribute most strongly to the fitted model while avoiding unsupported causal interpretations.

#### 8.3.1 Standardized Coefficient Analysis

I calculated standardized regression coefficients to compare the fitted contributions of the institutional predictors on a common scale.

The calculation used the original fitted coefficients and the standard deviations of the training predictors and target variable.

Each standardized coefficient represents the expected change in predicted mathematics proficiency, measured in target standard deviations, associated with a one-standard-deviation increase in the corresponding predictor, holding the other predictors constant.

This approach accounts for differences in the original measurement scales of expenditure, staffing, lunch eligibility, and broadband access.

The absolute values of standardized coefficients were used to examine relative magnitudes, while their signs retained the directions of the fitted relationships.

The results were interpreted as adjusted model relationships rather than causal effects.

Because the predictors may be correlated, standardized coefficient magnitudes were not treated as definitive measures of independent predictive importance.

```python

# ---------------------------------------------------------
# 8.3.1 Calculate Standardized Regression Coefficients
# ---------------------------------------------------------


# Calculate predictor standard deviations using training data
predictor_std = X_interpretation_train[
    interpretation_features
].std(ddof=0)

# Calculate the target standard deviation
target_std = y_interpretation_train.std(ddof=0)

# Extract coefficients from the selected fitted model
raw_coefficients = interpretation_model.coef_

# Calculate standardized coefficients
standardized_coefficients = (
    raw_coefficients
    * predictor_std.to_numpy()
    / target_std
)

# Organize the results
standardized_coefficient_results = pd.DataFrame({
    "Feature": interpretation_features,
    "Coefficient": raw_coefficients,
    "Predictor_SD": predictor_std.to_numpy(),
    "Standardized_Coefficient": standardized_coefficients
})

# Calculate absolute standardized coefficients
standardized_coefficient_results["Absolute_Standardized_Coefficient"] = (
    standardized_coefficient_results[
        "Standardized_Coefficient"
    ].abs()
)

# Add readable predictor names
standardized_coefficient_results["Predictor"] = (
    standardized_coefficient_results["Feature"].map(
        feature_labels
    )
)

# Sort by absolute standardized coefficient
standardized_coefficient_results = (
    standardized_coefficient_results
    .sort_values(
        by="Absolute_Standardized_Coefficient",
        ascending=False
    )
    .reset_index(drop=True)
)

# Display results
print("Standardized Linear Regression Coefficients")
print("-" * 85)

print(
    standardized_coefficient_results[
        [
            "Predictor",
            "Coefficient",
            "Standardized_Coefficient",
            "Absolute_Standardized_Coefficient"
        ]
    ].round(6).to_string(index=False)
)
```

Free or reduced-price lunch eligibility has the largest absolute standardized coefficient.

A one-standard-deviation increase in lunch eligibility corresponds to approximately 0.561 standard deviations lower predicted mathematics proficiency, holding the other predictors constant.

Per-pupil expenditure has the second-largest absolute standardized coefficient, approximately 0.161.

Student-teacher ratio and broadband access have much smaller standardized coefficients.

This suggests that lunch eligibility and expenditure contribute more strongly to the fitted linear relationships under the current model specification.

However, standardized coefficients do not tell us directly how much prediction error increases when a feature's information is disrupted.

#### Visualize standardized coefficients

```python

# ---------------------------------------------------------
# 8.3.2 Visualize Standardized Coefficients
# ---------------------------------------------------------

# Arrange results for horizontal plotting
standardized_plot_data = (
    standardized_coefficient_results
    .sort_values("Standardized_Coefficient")
)

# Set colors according to coefficient direction
bar_colors = [
    "steelblue" if value >= 0 else "darkorange"
    for value in standardized_plot_data[
        "Standardized_Coefficient"
    ]
]

# Create the chart
fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.barh(
    standardized_plot_data["Predictor"],
    standardized_plot_data["Standardized_Coefficient"],
    color=bar_colors
)

# Add a zero-reference line
ax.axvline(
    x=0,
    color="black",
    linewidth=1
)

# Add coefficient values beside bars
ax.bar_label(
    bars,
    fmt="%.4f",
    padding=4
)

# Add title and labels
ax.set_title(
    "Standardized Coefficients: Selected Linear Regression Model",
    fontsize=14
)

ax.set_xlabel("Standardized Regression Coefficient")
ax.set_ylabel("Institutional Predictor")

# Allow space for labels
ax.margins(x=0.2)

plt.tight_layout()
plt.show()
```

Lunch eligibility has the largest magnitude and a negative fitted relationship.

Per-pupil expenditure has the second-largest magnitude and a positive fitted relationship.

Student-teacher ratio and broadband access have comparatively small adjusted coefficients.

This is a model-specific finding, not evidence that the latter two characteristics are unimportant to education generally.

#### 8.3.2 Permutation Feature Importance

I used permutation feature importance to examine how strongly the selected Linear Regression model relies on each institutional predictor.

The method measures the change in prediction error when one predictor is randomly shuffled while the remaining predictors and fitted model are unchanged.

Mean Absolute Error was selected as the permutation importance metric to maintain consistency with the primary model evaluation criterion.

The analysis was performed using the training observations, with 20 random permutations per predictor and a fixed random state of 42.

For each feature, importance was measured as the increase in MAE caused by shuffling its values.

A larger positive importance indicates that disrupting the predictor leads to a greater increase in prediction error.

Values close to zero indicate that the fitted model's MAE changes little under the permutation procedure.

Negative values may occur due to random variation when shuffling slightly improves the measured prediction error.

Permutation importance measures model reliance under the selected dataset and scoring method. It does not establish statistical significance, causal influence, or independent feature importance when predictors are correlated.

```python

# ---------------------------------------------------------
# 8.3.3 Calculate Permutation Feature Importance
# ---------------------------------------------------------

from sklearn.inspection import permutation_importance

# Calculate training-based permutation importance
permutation_result = permutation_importance(
    estimator=interpretation_model,
    X=X_interpretation_train,
    y=y_interpretation_train,
    scoring="neg_mean_absolute_error",
    n_repeats=20,
    random_state=42,
    n_jobs=1
)

# Organize permutation importance results
permutation_importance_results = pd.DataFrame({
    "Feature": interpretation_features,
    "Mean_MAE_Increase": permutation_result.importances_mean,
    "SD_MAE_Increase": permutation_result.importances_std
})

# Add readable labels
permutation_importance_results["Predictor"] = (
    permutation_importance_results["Feature"].map(
        feature_labels
    )
)

# Sort by mean increase in MAE
permutation_importance_results = (
    permutation_importance_results
    .sort_values(
        by="Mean_MAE_Increase",
        ascending=False
    )
    .reset_index(drop=True)
)

# Display the results
print("Training-Based Permutation Feature Importance")
print("-" * 85)

print(
    permutation_importance_results[
        [
            "Predictor",
            "Mean_MAE_Increase",
            "SD_MAE_Increase"
        ]
    ].round(6).to_string(index=False)
)
```

Free or reduced-price lunch eligibility produced the largest increase in training MAE when shuffled.

Its importance was approximately 4.20 percentage points.

This means disrupting the lunch eligibility information increased the selected model's training MAE by approximately 4.20 percentage points, on average, across the 20 permutations.

Per-pupil expenditure produced a smaller but positive increase of approximately 0.36 percentage points.

Broadband access and student-teacher ratio produced changes close to zero.

Student-teacher ratio's slightly negative importance does not establish that the predictor is harmful. The result is tiny compared with the variation across permutations.

Finding: Under this training-based permutation procedure, the fitted Linear Regression model relies most strongly on lunch eligibility, followed by expenditure.

We should be careful not to confuse this with the amount of unique information each feature provides in an entirely new population.

#### Visualize permutation importance

```python

# ---------------------------------------------------------
# 8.3.4 Visualize Permutation Feature Importance
# ---------------------------------------------------------


# Sort results for horizontal plotting
permutation_plot_data = (
    permutation_importance_results
    .sort_values("Mean_MAE_Increase")
)

# Create the chart
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the mean MAE increase with error bars
bars = ax.barh(
    permutation_plot_data["Predictor"],
    permutation_plot_data["Mean_MAE_Increase"],
    xerr=permutation_plot_data["SD_MAE_Increase"],
    color="steelblue",
    capsize=4
)

# Add the zero-reference line
ax.axvline(
    x=0,
    color="black",
    linewidth=1
)

# Add chart labels
ax.set_title(
    "Permutation Feature Importance: Linear Regression",
    fontsize=14
)

ax.set_xlabel(
    "Mean Increase in Training MAE (Percentage Points)"
)

ax.set_ylabel("Institutional Predictor")

# Improve layout
plt.tight_layout()

# Display chart
plt.show()
```

Lunch eligibility dominates the permutation importance results for the current fitted model.

Expenditure ranks second, while the other two predictors show negligible changes in training MAE when shuffled individually.

The chart supports the standardized coefficient findings, but the two methods do not measure exactly the same thing.


#### 8.3.3 Predictor Correlation and Importance Limitations

I examined correlations among the four institutional predictors to understand whether overlapping information could affect the feature importance results.

Correlated predictors can complicate the interpretation of both standardized regression coefficients and permutation importance.

When predictors are strongly related, their fitted coefficients may be sensitive to the other variables included in the model.

Similarly, shuffling one predictor can disrupt relationships between variables and create combinations that are uncommon in the original data.

I therefore examined the training predictor correlation matrix before drawing conclusions about relative importance.

The objective was to identify potentially important dependencies and incorporate them into the interpretation of the feature contribution findings.

```python

# ---------------------------------------------------------
# 8.3.5 Examine Correlations Between Predictors
# ---------------------------------------------------------


# Calculate Pearson correlations on training predictors
predictor_correlation_matrix = (
    X_interpretation_train[
        interpretation_features
    ].corr()
)

# Replace technical labels with readable names
predictor_correlation_display = (
    predictor_correlation_matrix
    .rename(
        index=feature_labels,
        columns=feature_labels
    )
)

# Display the correlation matrix
print("Training Predictor Correlations")
print("-" * 75)

print(
    predictor_correlation_display
    .round(3)
    .to_string()
)

# Create the correlation heatmap
plt.figure(figsize=(9, 7))

sns.heatmap(
    predictor_correlation_display,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=0.5
)

plt.title(
    "Correlation Among Institutional Predictors",
    fontsize=14
)

plt.xticks(rotation=35, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

#### Compare our two importance methods

### 8.4 Prediction Error Analysis

Following the feature contribution analysis, I examined the prediction errors produced by the selected Linear Regression model.

Although the model achieved a held-out test MAE of approximately 11.01 percentage points, this overall metric does not fully describe its performance across individual institutions.

The objective of this subsection is to investigate the distribution, magnitude, and direction of prediction errors.

The analysis focuses on four areas:

1. Overall prediction error distribution.
2. Overestimation and underestimation patterns.
3. Institutions with the largest absolute prediction errors.
4. Prediction errors across different observed mathematics proficiency levels.

The investigation uses the 929 held-out test observations and predictions generated by the selected fitted model.

No additional training, hyperparameter tuning, or model selection will be performed using these observations.

The findings will help identify limitations of the selected model and support a more responsible interpretation of its predictive usefulness.

```python

# Check whether the required variables exist

required_variables = [
    "edu_clean",
    "X_test",
    "y_test",
    "selected_model"
]

for variable in required_variables:
    if variable in globals():
        print(f"{variable}: Available")
    else:
        print(f"{variable}: Missing")
```

#### Prepare the prediction error dataset

```python

# ---------------------------------------------------------
# 8.4.1 Prepare the Prediction Error Dataset
# ---------------------------------------------------------

# Generate predictions using the selected model
interpretation_test_predictions = selected_model.predict(X_test)

# Retrieve institution identifiers corresponding to test rows
test_institution_ids = edu_clean.loc[
    X_test.index,
    "institution_id"
]

# Create the prediction error DataFrame
prediction_error_data = pd.DataFrame({
    "Institution_ID": test_institution_ids.to_numpy(),
    "Actual_Math_Proficiency": y_test.to_numpy(),
    "Predicted_Math_Proficiency": interpretation_test_predictions
})

# Calculate signed prediction errors
prediction_error_data["Prediction_Error"] = (
    prediction_error_data["Actual_Math_Proficiency"]
    - prediction_error_data["Predicted_Math_Proficiency"]
)

# Calculate absolute errors
prediction_error_data["Absolute_Error"] = (
    prediction_error_data["Prediction_Error"].abs()
)

# Calculate squared errors
prediction_error_data["Squared_Error"] = (
    prediction_error_data["Prediction_Error"] ** 2
)

# Identify prediction error directions
prediction_error_data["Error_Direction"] = np.select(
    [
        prediction_error_data["Prediction_Error"] > 0,
        prediction_error_data["Prediction_Error"] < 0
    ],
    [
        "Underestimation",
        "Overestimation"
    ],
    default="Exact Prediction"
)

# Validate the observation count
assert len(prediction_error_data) == len(y_test), (
    "Prediction count does not match the test dataset."
)

# Display results
print("Prediction Error Dataset")
print("-" * 65)

print(f"Number of Institutions: {len(prediction_error_data)}")

print("\nFirst Five Observations:")

print(
    prediction_error_data[
        [
            "Institution_ID",
            "Actual_Math_Proficiency",
            "Predicted_Math_Proficiency",
            "Prediction_Error",
            "Absolute_Error"
        ]
    ].head().round(2).to_string(index=False)
)
```

#### 8.4.1 Descriptive Analysis of Prediction Errors

I calculated descriptive statistics for the prediction errors generated by the selected Linear Regression model.

The analysis included mean signed error, median signed error, standard deviation, Mean Absolute Error, Root Mean Squared Error, and selected absolute error percentiles.

Mean signed error was used to investigate the model's overall tendency toward underestimation or overestimation.

Absolute error statistics were used to examine the magnitude of prediction differences without allowing positive and negative errors to cancel each other.

The analysis was conducted on all 929 held-out test observations.

These statistics provide a more detailed understanding of predictive reliability than the overall evaluation metrics alone.

#### Calculate descriptive error statistics

```python

# ---------------------------------------------------------
# 8.4.2 Calculate Prediction Error Summary Statistics
# ---------------------------------------------------------


from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

# Extract prediction errors
signed_errors = prediction_error_data["Prediction_Error"]
absolute_errors = prediction_error_data["Absolute_Error"]

# Calculate summary statistics
error_summary = pd.Series({
    "Number of Institutions": len(prediction_error_data),
    "Mean Signed Error": signed_errors.mean(),
    "Median Signed Error": signed_errors.median(),
    "Error Standard Deviation": signed_errors.std(),
    "Mean Absolute Error": absolute_errors.mean(),
    "Median Absolute Error": absolute_errors.median(),
    "Root Mean Squared Error": np.sqrt(
        mean_squared_error(
            prediction_error_data["Actual_Math_Proficiency"],
            prediction_error_data["Predicted_Math_Proficiency"]
        )
    ),
    "75th Percentile Absolute Error": absolute_errors.quantile(0.75),
    "90th Percentile Absolute Error": absolute_errors.quantile(0.90),
    "95th Percentile Absolute Error": absolute_errors.quantile(0.95),
    "Maximum Absolute Error": absolute_errors.max()
})

# Calculate MAE independently for validation
calculated_test_mae = mean_absolute_error(
    prediction_error_data["Actual_Math_Proficiency"],
    prediction_error_data["Predicted_Math_Proficiency"]
)

# Confirm the MAE calculation is consistent
assert np.isclose(
    error_summary["Mean Absolute Error"],
    calculated_test_mae,
    atol=1e-8
), "The prediction error MAE calculation is inconsistent."

# Display results
print("Held-Out Prediction Error Statistics")
print("-" * 65)

print(error_summary.round(4).to_string())
```

The mean signed error is approximately +0.09 percentage points, which is close to zero.

This suggests that the model does not exhibit substantial overall signed bias across the held-out observations.

However, errors can cancel each other out. A near-zero average signed error does not mean the model predicts every institution accurately.

The 90th percentile absolute error is approximately 22.21 percentage points, meaning 90% of the held-out observations have absolute errors at or below that value, while approximately 10% have larger errors.

The maximum error is nearly 44 percentage points.

Finding: The model's average performance conceals some substantial institution-level prediction errors.

#### Visualize the distribution of prediction errors

```python

# ---------------------------------------------------------
# 8.4.3 Visualize the Prediction Error Distribution
# ---------------------------------------------------------

# Create the histogram
fig, ax = plt.subplots(figsize=(10, 6))

sns.histplot(
    data=prediction_error_data,
    x="Prediction_Error",
    bins=30,
    kde=True,
    color="steelblue",
    ax=ax
)

# Add the zero-error reference line
ax.axvline(
    x=0,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Zero Prediction Error"
)

# Add the mean error reference line
ax.axvline(
    x=signed_errors.mean(),
    color="darkgreen",
    linestyle=":",
    linewidth=2,
    label=f"Mean Error = {signed_errors.mean():.2f} pp"
)

# Add chart information
ax.set_title(
    "Distribution of Linear Regression Test Prediction Errors",
    fontsize=14
)

ax.set_xlabel(
    "Prediction Error: Actual - Predicted (Percentage Points)"
)

ax.set_ylabel("Number of Institutions")

ax.legend()

plt.tight_layout()
plt.show()
```

### 8.4.2 Overestimation and Underestimation Analysis

I examined the direction of prediction errors to determine how frequently the selected Linear Regression model overestimated or underestimated mathematics proficiency.

A positive signed error indicates that actual mathematics proficiency was higher than predicted, representing underestimation.

A negative signed error indicates that predicted proficiency was higher than the actual outcome, representing overestimation.

I calculated the number and percentage of institutions in each category.

```python

# ---------------------------------------------------------
# 8.4.4 Examine Overestimation and Underestimation
# ---------------------------------------------------------

# Count error directions
error_direction_summary = (
    prediction_error_data["Error_Direction"]
    .value_counts()
    .rename_axis("Error_Direction")
    .reset_index(name="Institution_Count")
)

# Calculate percentages
error_direction_summary["Percentage"] = (
    error_direction_summary["Institution_Count"]
    / len(prediction_error_data)
) * 100

# Display the summary
print("Prediction Error Direction Summary")
print("-" * 65)

print(
    error_direction_summary
    .round(2)
    .to_string(index=False)
)

print(
    f"\nMean Signed Error: "
    f"{signed_errors.mean():+.4f} percentage points"
)
```

The model underestimated slightly more institutions than it overestimated.

Finding: We do not observe a substantial overall tendency to consistently overestimate or underestimate mathematics proficiency across all 929 test institutions.

#### 8.4.3 Institutions with the Largest Prediction Errors

I identified the ten held-out institutions with the largest absolute prediction errors.

The analysis compared actual mathematics proficiency with the fitted model's predictions and ranked observations according to absolute error magnitude.

Institutional identifiers were retained to make the observations traceable within the analytical dataset.

The purpose was to assess the extent of the model's largest prediction discrepancies and identify whether errors involved substantial underestimation or overestimation.

These observations were not automatically treated as data errors or removed from the analysis.

#### Examine the frequency of large errors

```python

# ---------------------------------------------------------
# 8.4.6 Examine Large Prediction Error Frequencies
# ---------------------------------------------------------

# Define descriptive absolute error thresholds
error_thresholds = [5, 10, 15, 20, 25]

# Calculate how many institutions exceed each threshold
error_threshold_results = []

for threshold in error_thresholds:

    error_count = int(
        (prediction_error_data["Absolute_Error"] > threshold).sum()
    )

    error_percentage = (
        error_count / len(prediction_error_data)
    ) * 100

    error_threshold_results.append({
        "Absolute_Error_Threshold_pp": threshold,
        "Institutions_Above_Threshold": error_count,
        "Percentage_Above_Threshold": error_percentage
    })

# Create the summary table
error_threshold_summary = pd.DataFrame(
    error_threshold_results
)

print("Large Prediction Error Frequencies")
print("-" * 75)

print(
    error_threshold_summary
    .round(2)
    .to_string(index=False)
)
```

Almost half the test institutions had absolute prediction errors greater than 10 percentage points.

#### 8.4.4 Prediction Errors Across Mathematics Proficiency Levels

I investigated whether prediction errors differed across institutions with different observed mathematics proficiency levels.

The held-out institutions were grouped into four descriptive categories based on their actual mathematics proficiency: below 40%, 40% to below 60%, 60% to below 80%, and 80% or above.

For each group, I calculated the number of institutions, Mean Absolute Error, mean signed error, average actual proficiency, and average predicted proficiency.

The objective was to determine whether the selected model demonstrated different error patterns across the observed outcome range.

Positive mean signed errors indicate average underestimation, while negative values indicate average overestimation.

Because the groups were formed using the actual outcome, the findings are descriptive error diagnostics rather than independent estimates of model performance for groups that could be identified using predictors alone.

The results will be interpreted cautiously and will not be treated as evidence that the model is causally biased against any particular type of institution.

```python

# ---------------------------------------------------------
# 8.4.7 Analyze Errors Across Observed Proficiency Levels
# ---------------------------------------------------------


# Define descriptive mathematics proficiency groups
proficiency_labels = [
    "Below 40%",
    "40% to below 60%",
    "60% to below 80%",
    "80% and above"
]

# Assign each institution to a group using actual proficiency
prediction_error_data["Proficiency_Group"] = pd.cut(
    prediction_error_data["Actual_Math_Proficiency"],
    bins=[-np.inf, 40, 60, 80, np.inf],
    labels=proficiency_labels,
    right=False
)

# Summarize model performance within each group
proficiency_error_summary = (
    prediction_error_data
    .groupby("Proficiency_Group", observed=True)
    .agg(
        Institutions=("Institution_ID", "size"),
        MAE=("Absolute_Error", "mean"),
        Mean_Signed_Error=("Prediction_Error", "mean"),
        Mean_Actual=("Actual_Math_Proficiency", "mean"),
        Mean_Predicted=("Predicted_Math_Proficiency", "mean")
    )
    .reset_index()
)

# Confirm every test institution belongs to a group
assert proficiency_error_summary["Institutions"].sum() == len(
    prediction_error_data
), "Some test observations are missing from the group analysis."

# Display the summary
print("Prediction Errors by Actual Mathematics Proficiency")
print("-" * 100)

print(
    proficiency_error_summary
    .round(4)
    .to_string(index=False)
)
```

**Institutions with observed proficiency below 40**

The model overestimated mathematics proficiency by approximately 17.40 percentage points on average. Their average absolute errors were also relatively large.



**Institutions with observed proficiency of 80% and above**

The model underestimated mathematics proficiency by approximately 17.72 percentage points on average.



**Institutions with observed proficiency between 40% and 80%**

The model achieved smaller average absolute errors than it did for institutions at the low and high ends of the observed outcome range.



**Key finding:** The selected model tends to predict less extreme proficiency values than those actually observed at either end of the outcome range.

This is consistent with a pattern often described as regression toward the mean or prediction shrinkage toward the middle of the distribution.

However, because we're grouping observations by their actual outcomes, some apparent directional patterns can arise mathematically from imperfect predictions. These results should be interpreted alongside other diagnostics rather than treated as independent proof of systematic bias.

### 8.5 Model Limitations and Reliability

The previous interpretation analyses examined the coefficients, feature contributions, and prediction errors of the selected Linear Regression model.

Although the model achieved better predictive performance than the baseline, its held-out test MAE of approximately 11.01 percentage points indicates that meaningful prediction errors remain.

The prediction error analysis also identified substantial errors for some individual institutions and larger average errors among institutions at the extremes of observed mathematics proficiency.

These findings highlight the importance of assessing model reliability before interpreting its predictions for educational decision-making.

This subsection evaluates the selected model's statistical assumptions, predictor limitations, generalizability, and practical reliability.

The objective is to identify the conditions under which the model's predictions may be useful and clearly document situations where they should be interpreted with caution.

The findings will support responsible recommendations and help distinguish exploratory predictive analysis from validated institutional decision-making systems.

#### 8.5.1 Linear Regression Assumptions and Model Specification

The selected Linear Regression model estimates mathematics proficiency using an additive linear combination of four institutional predictors.

Although this approach supports interpretability, several assumptions and modelling considerations affect how its results should be understood.

**Linearity**

The model assumes that the expected mathematics proficiency outcome can be represented by linear contributions from the selected predictors.

Nonlinear relationships and interactions may not be adequately captured.

**Residual Independence**

The modelling approach does not explicitly account for potential dependence between institutions within the same state, district, or other shared educational environment.

Such dependence may affect the interpretation of model coefficients and the reliability of conventional statistical inference.

**Error Variance**

The model does not explicitly account for potential differences in prediction error variability across institutional characteristics or mathematics proficiency levels.

The earlier prediction error analysis identified larger average errors at the extremes of observed proficiency.

**Multicollinearity**

The predictor analysis identified a strong negative correlation between free or reduced-price lunch eligibility and broadband access.

This relationship may complicate the interpretation of their individual adjusted coefficients.

**Residual Normality**

Normality is relevant to certain regression inference procedures but is not required for computing Linear Regression predictions.

**Bounded Target**

Mathematics proficiency is measured as a percentage, while ordinary Linear Regression can theoretically generate predictions below 0% or above 100%.

**Conclusion**

The selected model provides a transparent predictive approach, but its assumptions and specification limitations must be considered when interpreting predictions and fitted relationships.

#### Check whether test predictions fall outside 0–100%

```python

# ---------------------------------------------------------
# 8.5.2 Check the Validity of Predicted Proficiency Values
# ---------------------------------------------------------


# Store the selected model's held-out predictions
predicted_proficiency = np.asarray(
    interpretation_test_predictions,
    dtype=float
)

# Identify predictions outside the valid percentage range
below_zero = predicted_proficiency < 0
above_hundred = predicted_proficiency > 100

# Count invalid percentage predictions
below_zero_count = int(below_zero.sum())
above_hundred_count = int(above_hundred.sum())

outside_range_count = (
    below_zero_count + above_hundred_count
)

# Calculate the percentage outside the valid range
outside_range_pct = (
    outside_range_count / len(predicted_proficiency)
) * 100

# Summarize prediction ranges
prediction_range_summary = pd.DataFrame({
    "Metric": [
        "Minimum Actual Proficiency",
        "Maximum Actual Proficiency",
        "Minimum Predicted Proficiency",
        "Maximum Predicted Proficiency",
        "Predictions Below 0%",
        "Predictions Above 100%",
        "Total Predictions Outside Valid Range",
        "Percentage Outside Valid Range"
    ],
    "Value": [
        float(y_interpretation_test.min()),
        float(y_interpretation_test.max()),
        float(predicted_proficiency.min()),
        float(predicted_proficiency.max()),
        below_zero_count,
        above_hundred_count,
        outside_range_count,
        outside_range_pct
    ]
})

# Display the results
print("Predicted Proficiency Range Assessment")
print("-" * 65)

print(
    prediction_range_summary
    .round(4)
    .to_string(index=False)
)
```

#### 8.5.3 Residual Patterns and Predictive Reliability

I examined the relationship between predicted mathematics proficiency and prediction residuals to identify patterns that may indicate limitations in the selected model.

Residuals were calculated as the difference between actual and predicted proficiency.

The analysis used the 929 held-out test observations.

A residual plot was created to examine whether errors were distributed around zero or displayed systematic patterns across predicted proficiency values.

The investigation provides descriptive evidence about model fit and error behaviour.

Any observed patterns will be considered alongside the earlier prediction error analysis and will not be treated as definitive proof that a statistical assumption has been satisfied or violated.

The model was not modified using the test residuals.

```python

# ---------------------------------------------------------
# 8.5.3 Visualize Test Residuals Against Predicted Values
# ---------------------------------------------------------

# Extract predicted proficiency and corresponding residuals
fitted_predictions = prediction_error_data[
    "Predicted_Math_Proficiency"
].to_numpy()

test_residuals = prediction_error_data[
    "Prediction_Error"
].to_numpy()

# Create the residual plot
fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(
    fitted_predictions,
    test_residuals,
    alpha=0.4,
    s=28,
    color="steelblue"
)

# Add the zero-residual reference line
ax.axhline(
    y=0,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Zero Residual"
)

# Add titles and axis labels
ax.set_title(
    "Linear Regression: Test Residuals vs Predicted Proficiency",
    fontsize=14
)

ax.set_xlabel("Predicted Mathematics Proficiency (%)")

ax.set_ylabel(
    "Residual: Actual - Predicted (Percentage Points)"
)

ax.legend()

plt.tight_layout()
plt.show()

# Display supporting residual statistics
print("Held-Out Residual Summary")
print("-" * 55)

print(f"Mean Residual: {test_residuals.mean():.4f} pp")

print(
    f"Residual Standard Deviation: "
    f"{test_residuals.std(ddof=1):.4f} pp"
)

print(
    f"Mean Absolute Residual: "
    f"{np.abs(test_residuals).mean():.4f} pp"
)
```

The mean residual is close to zero, indicating little overall signed bias in the held-out predictions.

However, the relatively large residual standard deviation shows that individual prediction errors vary substantially.

The residual plot should be inspected for visible curves, changing spread, or clusters.

Our earlier analysis also demonstrated that institutions with very low and very high observed mathematics proficiency tended to have larger errors.

Finding: Linear Regression provides useful average predictions relative to the baseline, but its remaining errors limit the reliability of predictions for individual institutions.

### 8.6 Model Interpretation Summary and Conclusion

The model interpretation stage examined the behaviour, feature contributions, prediction errors, and reliability of the selected Linear Regression model.

The model was trained to predict mathematics proficiency using four institutional characteristics: per-pupil expenditure, student-teacher ratio, free or reduced-price lunch eligibility, and broadband access.

The interpretation combined fitted coefficient analysis, standardized coefficient comparisons, permutation feature importance, and prediction error diagnostics.

The findings were evaluated alongside the selected model's previously established held-out test performance.

The objective of this subsection is to consolidate the interpretation results, identify the most relevant predictive relationships, and explain their implications for the original research question.

The conclusions distinguish predictive associations from causal effects and consider the limitations of using the model for educational decision-making.

#### Consolidate our feature interpretation results

```python

# ---------------------------------------------------------
# 8.6.1 Consolidate Model Interpretation Results
# ---------------------------------------------------------


# Extract original fitted coefficients
original_coefficient_summary = coefficient_results[
    ["Feature", "Coefficient"]
].copy()

# Extract standardized coefficients
standardized_summary = standardized_coefficient_results[
    ["Feature", "Standardized_Coefficient"]
].copy()

# Extract permutation feature importance
permutation_summary = permutation_importance_results[
    ["Feature", "Mean_MAE_Increase"]
].copy()

# Combine the interpretation results
final_feature_summary = (
    original_coefficient_summary
    .merge(
        standardized_summary,
        on="Feature",
        validate="one_to_one"
    )
    .merge(
        permutation_summary,
        on="Feature",
        validate="one_to_one"
    )
)

# Add readable predictor names
final_feature_summary["Predictor"] = (
    final_feature_summary["Feature"].map(feature_labels)
)

# Add the direction of the fitted coefficient
final_feature_summary["Direction"] = np.select(
    [
        final_feature_summary["Coefficient"] > 0,
        final_feature_summary["Coefficient"] < 0
    ],
    [
        "Positive",
        "Negative"
    ],
    default="Zero"
)

# Sort by absolute standardized coefficient
final_feature_summary["Absolute_Standardized_Coefficient"] = (
    final_feature_summary["Standardized_Coefficient"].abs()
)

final_feature_summary = (
    final_feature_summary
    .sort_values(
        by="Absolute_Standardized_Coefficient",
        ascending=False
    )
    .reset_index(drop=True)
)

# Select columns for presentation
final_feature_summary = final_feature_summary[
    [
        "Predictor",
        "Feature",
        "Coefficient",
        "Direction",
        "Standardized_Coefficient",
        "Mean_MAE_Increase",
        "Absolute_Standardized_Coefficient"
    ]
]

# Verify all four predictors are present
assert len(final_feature_summary) == 4, (
    "The feature interpretation table must contain four predictors."
)

# Display the main findings
print("Final Model Feature Interpretation Summary")
print("-" * 100)

print(
    final_feature_summary[
        [
            "Predictor",
            "Coefficient",
            "Direction",
            "Standardized_Coefficient",
            "Mean_MAE_Increase"
        ]
    ].round(6).to_string(index=False)
)
```

Both standardized coefficients and permutation importance identified free or reduced-price lunch eligibility as the strongest contributor within our selected model, followed by per-pupil expenditure.

Student-teacher ratio and broadband access showed relatively small individual contributions.

However, we also observed a strong negative correlation between lunch eligibility and broadband access. This means their information overlaps, which makes interpreting individual feature importance more complicated.

These findings describe model reliance, not which education policy would produce the greatest improvement.

#### 8.6.1 Summary of Coefficient and Feature Contribution Findings

The model interpretation identified different contributions from the four institutional predictors.

**Free or Reduced-Price Lunch Eligibility**

Lunch eligibility demonstrated a negative adjusted relationship with predicted mathematics proficiency.

Its original coefficient was approximately -0.3990, indicating that a one-percentage-point difference in eligibility corresponded to approximately 0.40 percentage points lower predicted mathematics proficiency, holding the other predictors constant.

#### Consolidate prediction reliability findings

```python

# ---------------------------------------------------------
# 8.6.2 Consolidate Final Prediction Reliability Findings
# ---------------------------------------------------------


# Extract the selected model's test performance
selected_model_test_results = test_results.loc[
    test_results["Model"] == selected_model_name
].iloc[0]

# Calculate error summary statistics
mean_signed_error = prediction_error_data[
    "Prediction_Error"
].mean()

median_absolute_error = prediction_error_data[
    "Absolute_Error"
].median()

maximum_absolute_error = prediction_error_data[
    "Absolute_Error"
].max()

# Count institutions with absolute error above 20 percentage points
large_error_count = int(
    (prediction_error_data["Absolute_Error"] > 20).sum()
)

large_error_percentage = (
    large_error_count / len(prediction_error_data)
) * 100

# Organize the reliability results
reliability_summary = pd.DataFrame({
    "Metric": [
        "Held-Out Test Institutions",
        "Test MAE",
        "Test RMSE",
        "Test R-squared",
        "Mean Signed Error",
        "Median Absolute Error",
        "Maximum Absolute Error",
        "Institutions with Error Above 20 pp",
        "Percentage with Error Above 20 pp"
    ],
    "Value": [
        len(prediction_error_data),
        float(selected_model_test_results["MAE"]),
        float(selected_model_test_results["RMSE"]),
        float(selected_model_test_results["R2"]),
        float(mean_signed_error),
        float(median_absolute_error),
        float(maximum_absolute_error),
        large_error_count,
        float(large_error_percentage)
    ]
})

# Confirm the test dataset remains unchanged
assert len(prediction_error_data) == 929, (
    "The held-out test dataset should contain 929 institutions."
)

# Confirm the mean absolute error matches evaluation
assert np.isclose(
    float(selected_model_test_results["MAE"]),
    prediction_error_data["Absolute_Error"].mean()
), "The test MAE does not match the prediction error analysis."

# Display the findings
print("Selected Model: Final Prediction Reliability Summary")
print("-" * 75)

print(reliability_summary.round(4).to_string(index=False))
```

Our selected model achieved better predictive performance than the baseline.

However, its test MAE indicates that institution-level predictions may differ considerably from actual mathematics proficiency.

The model also exhibited larger average errors at the low and high ends of the observed proficiency distribution.

We must distinguish an improvement over the baseline from sufficient precision for institutional decision-making. Our analysis established the former, but not the latter.

The negative adjusted association between lunch eligibility and mathematics proficiency highlights the importance of considering socioeconomic conditions when examining academic outcomes.

**1. Identify Institutions with Greater Support Needs**

This assessment should consider actual academic outcomes, available resources, and local institutional conditions.

Institutions facing greater educational challenges may benefit from further assessment of tutoring, learning support, and other academic interventions.

The effectiveness of specific interventions should be established through appropriate evaluation rather than inferred from the regression coefficients.

**4. Promote Equitable Resource Planning**

Resource planning should consider differences in institutional needs and socioeconomic conditions.

Funding and support decisions should not be based solely on predicted mathematics proficiency.

**Policy Implication**

The findings support investigating socioeconomic inequalities and assessing whether existing educational resources adequately address differences in student and institutional needs.

They do not establish which intervention would produce the largest academic improvement.

#### 9.3.3 Socioeconomic Conditions and Digital Access

The predictor correlation analysis identified a strong negative relationship between free or reduced-price lunch eligibility and broadband access, with a Pearson correlation of approximately -0.7583.

This indicates that institutions with higher lunch eligibility tended to report lower broadband access within the analysed training dataset.

The finding suggests that socioeconomic disadvantage and digital resource access may overlap.

However, the selected predictive model demonstrated only a small adjusted broadband coefficient after accounting for the other institutional predictors.

**Interpretation**

Broadband access and socioeconomic conditions should not be treated as independent institutional challenges without further investigation.

The strong correlation indicates overlapping information that complicates the interpretation of their individual predictive contributions.

**Policy Consideration**

Educational stakeholders should examine digital access alongside broader socioeconomic conditions when investigating educational opportunity.

**Limitation**

The observed correlation does not establish that socioeconomic disadvantage causes lower broadband access or that expanding broadband access alone would improve mathematics proficiency.

Further research is required to assess these relationships and the effectiveness of potential interventions.

**Recommendation**

**Responsible Use**

Socioeconomic information should support the identification of potential barriers and resource needs rather than justify negative judgments about disadvantaged institutions.

**Limitation**

The current observational analysis cannot determine the causal impact or cost-effectiveness of specific educational support programmes.

#### 9.3.5 Socioeconomic Inequality and Educational Opportunity: Conclusion

## 10. Final Recommendations and Conclusion

This section consolidates the findings of the Education Insights Lab project and presents the final recommendations, research limitations, opportunities for future work, and overall conclusions.

The conclusions are based on the statistical analysis, predictive modelling, and policy insights developed throughout the project.

### 10.1 Project Objectives and Key Findings Summary

#### 10.1.1 Project Objectives

The project investigated how institutional funding, student-teacher ratios, socioeconomic conditions, and broadband access are associated with mathematics proficiency across US educational institutions.

The analysis focused on K–12 institutions and addressed two objectives:

1. Examine the relationships between the four institutional characteristics and mathematics proficiency.
2. Evaluate how accurately mathematics proficiency can be predicted using these characteristics.

The project used a structured data science workflow involving data cleaning, exploratory analysis, statistical testing, predictive modelling, model evaluation, and interpretation.

The final analytical dataset contained **4,642 K–12 institutions** and **35 variables**.

#### 10.1.2 Summary of Key Findings

**1. Institutional Funding**

Per-pupil expenditure demonstrated a positive and statistically significant adjusted association with mathematics proficiency.

In the selected predictive model, a $1,000 difference in expenditure corresponded to approximately 0.69 percentage points higher predicted proficiency, holding other predictors constant.

**2. Socioeconomic Conditions**

Free or reduced-price lunch eligibility demonstrated a negative and statistically significant adjusted association with mathematics proficiency.

It was also the strongest contributor in the selected predictive model, followed by per-pupil expenditure.

**3. Student-Teacher Ratio and Broadband Access**

Both predictors demonstrated small positive fitted coefficients, but neither had a statistically significant adjusted association in the earlier explanatory regression.

Broadband access was strongly negatively correlated with lunch eligibility, indicating overlapping information between the predictors.

**4. Predictive Modelling**

Three models were evaluated: a mean-based baseline, Linear Regression, and Random Forest.

Linear Regression was selected based on the lowest mean five-fold cross-validation MAE.

Its held-out test performance was:

| Metric | Result |
|---|---:|
| MAE | 11.01 percentage points |
| RMSE | 13.64 percentage points |
| R² | 0.3139 |
| MAE reduction versus baseline | 17.45% |

**5. Model Reliability**

Although Linear Regression outperformed the baseline, substantial errors remained for some institutions.

Approximately 14.85% of held-out institutions had absolute prediction errors greater than 20 percentage points.

**Overall Finding**

Socioeconomic conditions and per-pupil expenditure demonstrated the strongest adjusted and predictive relationships with mathematics proficiency among the four selected characteristics.

The predictive model provided useful but limited accuracy.

These findings support further investigation of educational inequality and institutional resources but do not establish causal effects or justify high-stakes decisions based solely on model predictions.

### 10.3 Research Limitations

Despite successfully addressing the research objectives, the study has several limitations.

#### 10.3.1 Key Limitations

**1. Dataset Coverage**

The analysis focused on 4,642 K–12 institutions from the supplied US education dataset. Colleges and universities were excluded to maintain a consistent analytical population.

The findings may not represent all US educational institutions or educational systems outside the dataset.

**2. Limited Predictor Selection**

The predictive model included four institutional characteristics: per-pupil expenditure, student-teacher ratio, free or reduced-price lunch eligibility, and broadband access.

Other potentially relevant factors, including prior academic achievement, instructional quality, school leadership, and community characteristics, were not included.

**3. Observational Data and Causality**

The study identified statistical associations rather than causal relationships.

Consequently, the results cannot establish that changing institutional funding, staffing, or broadband access would directly improve mathematics proficiency.

**4. Correlated Predictors**

Lunch eligibility and broadband access demonstrated a strong negative correlation of approximately -0.76.

This overlapping information complicates the interpretation of their individual regression coefficients and predictive contributions.

**5. Model Specification and Predictive Accuracy**

The selected Linear Regression model assumes additive linear predictor relationships and may not capture complex educational patterns.

Its held-out test MAE was approximately 11.01 percentage points, with some institutions experiencing substantially larger prediction errors.

This limits its reliability for precise institution-level predictions.

**6. Geographic and Temporal Generalizability**

The evaluation used a random institutional train-test split and five-fold cross-validation.

The model was not independently validated on entirely unseen states or future academic years.

Its performance in those settings remains unverified.

**7. Institutional-Level Analysis**

The dataset contains institutional indicators rather than detailed individual student records.

Therefore, the findings cannot be used to explain individual students' academic performance or experiences.

**Overall Limitation**

The study provides useful evidence about institutional relationships and predictive patterns within the analysed dataset, but its findings should not be generalized beyond that scope without additional validation.

#### 10.3.2 Implications of the Research Limitations

The identified limitations affect how the findings should be interpreted and applied.

**Interpretation:** The statistical relationships and feature contributions should not be presented as causal effects.

**Prediction:** The selected model provides measurable improvements over the baseline but cannot guarantee accurate mathematics proficiency predictions for individual institutions.

**Generalization:** Findings should be restricted to the analysed K–12 dataset unless supported by additional geographic, temporal, or external validation.

**Policy Application:** Educational recommendations should be treated as priorities for further investigation rather than interventions whose effectiveness has been demonstrated.

**Conclusion**

These limitations do not invalidate the analysis but define the boundaries of its conclusions.

The findings provide a foundation for further educational research, while highlighting the need for additional predictors, broader validation, and appropriate causal evaluation before operational policy use.

The current study provides a reproducible foundation for these future investigations but does not establish their outcomes.

### 10.5 Final Research Question Answer

**Primary Research Question**

How are institutional funding, student-teacher ratios, socioeconomic conditions, and broadband access associated with mathematics proficiency across US educational institutions, and how accurately can mathematics proficiency be predicted using institutional characteristics?

The research question was investigated using 4,642 US K–12 institutions through exploratory data analysis, statistical regression, and machine learning.

#### 10.5.1 Relationships Between Institutional Characteristics and Mathematics Proficiency

The statistical analysis identified the following adjusted relationships:

**Institutional Funding:** Per-pupil expenditure demonstrated a positive and statistically significant association with mathematics proficiency. The selected predictive model indicated approximately 0.69 percentage points higher predicted proficiency for a $1,000 difference in expenditure, holding other predictors constant.

**Socioeconomic Conditions:** Free or reduced-price lunch eligibility demonstrated a negative and statistically significant adjusted association with mathematics proficiency. It also emerged as the strongest contributor in the selected predictive model.

**Student-Teacher Ratio:** Student-teacher ratio demonstrated a small positive adjusted coefficient, but the explanatory regression did not establish a statistically significant association.

**Broadband Access:** Broadband access demonstrated a small positive adjusted coefficient, but its association was not statistically significant in the explanatory regression. Its strong negative correlation with lunch eligibility also indicated overlapping predictor information.

**Conclusion**

Among the four selected characteristics, socioeconomic disadvantage and per-pupil expenditure demonstrated the strongest evidence of adjusted relationships with mathematics proficiency.

These findings represent statistical associations within the analysed K–12 dataset and do not establish causal effects.

#### 10.5.2 Predictive Accuracy and Overall Research Answer

Three regression models were developed and evaluated: a mean-based Dummy Regressor, Linear Regression, and Random Forest Regressor.

Linear Regression was selected using the lowest mean five-fold cross-validation MAE.

The selected model achieved the following held-out test results:

| Evaluation Metric | Result |
|---|---:|
| Mean Absolute Error | 11.01 percentage points |
| Root Mean Squared Error | 13.64 percentage points |
| R-squared | 0.3139 |
| MAE reduction compared with baseline | 17.45% |

The findings demonstrate that the four selected institutional characteristics provide useful predictive information about mathematics proficiency.

However, the remaining prediction errors indicate that these predictors do not fully account for institutional differences in academic outcomes.

**Final Answer to the Research Question**

Institutional expenditure and socioeconomic conditions demonstrated meaningful adjusted associations with mathematics proficiency, while student-teacher ratio and broadband access showed weaker adjusted relationships in the analysed models.

Mathematics proficiency could be predicted more accurately using these institutional characteristics than by relying on the average proficiency alone.

Linear Regression provided the strongest validation performance among the evaluated models and achieved an average absolute test prediction error of approximately 11 percentage points.

**Overall Conclusion**

The study identified relevant institutional relationships and demonstrated measurable but limited predictive accuracy.

The findings support further investigation of educational inequality and institutional resources while emphasizing that observational associations and model predictions should not be interpreted as causal policy effects.

### 10.6 Overall Project Conclusion

The Education Insights Lab project investigated the relationships between institutional characteristics and mathematics proficiency across 4,642 US K–12 institutions.

The analysis followed a structured data science workflow involving data understanding, cleaning, exploratory analysis, statistical testing, predictive modelling, model evaluation, and interpretation.

The statistical findings identified a positive adjusted association between per-pupil expenditure and mathematics proficiency and a negative adjusted association between free or reduced-price lunch eligibility and mathematics proficiency.

These two characteristics also demonstrated the strongest contributions within the selected predictive model.

Student-teacher ratio and broadband access showed weaker adjusted relationships after accounting for the other predictors.

For the predictive objective, three regression models were evaluated: a mean-based baseline, Linear Regression, and Random Forest.

Linear Regression was selected based on its superior cross-validation performance and interpretability.

The selected model achieved a held-out test MAE of **11.01 percentage points**, RMSE of **13.64**, and R-squared of **0.3139**, reducing average absolute prediction error by **17.45%** compared with the baseline.

Although these results demonstrated measurable predictive value, substantial errors remained for some institutions.

The findings therefore support the use of institutional characteristics for exploratory educational analysis while highlighting the limitations of using the selected model for precise institution-level predictions.

**Overall Conclusion**

Socioeconomic conditions and per-pupil expenditure were the most prominent institutional characteristics associated with mathematics proficiency in this analysis.

The findings provide evidence-informed insights for investigating educational inequality, institutional funding, and resource allocation.

However, the study's observational design and remaining prediction errors mean that the results should not be interpreted as causal effects or used independently for high-stakes educational decisions.

The analysis identified meaningful institutional associations and demonstrated that the selected predictors improved prediction performance relative to a mean-based baseline.

The findings also highlighted the importance of model validation, careful interpretation, and recognition of research limitations.

Educational stakeholders should investigate socioeconomic disparities and funding adequacy, assess institutional support needs, and evaluate proposed educational interventions using appropriate evidence.

Predictive modelling should complement actual educational outcomes and professional judgment rather than replace them.
